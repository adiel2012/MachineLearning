## 🔬 **Why Ensemble Methods Work**

### **1. Bias-Variance Decomposition**
The expected prediction error can be decomposed as:
$$E[(y - \hat{f}(x))^2] = \text{Bias}^2 + \text{Variance} + \text{Irreducible Error}$$

### **2. Ensemble Benefits**
- **Bagging**: Reduces variance by averaging predictions
- **Boosting**: Reduces bias by focusing on hard examples
- **Stacking**: Combines different model strengths

### **3. Diversity-Accuracy Tradeoff**
Successful ensembles balance:
- **Individual Accuracy**: Each model should be reasonably good
- **Diversity**: Models should make different types of errors

# 🌲 **Part II: Bagging Methods - Comprehensive Implementation**

## 📚 **Theoretical Foundation**

**Bootstrap Aggregating (Bagging)** is a powerful ensemble technique that combines the predictions of multiple models trained on different bootstrap samples of the training data.

### **Mathematical Foundation**

For a training dataset $D = \{(x_i, y_i)\}_{i=1}^n$, bagging generates $B$ bootstrap samples:
- Each bootstrap sample $D_b$ contains $n$ instances sampled with replacement from $D$
- Approximately 63.2% of original instances appear in each bootstrap sample
- Train base learner $f_b$ on each $D_b$

**Aggregation Rules:**
- **Regression**: $\hat{y} = \frac{1}{B}\sum_{b=1}^B f_b(x)$
- **Classification**: $\hat{y} = \text{mode}\{f_1(x), f_2(x), ..., f_B(x)\}$

### **Variance Reduction Property**

If base learners are independent with variance $\sigma^2$, the ensemble variance is:
$$\text{Var}(\text{ensemble}) = \frac{\sigma^2}{B}$$

For correlated learners with correlation $\rho$:
$$\text{Var}(\text{ensemble}) = \rho\sigma^2 + \frac{1-\rho}{B}\sigma^2$$

### **Out-of-Bag (OOB) Evaluation**

Since each bootstrap sample excludes ~36.8% of the original data:
- **OOB samples** serve as a validation set for each base learner
- **OOB score** provides an unbiased estimate of ensemble performance
- No need for separate validation set!

```python
# 🌲 BAGGING ENSEMBLE IMPLEMENTATION

class BaggingEnsemble:
    """Advanced Bagging implementation with comprehensive analysis"""
    
    def __init__(self, base_estimator=None, n_estimators: int = 100, 
                 max_samples: float = 1.0, max_features: float = 1.0,
                 bootstrap: bool = True, random_state: int = 42):
        
        self.base_estimator = base_estimator or DecisionTreeClassifier(max_depth=5)
        self.n_estimators = n_estimators
        self.max_samples = max_samples
        self.max_features = max_features
        self.bootstrap = bootstrap
        self.random_state = random_state
        
        # Storage for ensemble
        self.estimators_ = []
        self.estimators_features_ = []
        self.estimators_samples_ = []
        self.feature_importances_ = None
        
        # Training metrics
        self.oob_score_ = None
        self.training_history = {
            'oob_scores': [],
            'individual_scores': [],
            'diversity_scores': []
        }
    
    def fit(self, X: np.ndarray, y: np.ndarray):
        """Fit the bagging ensemble with comprehensive tracking"""
        np.random.seed(self.random_state)
        
        n_samples, n_features = X.shape
        max_samples = int(self.max_samples * n_samples)
        max_features = int(self.max_features * n_features)
        
        # Track out-of-bag predictions for OOB score
        oob_predictions = np.zeros((n_samples, len(np.unique(y))))
        oob_counts = np.zeros(n_samples)
        
        print(f"🌲 Training Bagging Ensemble with {self.n_estimators} estimators")
        print(f"📊 Bootstrap samples: {max_samples}/{n_samples}")
        print(f"🔍 Features per estimator: {max_features}/{n_features}")
        
        for i in range(self.n_estimators):
            # Bootstrap sampling
            if self.bootstrap:
                indices = np.random.choice(n_samples, max_samples, replace=True)
            else:
                indices = np.random.choice(n_samples, max_samples, replace=False)
            
            # Feature sampling
            feature_indices = np.random.choice(n_features, max_features, replace=False)
            
            # Train estimator
            estimator = self.base_estimator.__class__(**self.base_estimator.get_params())
            X_sample = X[np.ix_(indices, feature_indices)]
            y_sample = y[indices]
            
            estimator.fit(X_sample, y_sample)
            
            # Store estimator and indices
            self.estimators_.append(estimator)
            self.estimators_samples_.append(indices)
            self.estimators_features_.append(feature_indices)
            
            # Calculate OOB predictions
            oob_indices = np.setdiff1d(np.arange(n_samples), indices)
            if len(oob_indices) > 0:
                X_oob = X[np.ix_(oob_indices, feature_indices)]
                if hasattr(estimator, 'predict_proba'):
                    oob_pred_proba = estimator.predict_proba(X_oob)
                    oob_predictions[oob_indices] += oob_pred_proba
                else:
                    oob_pred = estimator.predict(X_oob)
                    for idx, pred in zip(oob_indices, oob_pred):
                        oob_predictions[idx, pred] += 1
                
                oob_counts[oob_indices] += 1
            
            # Progress tracking
            if (i + 1) % (self.n_estimators // 10) == 0:
                progress = (i + 1) / self.n_estimators * 100
                print(f"   Progress: {progress:.0f}% ({i+1}/{self.n_estimators})")
        
        # Calculate OOB score
        oob_mask = oob_counts > 0
        if np.sum(oob_mask) > 0:
            oob_predictions_final = np.argmax(oob_predictions[oob_mask], axis=1)
            self.oob_score_ = accuracy_score(y[oob_mask], oob_predictions_final)
        
        # Calculate feature importances
        self._calculate_feature_importances(n_features)
        
        print(f"✅ Training completed!")
        print(f"📈 OOB Score: {self.oob_score_:.4f}" if self.oob_score_ else "📈 OOB Score: Not available")
        
        return self
    
    def predict(self, X: np.ndarray) -> np.ndarray:
        """Make predictions using majority voting"""
        predictions = np.zeros((X.shape[0], self.n_estimators))
        
        for i, (estimator, features) in enumerate(zip(self.estimators_, self.estimators_features_)):
            X_subset = X[:, features]
            predictions[:, i] = estimator.predict(X_subset)
        
        # Majority voting
        final_predictions = np.array([stats.mode(row)[0][0] for row in predictions])
        return final_predictions
    
    def predict_proba(self, X: np.ndarray) -> np.ndarray:
        """Predict class probabilities"""
        if not hasattr(self.base_estimator, 'predict_proba'):
            raise ValueError("Base estimator doesn't support probability prediction")
        
        n_classes = len(np.unique([est.classes_ for est in self.estimators_][0]))
        probabilities = np.zeros((X.shape[0], n_classes))
        
        for estimator, features in zip(self.estimators_, self.estimators_features_):
            X_subset = X[:, features]
            probabilities += estimator.predict_proba(X_subset)
        
        return probabilities / self.n_estimators
    
    def _calculate_feature_importances(self, n_features: int):
        """Calculate average feature importances"""
        if hasattr(self.base_estimator, 'feature_importances_'):
            importances = np.zeros(n_features)
            
            for estimator, features in zip(self.estimators_, self.estimators_features_):
                if hasattr(estimator, 'feature_importances_'):
                    importances[features] += estimator.feature_importances_
            
            self.feature_importances_ = importances / self.n_estimators
    
    def calculate_diversity_metrics(self, X: np.ndarray, y: np.ndarray) -> Dict[str, float]:
        """Calculate ensemble diversity metrics"""
        predictions = []
        
        for estimator, features in zip(self.estimators_, self.estimators_features_):
            X_subset = X[:, features]
            pred = estimator.predict(X_subset)
            predictions.append(pred)
        
        predictions = np.array(predictions).T  # Shape: (n_samples, n_estimators)
        
        # Q-statistic (average pairwise diversity)
        q_statistics = []
        for i in range(self.n_estimators):
            for j in range(i + 1, self.n_estimators):
                pred_i, pred_j = predictions[:, i], predictions[:, j]
                
                # Contingency table
                n11 = np.sum((pred_i == y) & (pred_j == y))
                n10 = np.sum((pred_i == y) & (pred_j != y))
                n01 = np.sum((pred_i != y) & (pred_j == y))
                n00 = np.sum((pred_i != y) & (pred_j != y))
                
                # Q-statistic
                if (n11 * n00 + n10 * n01) != 0:
                    q = (n11 * n00 - n10 * n01) / (n11 * n00 + n10 * n01)
                    q_statistics.append(q)
        
        # Disagreement measure
        disagreements = []
        for i in range(self.n_estimators):
            for j in range(i + 1, self.n_estimators):
                disagreement = np.mean(predictions[:, i] != predictions[:, j])
                disagreements.append(disagreement)
        
        return {
            'q_statistic_mean': np.mean(q_statistics) if q_statistics else 0,
            'q_statistic_std': np.std(q_statistics) if q_statistics else 0,
            'disagreement_mean': np.mean(disagreements),
            'disagreement_std': np.std(disagreements)
        }

print("🌲 Advanced Bagging Ensemble implementation ready!")
print("   📊 Comprehensive OOB scoring and diversity analysis")
print("   🔍 Feature and sample tracking for transparency")
```

```python
# 🏗️ STACKING AND VOTING ENSEMBLES

class StackingEnsemble:
    """
    Advanced Stacking (Stacked Generalization) Implementation
    
    Stacking uses a meta-learner to combine predictions from multiple base learners,
    creating a two-level learning architecture that often outperforms simple voting.
    """
    
    def __init__(self, base_estimators: List, meta_estimator=None, cv: int = 5, 
                 use_probabilities: bool = True, random_state: int = 42):
        """
        Initialize Stacking Ensemble
        
        Args:
            base_estimators: List of (name, estimator) tuples for level-0 models
            meta_estimator: Level-1 meta-learner (default: LogisticRegression)
            cv: Number of CV folds for generating meta-features
            use_probabilities: Use predict_proba instead of predict for base models
            random_state: Random seed for reproducibility
        """
        self.base_estimators = base_estimators
        self.meta_estimator = meta_estimator or LogisticRegression(random_state=random_state)
        self.cv = cv
        self.use_probabilities = use_probabilities
        self.random_state = random_state
        self.fitted_base_estimators_ = []
        self.meta_features_ = None
        
    def _get_base_predictions(self, estimator, X: np.ndarray) -> np.ndarray:
        """Get predictions from base estimator"""
        if self.use_probabilities and hasattr(estimator, 'predict_proba'):
            preds = estimator.predict_proba(X)
            # For binary classification, use only positive class probability
            if preds.shape[1] == 2:
                return preds[:, 1]
            return preds
        else:
            return estimator.predict(X)
    
    def fit(self, X: np.ndarray, y: np.ndarray):
        """
        Fit the stacking ensemble using cross-validation
        
        Process:
        1. Split data into K folds
        2. For each fold, train base models on other folds
        3. Generate out-of-fold predictions as meta-features
        4. Train meta-learner on meta-features
        5. Retrain all base models on full dataset
        """
        from sklearn.model_selection import KFold, StratifiedKFold
        from sklearn.base import clone
        
        n_samples = X.shape[0]
        n_base_estimators = len(self.base_estimators)
        
        # Determine if classification or regression
        is_classification = len(np.unique(y)) < 0.1 * len(y)
        
        # Initialize meta-features matrix
        if self.use_probabilities and is_classification:
            n_classes = len(np.unique(y))
            if n_classes == 2:
                self.meta_features_ = np.zeros((n_samples, n_base_estimators))
            else:
                self.meta_features_ = np.zeros((n_samples, n_base_estimators * n_classes))
        else:
            self.meta_features_ = np.zeros((n_samples, n_base_estimators))
        
        # Set up cross-validation
        if is_classification:
            kf = StratifiedKFold(n_splits=self.cv, shuffle=True, random_state=self.random_state)
        else:
            kf = KFold(n_splits=self.cv, shuffle=True, random_state=self.random_state)
        
        # Generate meta-features using cross-validation
        print(f"📊 Generating meta-features using {self.cv}-fold cross-validation...")
        
        for fold_idx, (train_idx, val_idx) in enumerate(kf.split(X, y)):
            print(f"   Processing fold {fold_idx + 1}/{self.cv}...")
            
            X_fold_train, X_fold_val = X[train_idx], X[val_idx]
            y_fold_train = y[train_idx]
            
            # Train and predict with each base estimator
            col_idx = 0
            for name, base_est in self.base_estimators:
                # Clone and fit base estimator
                fold_estimator = clone(base_est)
                fold_estimator.fit(X_fold_train, y_fold_train)
                
                # Get out-of-fold predictions
                fold_preds = self._get_base_predictions(fold_estimator, X_fold_val)
                
                # Store in meta-features matrix
                if fold_preds.ndim == 1:
                    self.meta_features_[val_idx, col_idx] = fold_preds
                    col_idx += 1
                else:
                    n_cols = fold_preds.shape[1]
                    self.meta_features_[val_idx, col_idx:col_idx + n_cols] = fold_preds
                    col_idx += n_cols
        
        # Train meta-learner on meta-features
        print("🎯 Training meta-learner...")
        self.meta_estimator.fit(self.meta_features_, y)
        
        # Retrain base estimators on full dataset
        print("🔄 Retraining base estimators on full dataset...")
        self.fitted_base_estimators_ = []
        for name, base_est in self.base_estimators:
            fitted_estimator = clone(base_est)
            fitted_estimator.fit(X, y)
            self.fitted_base_estimators_.append((name, fitted_estimator))
        
        return self
    
    def predict(self, X: np.ndarray) -> np.ndarray:
        """Generate stacked predictions"""
        # Get base predictions
        meta_features = self._generate_meta_features(X)
        
        # Meta-learner prediction
        return self.meta_estimator.predict(meta_features)
    
    def predict_proba(self, X: np.ndarray) -> np.ndarray:
        """Generate stacked prediction probabilities"""
        if not hasattr(self.meta_estimator, 'predict_proba'):
            raise ValueError("Meta-estimator doesn't support predict_proba")
        
        # Get base predictions
        meta_features = self._generate_meta_features(X)
        
        # Meta-learner prediction probabilities
        return self.meta_estimator.predict_proba(meta_features)
    
    def _generate_meta_features(self, X: np.ndarray) -> np.ndarray:
        """Generate meta-features from base estimator predictions"""
        n_samples = X.shape[0]
        
        # Initialize meta-features matrix with same shape as training
        meta_features = np.zeros((n_samples, self.meta_features_.shape[1]))
        
        col_idx = 0
        for name, fitted_estimator in self.fitted_base_estimators_:
            preds = self._get_base_predictions(fitted_estimator, X)
            
            if preds.ndim == 1:
                meta_features[:, col_idx] = preds
                col_idx += 1
            else:
                n_cols = preds.shape[1]
                meta_features[:, col_idx:col_idx + n_cols] = preds
                col_idx += n_cols
        
        return meta_features
    
    def get_base_predictions_detailed(self, X: np.ndarray) -> Dict[str, np.ndarray]:
        """Get detailed predictions from all base estimators"""
        base_predictions = {}
        
        for name, fitted_estimator in self.fitted_base_estimators_:
            if self.use_probabilities and hasattr(fitted_estimator, 'predict_proba'):
                base_predictions[f"{name}_proba"] = fitted_estimator.predict_proba(X)
            base_predictions[f"{name}_pred"] = fitted_estimator.predict(X)
        
        return base_predictions


class VotingEnsemble:
    """
    Advanced Voting Ensemble Implementation
    
    Supports both hard voting (majority vote) and soft voting (average probabilities)
    with weighted voting and confidence-based weighting.
    """
    
    def __init__(self, estimators: List, voting: str = 'soft', weights: List[float] = None,
                 confidence_weighting: bool = False, random_state: int = 42):
        """
        Initialize Voting Ensemble
        
        Args:
            estimators: List of (name, estimator) tuples
            voting: 'hard' or 'soft' voting strategy
            weights: Manual weights for each estimator
            confidence_weighting: Use prediction confidence as weights
            random_state: Random seed
        """
        self.estimators = estimators
        self.voting = voting
        self.weights = weights
        self.confidence_weighting = confidence_weighting
        self.random_state = random_state
        self.fitted_estimators_ = []
        self.learned_weights_ = None
        
    def fit(self, X: np.ndarray, y: np.ndarray):
        """Fit all base estimators"""
        print(f"🗳️ Training {len(self.estimators)} base estimators for voting...")
        
        from sklearn.base import clone
        self.fitted_estimators_ = []
        
        for i, (name, estimator) in enumerate(self.estimators):
            print(f"   Training {name} ({i+1}/{len(self.estimators)})...")
            fitted_est = clone(estimator)
            fitted_est.fit(X, y)
            self.fitted_estimators_.append((name, fitted_est))
        
        # Learn confidence-based weights if requested
        if self.confidence_weighting:
            self._learn_confidence_weights(X, y)
        
        return self
    
    def _learn_confidence_weights(self, X: np.ndarray, y: np.ndarray):
        """Learn weights based on prediction confidence during training"""
        from sklearn.model_selection import cross_val_score
        
        print("📊 Learning confidence-based weights...")
        scores = []
        
        for name, fitted_est in self.fitted_estimators_:
            # Use cross-validation score as weight proxy
            cv_scores = cross_val_score(fitted_est, X, y, cv=5, scoring='accuracy')
            scores.append(cv_scores.mean())
        
        # Convert scores to weights (higher score = higher weight)
        scores = np.array(scores)
        self.learned_weights_ = scores / scores.sum()
        
        print("   Learned weights:")
        for (name, _), weight in zip(self.fitted_estimators_, self.learned_weights_):
            print(f"     {name}: {weight:.3f}")
    
    def predict(self, X: np.ndarray) -> np.ndarray:
        """Generate ensemble predictions using voting"""
        if self.voting == 'hard':
            return self._hard_voting_predict(X)
        else:
            return self._soft_voting_predict(X)
    
    def _hard_voting_predict(self, X: np.ndarray) -> np.ndarray:
        """Hard voting: majority vote"""
        predictions = []
        
        for name, fitted_est in self.fitted_estimators_:
            pred = fitted_est.predict(X)
            predictions.append(pred)
        
        predictions = np.array(predictions).T  # Shape: (n_samples, n_estimators)
        
        # Apply weights if provided
        weights = self._get_effective_weights()
        if weights is not None:
            # Weighted voting using scipy mode with weights
            from scipy.stats import mode
            final_predictions = []
            for i in range(len(X)):
                sample_preds = predictions[i]
                # Simple weighted majority vote
                unique_preds, counts = np.unique(sample_preds, return_counts=True)
                weighted_counts = []
                for pred in unique_preds:
                    mask = sample_preds == pred
                    weighted_count = weights[mask].sum()
                    weighted_counts.append(weighted_count)
                
                final_pred = unique_preds[np.argmax(weighted_counts)]
                final_predictions.append(final_pred)
            
            return np.array(final_predictions)
        else:
            # Simple majority vote
            from scipy.stats import mode
            result = mode(predictions, axis=1)
            return result.mode.flatten()
    
    def _soft_voting_predict(self, X: np.ndarray) -> np.ndarray:
        """Soft voting: average probabilities"""
        if not all(hasattr(est, 'predict_proba') for _, est in self.fitted_estimators_):
            raise ValueError("All estimators must support predict_proba for soft voting")
        
        probabilities = []
        weights = self._get_effective_weights()
        
        for i, (name, fitted_est) in enumerate(self.fitted_estimators_):
            prob = fitted_est.predict_proba(X)
            if weights is not None:
                prob = prob * weights[i]
            probabilities.append(prob)
        
        # Average probabilities
        avg_prob = np.mean(probabilities, axis=0)
        
        # Return class with highest probability
        return np.argmax(avg_prob, axis=1)
    
    def predict_proba(self, X: np.ndarray) -> np.ndarray:
        """Generate ensemble prediction probabilities"""
        if not all(hasattr(est, 'predict_proba') for _, est in self.fitted_estimators_):
            raise ValueError("All estimators must support predict_proba")
        
        probabilities = []
        weights = self._get_effective_weights()
        
        for i, (name, fitted_est) in enumerate(self.fitted_estimators_):
            prob = fitted_est.predict_proba(X)
            if weights is not None:
                prob = prob * weights[i]
            probabilities.append(prob)
        
        # Average probabilities
        avg_prob = np.mean(probabilities, axis=0)
        
        # Normalize to ensure probabilities sum to 1
        return avg_prob / avg_prob.sum(axis=1, keepdims=True)
    
    def _get_effective_weights(self):
        """Get the effective weights to use"""
        if self.confidence_weighting and self.learned_weights_ is not None:
            return self.learned_weights_
        return self.weights
    
    def get_individual_predictions(self, X: np.ndarray) -> Dict[str, np.ndarray]:
        """Get predictions from each individual estimator"""
        predictions = {}
        
        for name, fitted_est in self.fitted_estimators_:
            predictions[f"{name}_pred"] = fitted_est.predict(X)
            if hasattr(fitted_est, 'predict_proba'):
                predictions[f"{name}_proba"] = fitted_est.predict_proba(X)
        
        return predictions


print("🏗️ Advanced Stacking and Voting Ensemble implementations ready!")
print("   📊 StackingEnsemble: Multi-level learning with meta-learner")
print("   🗳️ VotingEnsemble: Hard/soft voting with confidence weighting")
```

# 🚀 **Part III: Boosting Methods - Advanced Implementation**

## 📚 **Theoretical Foundation**

**Boosting** is a sequential ensemble technique that combines weak learners to create a strong learner. Unlike bagging, boosting trains models sequentially, with each model learning from the mistakes of previous ones.

### **Mathematical Foundation**

**AdaBoost Algorithm:**
1. Initialize weights: $w_i^{(1)} = \frac{1}{n}$ for $i = 1, ..., n$
2. For $t = 1$ to $T$:
   - Train weak learner $h_t$ on weighted dataset
   - Calculate weighted error: $\epsilon_t = \sum_{i=1}^n w_i^{(t)} \mathbf{1}[y_i \neq h_t(x_i)]$
   - Calculate learner weight: $\alpha_t = \frac{1}{2}\ln\left(\frac{1-\epsilon_t}{\epsilon_t}\right)$
   - Update sample weights: $w_i^{(t+1)} = w_i^{(t)} \exp(-\alpha_t y_i h_t(x_i))$
   - Normalize weights: $w_i^{(t+1)} = \frac{w_i^{(t+1)}}{\sum_{j=1}^n w_j^{(t+1)}}$

3. Final prediction: $H(x) = \text{sign}\left(\sum_{t=1}^T \alpha_t h_t(x)\right)$

### **Bias-Variance Decomposition for Boosting**

Boosting primarily reduces **bias** by focusing on hard-to-classify samples:
- **Bias reduction**: Sequential learning corrects systematic errors
- **Variance**: Can increase with overfitting, especially with deep trees
- **Key insight**: Boosting weak learners → strong learner

### **Gradient Boosting Framework**

Gradient Boosting minimizes a loss function $L(y, F(x))$ using functional gradient descent:

1. Initialize: $F_0(x) = \arg\min_\gamma \sum_{i=1}^n L(y_i, \gamma)$
2. For $m = 1$ to $M$:
   - Calculate negative gradients: $r_{im} = -\left[\frac{\partial L(y_i, F(x_i))}{\partial F(x_i)}\right]_{F=F_{m-1}}$
   - Fit base learner $h_m$ to pseudo-residuals $r_{im}$
   - Find optimal step size: $\gamma_m = \arg\min_\gamma \sum_{i=1}^n L(y_i, F_{m-1}(x_i) + \gamma h_m(x_i))$
   - Update model: $F_m(x) = F_{m-1}(x) + \gamma_m h_m(x)$

```python
# 🚀 ADVANCED BOOSTING IMPLEMENTATIONS

class AdaBoostEnsemble:
    """
    Advanced AdaBoost Implementation from scratch
    
    AdaBoost (Adaptive Boosting) sequentially combines weak learners,
    giving higher weight to previously misclassified samples.
    """
    
    def __init__(self, base_estimator=None, n_estimators: int = 50, 
                 learning_rate: float = 1.0, random_state: int = 42):
        """
        Initialize AdaBoost Ensemble
        
        Args:
            base_estimator: Weak learner (default: DecisionTreeClassifier with max_depth=1)
            n_estimators: Number of weak learners
            learning_rate: Learning rate (shrinkage parameter)
            random_state: Random seed
        """
        self.base_estimator = base_estimator or DecisionTreeClassifier(max_depth=1, random_state=random_state)
        self.n_estimators = n_estimators
        self.learning_rate = learning_rate
        self.random_state = random_state
        
        # To be filled during training
        self.estimators_ = []
        self.estimator_weights_ = []
        self.estimator_errors_ = []
        self.sample_weight_history_ = []
        
    def fit(self, X: np.ndarray, y: np.ndarray):
        """
        Train AdaBoost ensemble
        
        Returns detailed training information for analysis
        """
        from sklearn.base import clone
        from sklearn.tree import DecisionTreeClassifier
        
        n_samples = len(X)
        
        # Convert labels to {-1, 1} for AdaBoost math
        classes = np.unique(y)
        if len(classes) != 2:
            raise ValueError("AdaBoost only supports binary classification")
        
        y_binary = np.where(y == classes[0], -1, 1)
        
        # Initialize sample weights uniformly
        sample_weights = np.ones(n_samples) / n_samples
        self.sample_weight_history_.append(sample_weights.copy())
        
        print(f"🚀 Training AdaBoost with {self.n_estimators} estimators...")
        
        for t in range(self.n_estimators):
            print(f"   Training estimator {t+1}/{self.n_estimators}...")
            
            # Clone and train weak learner on weighted samples
            estimator = clone(self.base_estimator)
            estimator.fit(X, y, sample_weight=sample_weights)
            
            # Get predictions
            predictions = estimator.predict(X)
            pred_binary = np.where(predictions == classes[0], -1, 1)
            
            # Calculate weighted error
            incorrect = pred_binary != y_binary
            error = np.sum(sample_weights * incorrect) / np.sum(sample_weights)
            
            # Stop if perfect classifier or worse than random
            if error <= 0:
                print(f"   Perfect classifier found at iteration {t+1}!")
                if t == 0:
                    # If first estimator is perfect, use it alone
                    self.estimators_ = [estimator]
                    self.estimator_weights_ = [1.0]
                    self.estimator_errors_ = [0.0]
                break
                
            if error >= 0.5:
                print(f"   Early stopping: error {error:.4f} >= 0.5 at iteration {t+1}")
                break
            
            # Calculate estimator weight (alpha)
            alpha = self.learning_rate * 0.5 * np.log((1 - error) / error)
            
            # Store estimator and its weight
            self.estimators_.append(estimator)
            self.estimator_weights_.append(alpha)
            self.estimator_errors_.append(error)
            
            # Update sample weights
            sample_weights *= np.exp(-alpha * y_binary * pred_binary)
            sample_weights /= np.sum(sample_weights)  # Normalize
            
            self.sample_weight_history_.append(sample_weights.copy())
            
            print(f"     Error: {error:.4f}, Alpha: {alpha:.4f}")
        
        print(f"✅ AdaBoost training complete with {len(self.estimators_)} estimators")
        return self
        
    def predict(self, X: np.ndarray) -> np.ndarray:
        """Make predictions using weighted majority vote"""
        if not self.estimators_:
            raise ValueError("Estimator not fitted yet")
        
        # Get weighted predictions
        weighted_sum = np.zeros(len(X))
        
        for estimator, weight in zip(self.estimators_, self.estimator_weights_):
            pred = estimator.predict(X)
            # Convert to {-1, 1} and add weighted contribution
            pred_binary = np.where(pred == estimator.classes_[0], -1, 1)
            weighted_sum += weight * pred_binary
        
        # Convert back to original class labels
        predictions = np.where(weighted_sum >= 0, self.estimators_[0].classes_[1], 
                              self.estimators_[0].classes_[0])
        return predictions
    
    def predict_proba(self, X: np.ndarray) -> np.ndarray:
        """Estimate class probabilities"""
        if not self.estimators_:
            raise ValueError("Estimator not fitted yet")
        
        # Get weighted sum
        weighted_sum = np.zeros(len(X))
        for estimator, weight in zip(self.estimators_, self.estimator_weights_):
            pred = estimator.predict(X)
            pred_binary = np.where(pred == estimator.classes_[0], -1, 1)
            weighted_sum += weight * pred_binary
        
        # Convert to probabilities using sigmoid-like transformation
        prob_class_1 = 1 / (1 + np.exp(-2 * weighted_sum))
        prob_class_0 = 1 - prob_class_1
        
        return np.column_stack((prob_class_0, prob_class_1))
    
    def staged_predict(self, X: np.ndarray):
        """Return predictions at each boosting stage"""
        staged_predictions = []
        weighted_sum = np.zeros(len(X))
        
        for i, (estimator, weight) in enumerate(zip(self.estimators_, self.estimator_weights_)):
            pred = estimator.predict(X)
            pred_binary = np.where(pred == estimator.classes_[0], -1, 1)
            weighted_sum += weight * pred_binary
            
            # Convert to class predictions
            stage_pred = np.where(weighted_sum >= 0, estimator.classes_[1], estimator.classes_[0])
            staged_predictions.append(stage_pred)
            
        return staged_predictions
    
    def get_feature_importance(self) -> np.ndarray:
        """Compute feature importance from weak learners"""
        if not self.estimators_:
            raise ValueError("Estimator not fitted yet")
        
        importances = np.zeros(self.estimators_[0].n_features_in_)
        
        for estimator, weight in zip(self.estimators_, self.estimator_weights_):
            if hasattr(estimator, 'feature_importances_'):
                importances += weight * estimator.feature_importances_
        
        # Normalize
        importances /= np.sum(importances) if np.sum(importances) > 0 else 1
        return importances


class GradientBoostingEnsemble:
    """
    Advanced Gradient Boosting Implementation
    
    Implements gradient boosting with multiple loss functions and regularization.
    """
    
    def __init__(self, loss: str = 'deviance', learning_rate: float = 0.1, 
                 n_estimators: int = 100, max_depth: int = 3, 
                 min_samples_split: int = 2, min_samples_leaf: int = 1,
                 subsample: float = 1.0, random_state: int = 42):
        """
        Initialize Gradient Boosting Ensemble
        
        Args:
            loss: Loss function ('deviance' for classification, 'squared_error' for regression)
            learning_rate: Learning rate (shrinkage)
            n_estimators: Number of boosting stages
            max_depth: Maximum depth of trees
            min_samples_split: Minimum samples required to split
            min_samples_leaf: Minimum samples required at leaf
            subsample: Fraction of samples used for each tree
            random_state: Random seed
        """
        self.loss = loss
        self.learning_rate = learning_rate
        self.n_estimators = n_estimators
        self.max_depth = max_depth
        self.min_samples_split = min_samples_split
        self.min_samples_leaf = min_samples_leaf
        self.subsample = subsample
        self.random_state = random_state
        
        # To be filled during training
        self.estimators_ = []
        self.train_score_ = []
        self.oob_improvement_ = []
        self.initial_prediction_ = None
        
    def _get_loss_function(self):
        """Get loss function and its gradient"""
        if self.loss == 'deviance':
            # Logistic loss for binary classification
            def loss_fn(y_true, y_pred):
                y_pred = np.clip(y_pred, -250, 250)  # Prevent overflow
                return np.log(1 + np.exp(-y_true * y_pred))
            
            def gradient_fn(y_true, y_pred):
                y_pred = np.clip(y_pred, -250, 250)
                return -y_true / (1 + np.exp(y_true * y_pred))
            
            return loss_fn, gradient_fn
        
        elif self.loss == 'squared_error':
            # Squared loss for regression
            def loss_fn(y_true, y_pred):
                return 0.5 * (y_true - y_pred) ** 2
            
            def gradient_fn(y_true, y_pred):
                return y_true - y_pred
            
            return loss_fn, gradient_fn
        
        else:
            raise ValueError(f"Unknown loss function: {self.loss}")
    
    def fit(self, X: np.ndarray, y: np.ndarray):
        """Train Gradient Boosting ensemble"""
        np.random.seed(self.random_state)
        
        # Convert to binary labels for classification
        if self.loss == 'deviance':
            classes = np.unique(y)
            if len(classes) != 2:
                raise ValueError("Only binary classification supported")
            y_binary = np.where(y == classes[0], -1, 1)
            y_work = y_binary.astype(float)
        else:
            y_work = y.astype(float)
        
        n_samples = len(X)
        loss_fn, gradient_fn = self._get_loss_function()
        
        # Initialize with constant prediction
        if self.loss == 'deviance':
            # Initialize with log-odds
            pos_rate = np.sum(y_work == 1) / len(y_work)
            pos_rate = np.clip(pos_rate, 1e-15, 1 - 1e-15)  # Avoid log(0)
            self.initial_prediction_ = np.log(pos_rate / (1 - pos_rate))
        else:
            self.initial_prediction_ = np.mean(y_work)
        
        # Initialize predictions
        predictions = np.full(n_samples, self.initial_prediction_)
        
        print(f"🚀 Training Gradient Boosting with {self.n_estimators} estimators...")
        print(f"   Initial prediction: {self.initial_prediction_:.4f}")
        
        for stage in range(self.n_estimators):
            print(f"   Training stage {stage+1}/{self.n_estimators}...")
            
            # Calculate negative gradients (pseudo-residuals)
            residuals = -gradient_fn(y_work, predictions)
            
            # Subsample for stochastic gradient boosting
            if self.subsample < 1.0:
                n_subset = int(self.subsample * n_samples)
                indices = np.random.choice(n_samples, n_subset, replace=False)
                X_subset = X[indices]
                residuals_subset = residuals[indices]
            else:
                X_subset = X
                residuals_subset = residuals
                indices = np.arange(n_samples)
            
            # Fit base learner to residuals
            tree = DecisionTreeRegressor(
                max_depth=self.max_depth,
                min_samples_split=self.min_samples_split,
                min_samples_leaf=self.min_samples_leaf,
                random_state=self.random_state + stage
            )
            tree.fit(X_subset, residuals_subset)
            
            # Update predictions
            tree_pred = tree.predict(X)
            predictions += self.learning_rate * tree_pred
            
            # Store tree and training score
            self.estimators_.append(tree)
            train_loss = np.mean(loss_fn(y_work, predictions))
            self.train_score_.append(train_loss)
            
            if stage % 10 == 0:
                print(f"     Stage {stage+1}: Train loss = {train_loss:.4f}")
        
        print(f"✅ Gradient Boosting training complete!")
        return self
    
    def predict(self, X: np.ndarray) -> np.ndarray:
        """Make predictions"""
        if not self.estimators_:
            raise ValueError("Estimator not fitted yet")
        
        # Start with initial prediction
        predictions = np.full(len(X), self.initial_prediction_)
        
        # Add contributions from all trees
        for tree in self.estimators_:
            predictions += self.learning_rate * tree.predict(X)
        
        # Convert to class labels for classification
        if self.loss == 'deviance':
            # Convert from log-odds to probabilities, then to classes
            probs = 1 / (1 + np.exp(-predictions))
            return (probs >= 0.5).astype(int)
        else:
            return predictions
    
    def predict_proba(self, X: np.ndarray) -> np.ndarray:
        """Predict class probabilities (for classification only)"""
        if self.loss != 'deviance':
            raise ValueError("predict_proba only available for classification")
        
        # Start with initial prediction
        predictions = np.full(len(X), self.initial_prediction_)
        
        # Add contributions from all trees
        for tree in self.estimators_:
            predictions += self.learning_rate * tree.predict(X)
        
        # Convert to probabilities
        probs_class_1 = 1 / (1 + np.exp(-predictions))
        probs_class_0 = 1 - probs_class_1
        
        return np.column_stack((probs_class_0, probs_class_1))
    
    def staged_predict(self, X: np.ndarray):
        """Return predictions at each boosting stage"""
        staged_predictions = []
        predictions = np.full(len(X), self.initial_prediction_)
        
        for tree in self.estimators_:
            predictions += self.learning_rate * tree.predict(X)
            
            if self.loss == 'deviance':
                probs = 1 / (1 + np.exp(-predictions))
                stage_pred = (probs >= 0.5).astype(int)
            else:
                stage_pred = predictions.copy()
            
            staged_predictions.append(stage_pred)
        
        return staged_predictions
    
    def get_feature_importance(self) -> np.ndarray:
        """Compute feature importance"""
        if not self.estimators_:
            raise ValueError("Estimator not fitted yet")
        
        importances = np.zeros(self.estimators_[0].n_features_in_)
        
        for tree in self.estimators_:
            importances += tree.feature_importances_
        
        # Normalize
        importances /= len(self.estimators_)
        return importances


print("🚀 Advanced Boosting implementations ready!")
print("   📈 AdaBoostEnsemble: Adaptive boosting with weight tracking")
print("   🎯 GradientBoostingEnsemble: Gradient boosting with multiple loss functions")
```
