# Mathematical Basis of Generative AI

Generative AI refers to a class of artificial intelligence models that can generate new content, such as images, text, or music, that is similar to existing data. The mathematical basis of generative AI involves several key concepts and techniques from probability theory, statistics, and machine learning. Here are the foundational mathematical principles:

## 1. Probability Theory
- **Probability Distributions**: Generative models learn the probability distribution of the training data. Common distributions include Gaussian (normal), Bernoulli, and Multinomial distributions.
- **Joint and Marginal Probability**: Understanding the joint probability of multiple variables and their marginal probabilities is crucial for modeling complex data distributions.

## 2. Statistics
- **Maximum Likelihood Estimation (MLE)**: MLE is used to estimate the parameters of a probability distribution by maximizing the likelihood function.
- **Bayesian Inference**: Bayesian methods incorporate prior knowledge into the model and update this knowledge as more data becomes available.

## 3. Machine Learning
- **Neural Networks**: Neural networks, particularly deep learning models, are the backbone of modern generative AI. They can model complex data distributions through layers of nonlinear transformations.
- **Optimization**: Techniques like gradient descent are used to optimize the parameters of generative models to fit the training data.

## 4. Specific Generative Models
- **Generative Adversarial Networks (GANs)**: GANs consist of two neural networks, a generator and a discriminator, that are trained simultaneously. The generator creates fake data, while the discriminator tries to distinguish between real and fake data. The objective is a minimax game where the generator aims to fool the discriminator.
- **Variational Autoencoders (VAEs)**: VAEs use a probabilistic approach to model data. They consist of an encoder that maps input data to a latent space and a decoder that reconstructs the data from the latent space. VAEs optimize a loss function that includes a reconstruction term and a regularization term based on the Kullback-Leibler divergence.
- **Autoregressive Models**: Models like PixelRNN, PixelCNN, and Transformer-based models generate data sequentially, predicting each part of the data conditioned on the previous parts. These models use techniques like masked convolutions and attention mechanisms.
- **Flow-based Models**: These models, such as Normalizing Flows, learn invertible transformations that map complex data distributions to simple ones (e.g., Gaussian). They use the change of variables formula to compute exact likelihoods.

## 5. Latent Variable Models
- **Latent Variables**: These are unobserved variables that can explain the observed data. Generative models often use latent variables to capture underlying structure in the data.
- **Expectation-Maximization (EM) Algorithm**: EM is a method for finding maximum likelihood estimates in models with latent variables. It alternates between estimating the latent variables given the current parameters (E-step) and updating the parameters given the latent variables (M-step).

## 6. Information Theory
- **Entropy and Mutual Information**: These concepts measure the uncertainty and the amount of information shared between variables, respectively. They are often used to design objective functions for generative models.

In summary, the mathematical basis of generative AI involves a combination of probability theory, statistics, optimization, and various machine learning techniques. By leveraging these principles, generative models can learn to produce new data that closely resembles the training data.


# Generative Adversarial Networks (GANs)

Generative Adversarial Networks (GANs) are a class of generative models introduced by Ian Goodfellow and his colleagues in 2014 . GANs consist of two neural networks, a generator and a discriminator, which are trained simultaneously through adversarial learning. The mathematical explanation of GANs involves understanding the optimization problem they solve and the training dynamics between the generator and the discriminator.

## Components of GANs

1. **Generator (G)**:
   - The generator is a neural network that takes a random noise vector $z$ (sampled from a prior distribution, typically Gaussian or uniform) and maps it to the data space. The objective of the generator is to produce data that is indistinguishable from the real data.
   - The generator is denoted as $G(z; \theta_G)$, where $\theta_G$ represents the parameters of the generator.

2. **Discriminator (D)**:
   - The discriminator is a neural network that takes an input (either real data or generated data) and outputs a probability indicating whether the input is real or fake.
   - The discriminator is denoted as $D(x; \theta_D)$, where $\theta_D$ represents the parameters of the discriminator.

## Objective Function

The training process of GANs can be understood as a two-player minimax game with the following value function $V(D, G)$:

$$ \min_G \max_D V(D, G) = \mathbb{E}_{x \sim p_{\text{data}}(x)} [\log D(x)] + \mathbb{E}_{z \sim p_z(z)} [\log (1 - D(G(z)))] $$

Here:
- $p_{\text{data}}(x)$ is the distribution of the real data.
- $p_z(z)$ is the prior distribution of the noise vector $z$.

## Discriminator Objective

The discriminator aims to maximize the probability of correctly classifying real and generated data. This can be expressed as:

$$ \max_D V(D, G) = \mathbb{E}_{x \sim p_{\text{data}}(x)} [\log D(x)] + \mathbb{E}_{z \sim p_z(z)} [\log (1 - D(G(z)))] $$

## Generator Objective

The generator aims to minimize the probability that the discriminator correctly classifies the generated data as fake. This can be expressed as:

$$ \min_G V(D, G) = \mathbb{E}_{z \sim p_z(z)} [\log (1 - D(G(z)))] $$

However, in practice, minimizing $\log (1 - D(G(z)))$ can lead to slow gradients when $D(G(z))$ is near 0. Therefore, an alternative objective for the generator is often used:

$$ \max_G \mathbb{E}_{z \sim p_z(z)} [\log D(G(z))] $$

This alternative objective aims to maximize the log probability of the discriminator being mistaken, which provides stronger gradients early in training.

## Training Process

1. **Discriminator Update**:
   - Sample a minibatch of $m$ real data examples $\{x^{(1)}, x^{(2)}, \dots, x^{(m)}\}$ from the data distribution $p_{\text{data}}(x)$.
   - Sample a minibatch of $m$ noise vectors $\{z^{(1)}, z^{(2)}, \dots, z^{(m)}\}$ from the prior distribution $p_z(z)$.
   - Update the discriminator by ascending its stochastic gradient:

$$ \nabla_{\theta_D} \frac{1}{m} \sum_{i=1}^{m} \left[ \log D(x^{(i)}) + \log (1 - D(G(z^{(i)}))) \right] $$

2. **Generator Update**:
   - Sample a minibatch of $m$ noise vectors $\{z^{(1)}, z^{(2)}, \dots, z^{(m)}\}$ from the prior distribution $p_z(z)$.
   - Update the generator by descending its stochastic gradient:

$$ \nabla_{\theta_G} \frac{1}{m} \sum_{i=1}^{m} \log D(G(z^{(i)})) $$

## Convergence

In theory, GANs reach a Nash equilibrium when the generator captures the data distribution perfectly. At this point, the discriminator cannot distinguish between real and generated data, meaning $D(x) = 0.5$ for all $x$. However, in practice, achieving this equilibrium is challenging, and training GANs can be unstable. Various techniques, such as feature matching, batch normalization, and Wasserstein GANs (WGANs), have been developed to improve the stability and performance of GAN training  .

## Summary

GANs are powerful generative models that frame the training process as a game between a generator and a discriminator. The generator aims to produce realistic data, while the discriminator aims to distinguish between real and fake data. Through this adversarial process, GANs learn to generate high-quality data that mimics the underlying data distribution.

```python
# Complete working GAN implementation with file-based visualization
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision.utils import make_grid, save_image
import matplotlib.pyplot as plt
import os
from tqdm import tqdm
import time

print("✓ All libraries imported successfully!")

# Check device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"✓ Using device: {device}")

# Create directories
os.makedirs('gan_models', exist_ok=True)
os.makedirs('gan_outputs', exist_ok=True)

# Configuration
class GANConfig:
    def __init__(self):
        self.latent_dim = 100
        self.batch_size = 64
        self.learning_rate = 0.0002
        self.num_epochs = 10
        self.beta1 = 0.5
        self.beta2 = 0.999
        self.sample_interval = 200
        self.checkpoint_interval = 5
        
config = GANConfig()

# File-based tensor display function that avoids numpy conversion
def show_tensor_images(images, title="Generated Images", nrow=5):
    """Display tensor images by saving to file first"""
    save_image(images, 'temp_display.png', nrow=nrow, normalize=True)
    
    from PIL import Image
    img = Image.open('temp_display.png')
    
    plt.figure(figsize=(8, 8))
    plt.imshow(img)
    plt.title(title)
    plt.axis('off')
    plt.show()
    
    # Clean up
    try:
        os.remove('temp_display.png')
    except:
        pass

# Synthetic dataset class
class SyntheticMNIST:
    def __init__(self, num_samples=3000):
        self.num_samples = num_samples
        print(f"Generating {num_samples} synthetic images...")
        self.data = self._generate_data()
    
    def _generate_data(self):
        data = []
        for i in range(self.num_samples):
            img = torch.zeros(1, 28, 28)
            pattern_type = torch.randint(0, 4, (1,)).item()
            
            if pattern_type == 0:  # Circle
                center = 14
                for r in range(6, 10):
                    for angle in range(0, 360, 10):
                        x = int(center + r * torch.cos(torch.tensor(angle * 3.14159 / 180)))
                        y = int(center + r * torch.sin(torch.tensor(angle * 3.14159 / 180)))
                        if 0 <= x < 28 and 0 <= y < 28:
                            img[0, y, x] = 1.0
            
            elif pattern_type == 1:  # Vertical line
                x_pos = torch.randint(12, 16, (1,)).item()
                img[0, 4:24, x_pos:x_pos+2] = 1.0
            
            elif pattern_type == 2:  # Horizontal line
                y_pos = torch.randint(12, 16, (1,)).item()
                img[0, y_pos:y_pos+2, 4:24] = 1.0
            
            else:  # Random shapes
                for _ in range(2):
                    x1 = torch.randint(5, 20, (1,)).item()
                    y1 = torch.randint(5, 20, (1,)).item()
                    w = torch.randint(3, 6, (1,)).item()
                    h = torch.randint(3, 6, (1,)).item()
                    img[0, y1:y1+h, x1:x1+w] = torch.rand(h, w) * 0.8
            
            # Add noise and normalize
            img = img + torch.randn_like(img) * 0.05
            img = torch.clamp(img, 0, 1)
            img = (img - 0.5) / 0.5  # Normalize to [-1, 1]
            
            data.append(img)
        
        return torch.stack(data)
    
    def __len__(self):
        return self.num_samples
    
    def __getitem__(self, idx):
        return self.data[idx], 0

# Create dataset and loader
print("Creating dataset...")
dataset = SyntheticMNIST(3000)
train_loader = torch.utils.data.DataLoader(
    dataset, batch_size=config.batch_size, shuffle=True, 
    num_workers=0, drop_last=True
)

print(f"✓ Dataset: {len(dataset)} samples, {len(train_loader)} batches")

# Show samples using file-based display
print("Displaying sample images...")
sample_batch, _ = next(iter(train_loader))
show_tensor_images(sample_batch[:16], "Sample Synthetic Dataset Images", nrow=4)

print("✓ Sample display successful!")
```

```python
# Define GAN models and setup
class Generator(nn.Module):
    def __init__(self, latent_dim=100):
        super(Generator, self).__init__()
        self.latent_dim = latent_dim
        
        self.model = nn.Sequential(
            nn.Linear(latent_dim, 256),
            nn.BatchNorm1d(256),
            nn.LeakyReLU(0.2, inplace=True),
            
            nn.Linear(256, 512),
            nn.BatchNorm1d(512),
            nn.LeakyReLU(0.2, inplace=True),
            
            nn.Linear(512, 1024),
            nn.BatchNorm1d(1024),
            nn.LeakyReLU(0.2, inplace=True),
            
            nn.Linear(1024, 784),
            nn.Tanh()
        )

    def forward(self, z):
        if z.dim() == 1:
            z = z.unsqueeze(0)
        img = self.model(z)
        return img.view(img.size(0), 1, 28, 28)

class Discriminator(nn.Module):
    def __init__(self):
        super(Discriminator, self).__init__()
        
        self.model = nn.Sequential(
            nn.Linear(784, 512),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Dropout(0.3),
            
            nn.Linear(512, 256),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Dropout(0.3),
            
            nn.Linear(256, 128),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Dropout(0.3),
            
            nn.Linear(128, 1),
            nn.Sigmoid()
        )

    def forward(self, img):
        return self.model(img.view(img.size(0), -1))

# Initialize models
generator = Generator(config.latent_dim).to(device)
discriminator = Discriminator().to(device)

print(f"✓ Generator parameters: {sum(p.numel() for p in generator.parameters()):,}")
print(f"✓ Discriminator parameters: {sum(p.numel() for p in discriminator.parameters()):,}")

# Loss and optimizers
adversarial_loss = nn.BCELoss()
optimizer_G = optim.Adam(generator.parameters(), lr=config.learning_rate, betas=(config.beta1, config.beta2))
optimizer_D = optim.Adam(discriminator.parameters(), lr=config.learning_rate, betas=(config.beta1, config.beta2))

# Fixed noise for consistent generation
fixed_noise = torch.randn(25, config.latent_dim, device=device)

# Helper functions
def save_checkpoint(epoch, g_loss, d_loss):
    torch.save({
        'epoch': epoch,
        'generator_state_dict': generator.state_dict(),
        'discriminator_state_dict': discriminator.state_dict(),
        'optimizer_G_state_dict': optimizer_G.state_dict(),
        'optimizer_D_state_dict': optimizer_D.state_dict(),
        'g_loss': g_loss, 'd_loss': d_loss
    }, f'gan_models/checkpoint_epoch_{epoch}.pth')

# Test generator before training
print("Testing generator...")
with torch.no_grad():
    test_imgs = generator(fixed_noise[:4])
    show_tensor_images(test_imgs, "Initial Generator Output (Random Noise)")

print("✓ Models ready for training!")
```

```python
# Complete training and evaluation functions
def train_gan():
    print(f"🚀 Starting GAN training for {config.num_epochs} epochs")
    print(f"Device: {device} | Batch size: {config.batch_size}")
    
    G_losses = []
    D_losses = []
    start_time = time.time()
    
    for epoch in range(config.num_epochs):
        epoch_g_loss = 0.0
        epoch_d_loss = 0.0
        
        pbar = tqdm(train_loader, desc=f'Epoch {epoch+1}/{config.num_epochs}')
        
        for batch_idx, (real_imgs, _) in enumerate(pbar):
            batch_size = real_imgs.size(0)
            real_imgs = real_imgs.to(device)
            
            # Labels with smoothing for stability
            real_labels = torch.ones(batch_size, 1, device=device) * 0.9
            fake_labels = torch.zeros(batch_size, 1, device=device)
            
            # Train Discriminator
            optimizer_D.zero_grad()
            
            # Real images
            real_output = discriminator(real_imgs)
            d_loss_real = adversarial_loss(real_output, real_labels)
            
            # Fake images
            noise = torch.randn(batch_size, config.latent_dim, device=device)
            fake_imgs = generator(noise)
            fake_output = discriminator(fake_imgs.detach())
            d_loss_fake = adversarial_loss(fake_output, fake_labels)
            
            d_loss = (d_loss_real + d_loss_fake) / 2
            d_loss.backward()
            optimizer_D.step()
            
            # Train Generator
            optimizer_G.zero_grad()
            fake_output = discriminator(fake_imgs)
            g_loss = adversarial_loss(fake_output, real_labels)
            g_loss.backward()
            optimizer_G.step()
            
            # Update losses
            epoch_g_loss += g_loss.item()
            epoch_d_loss += d_loss.item()
            
            # Update progress bar
            pbar.set_postfix({
                'D_loss': f'{d_loss.item():.4f}',
                'G_loss': f'{g_loss.item():.4f}',
                'D(real)': f'{real_output.mean().item():.3f}',
                'D(fake)': f'{fake_output.mean().item():.3f}'
            })
        
        # Calculate epoch averages
        avg_g_loss = epoch_g_loss / len(train_loader)
        avg_d_loss = epoch_d_loss / len(train_loader)
        
        G_losses.append(avg_g_loss)
        D_losses.append(avg_d_loss)
        
        # Show progress using file-based visualization
        with torch.no_grad():
            fake_samples = generator(fixed_noise)
            show_tensor_images(fake_samples, f"Epoch {epoch+1} - Generated Images")
        
        # Save checkpoint
        if (epoch + 1) % config.checkpoint_interval == 0:
            save_checkpoint(epoch + 1, avg_g_loss, avg_d_loss)
        
        print(f"Epoch {epoch+1}: G_loss={avg_g_loss:.4f}, D_loss={avg_d_loss:.4f}")
    
    total_time = time.time() - start_time
    print(f"\\n🎉 Training completed in {total_time/60:.2f} minutes!")
    
    return G_losses, D_losses

# Plotting function without numpy dependency
def plot_losses(G_losses, D_losses):
    if not G_losses or not D_losses:
        print("No loss data to plot")
        return
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))
    epochs = list(range(1, len(G_losses) + 1))
    
    # Raw losses
    ax1.plot(epochs, G_losses, 'b-', label='Generator', linewidth=2)
    ax1.plot(epochs, D_losses, 'r-', label='Discriminator', linewidth=2)
    ax1.set_xlabel('Epoch')
    ax1.set_ylabel('Loss')
    ax1.set_title('Training Losses')
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    # Simple smoothed version
    if len(G_losses) >= 3:
        smooth_g = []
        smooth_d = []
        for i in range(1, len(G_losses) - 1):
            smooth_g.append((G_losses[i-1] + G_losses[i] + G_losses[i+1]) / 3)
            smooth_d.append((D_losses[i-1] + D_losses[i] + D_losses[i+1]) / 3)
        
        smooth_epochs = list(range(2, len(smooth_g) + 2))
        ax2.plot(smooth_epochs, smooth_g, 'b-', label='Generator (Smoothed)', linewidth=2)
        ax2.plot(smooth_epochs, smooth_d, 'r-', label='Discriminator (Smoothed)', linewidth=2)
        ax2.set_title('Smoothed Losses')
    else:
        ax2.plot(epochs, G_losses, 'b-', label='Generator', linewidth=2)
        ax2.plot(epochs, D_losses, 'r-', label='Discriminator', linewidth=2)
        ax2.set_title('Training Losses')
    
    ax2.set_xlabel('Epoch')
    ax2.set_ylabel('Loss')
    ax2.legend()
    ax2.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()

def evaluate_generator(num_samples=64):
    print(f"Generating {num_samples} samples for evaluation...")
    
    generator.eval()
    all_samples = []
    
    with torch.no_grad():
        for i in range(0, num_samples, config.batch_size):
            batch_size = min(config.batch_size, num_samples - i)
            noise = torch.randn(batch_size, config.latent_dim, device=device)
            samples = generator(noise).cpu()
            all_samples.append(samples)
    
    all_samples = torch.cat(all_samples, dim=0)[:num_samples]
    
    # Calculate stats using tensor operations
    mean_val = all_samples.mean().item()
    std_val = all_samples.std().item()
    min_val = all_samples.min().item()
    max_val = all_samples.max().item()
    
    print(f"Sample statistics:")
    print(f"  Mean: {mean_val:.4f}")
    print(f"  Std: {std_val:.4f}")
    print(f"  Min: {min_val:.4f}")
    print(f"  Max: {max_val:.4f}")
    
    # Display using file-based visualization
    show_tensor_images(all_samples[:64], f"Generated Samples ({len(all_samples)} total)", nrow=8)
    
    generator.train()
    return all_samples

def compare_real_vs_fake():
    real_batch, _ = next(iter(train_loader))
    
    generator.eval()
    with torch.no_grad():
        noise = torch.randn(16, config.latent_dim, device=device)
        fake_batch = generator(noise).cpu()
    generator.train()
    
    # Save both as separate images and display side by side
    save_image(real_batch[:16], 'temp_real.png', nrow=4, normalize=True)
    save_image(fake_batch, 'temp_fake.png', nrow=4, normalize=True)
    
    from PIL import Image
    real_img = Image.open('temp_real.png')
    fake_img = Image.open('temp_fake.png')
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 8))
    
    ax1.imshow(real_img)
    ax1.set_title('Real Samples (Dataset)')
    ax1.axis('off')
    
    ax2.imshow(fake_img)
    ax2.set_title('Generated Samples')
    ax2.axis('off')
    
    plt.tight_layout()
    plt.show()
    
    # Clean up
    try:
        os.remove('temp_real.png')
        os.remove('temp_fake.png')
    except:
        pass

print("✓ All training functions ready!")
print("\\n" + "="*50)
print("🎯 Commands:")
print("• Train GAN:         G_losses, D_losses = train_gan()")
print("• Plot losses:       plot_losses(G_losses, D_losses)")
print("• Evaluate:          evaluate_generator()")
print("• Compare samples:   compare_real_vs_fake()")
print("="*50)
print("\\n🚀 Ready to train! Run: G_losses, D_losses = train_gan()")
```

```python
# Optional: WGAN-GP Implementation for advanced users
# This cell provides an alternative WGAN-GP implementation for improved stability

class WGANDiscriminator(nn.Module):
    def __init__(self):
        super(WGANDiscriminator, self).__init__()
        
        self.model = nn.Sequential(
            nn.Linear(784, 512),
            nn.LeakyReLU(0.2, inplace=True),
            
            nn.Linear(512, 256),
            nn.LeakyReLU(0.2, inplace=True),
            
            nn.Linear(256, 128),
            nn.LeakyReLU(0.2, inplace=True),
            
            nn.Linear(128, 1)
            # No sigmoid for WGAN
        )

    def forward(self, img):
        return self.model(img.view(img.size(0), -1))

def gradient_penalty(discriminator, real_samples, fake_samples, device):
    """Calculates gradient penalty for WGAN-GP"""
    batch_size = real_samples.size(0)
    alpha = torch.rand(batch_size, 1, 1, 1, device=device)
    interpolates = (alpha * real_samples + ((1 - alpha) * fake_samples)).requires_grad_(True)
    
    d_interpolates = discriminator(interpolates)
    gradients = torch.autograd.grad(
        outputs=d_interpolates,
        inputs=interpolates,
        grad_outputs=torch.ones_like(d_interpolates, device=device),
        create_graph=True,
        retain_graph=True,
        only_inputs=True
    )[0]
    
    gradients = gradients.view(gradients.size(0), -1)
    gradient_penalty = ((gradients.norm(2, dim=1) - 1) ** 2).mean()
    return gradient_penalty

def train_wgan_gp():
    """Train using WGAN-GP for improved stability"""
    wgan_discriminator = WGANDiscriminator().to(device)
    wgan_optimizer_D = optim.Adam(wgan_discriminator.parameters(), lr=config.learning_rate, betas=(0.0, 0.9))
    
    lambda_gp = 10
    n_critic = 5
    
    G_losses = []
    D_losses = []
    start_time = time.time()
    
    print(f"Starting WGAN-GP training for {config.num_epochs} epochs...")
    
    for epoch in range(config.num_epochs):
        epoch_g_loss = 0.0
        epoch_d_loss = 0.0
        
        pbar = tqdm(train_loader, desc=f'WGAN-GP Epoch {epoch+1}/{config.num_epochs}')
        
        for batch_idx, (real_imgs, _) in enumerate(pbar):
            batch_size = real_imgs.size(0)
            real_imgs = real_imgs.to(device)
            
            # Train Discriminator multiple times
            for _ in range(n_critic):
                wgan_optimizer_D.zero_grad()
                noise = torch.randn(batch_size, config.latent_dim, device=device)
                fake_imgs = generator(noise)
                
                real_validity = wgan_discriminator(real_imgs)
                fake_validity = wgan_discriminator(fake_imgs.detach())
                gp = gradient_penalty(wgan_discriminator, real_imgs, fake_imgs, device)
                
                d_loss = -torch.mean(real_validity) + torch.mean(fake_validity) + lambda_gp * gp
                d_loss.backward()
                wgan_optimizer_D.step()
            
            # Train Generator
            if batch_idx % n_critic == 0:
                optimizer_G.zero_grad()
                noise = torch.randn(batch_size, config.latent_dim, device=device)
                fake_imgs = generator(noise)
                fake_validity = wgan_discriminator(fake_imgs)
                g_loss = -torch.mean(fake_validity)
                g_loss.backward()
                optimizer_G.step()
                epoch_g_loss += g_loss.item()
            
            epoch_d_loss += d_loss.item()
            
            pbar.set_postfix({
                'D_loss': f'{d_loss.item():.4f}',
                'G_loss': f'{g_loss.item():.4f}' if 'g_loss' in locals() else 'N/A',
                'GP': f'{gp.item():.4f}'
            })
        
        avg_g_loss = epoch_g_loss / (len(train_loader) // n_critic)
        avg_d_loss = epoch_d_loss / len(train_loader)
        
        G_losses.append(avg_g_loss)
        D_losses.append(avg_d_loss)
        
        with torch.no_grad():
            fake_imgs = generator(fixed_noise)
            show_tensor_images(fake_imgs, f"WGAN-GP Epoch {epoch+1}")
        
        print(f"WGAN-GP Epoch {epoch+1}: G_loss={avg_g_loss:.4f}, D_loss={avg_d_loss:.4f}")
    
    total_time = time.time() - start_time
    print(f"WGAN-GP training completed in {total_time/60:.2f} minutes")
    
    return G_losses, D_losses, wgan_discriminator

print("✓ WGAN-GP implementation available!")
print("To use: G_losses, D_losses, wgan_disc = train_wgan_gp()")
```
