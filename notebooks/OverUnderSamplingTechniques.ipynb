## 🎯 **COMPREHENSIVE BEST PRACTICES GUIDE**

### **1. Method Selection Framework**

#### **By Imbalance Severity:**
- **Mild (1:2 to 1:4)**: Random sampling, simple SMOTE
- **Moderate (1:5 to 1:20)**: SMOTE, BorderlineSMOTE, ADASYN
- **Severe (1:20 to 1:100)**: ADASYN, hybrid methods, cost-sensitive
- **Extreme (>1:100)**: Ensemble methods, specialized techniques

#### **By Data Characteristics:**
- **Clean, well-separated**: Standard SMOTE
- **Noisy, overlapping**: Hybrid methods (SMOTETomek, SMOTEENN)
- **High-dimensional**: SVMSMOTE, dimensionality reduction first
- **Mixed data types**: SMOTE-NC, careful preprocessing

#### **By Application Requirements:**
- **High recall needed**: Oversampling methods, cost-sensitive
- **High precision needed**: Conservative undersampling
- **Balanced performance**: Hybrid approaches
- **Real-time constraints**: Simple methods, pre-computed samples

### **3. Evaluation Guidelines**

#### **Metrics Selection:**
- **Primary**: F1-Score, AUC-ROC, AUC-PR
- **Class-specific**: Precision, Recall per class
- **Business**: Cost-weighted metrics, domain-specific measures

#### **Validation Strategy:**
- **Time-series data**: Time-based splits
- **Stratified sampling**: Maintain class proportions
- **Nested CV**: Hyperparameter tuning + performance estimation

### **4. Common Pitfalls and Solutions**

#### **❌ Common Mistakes:**
1. **Applying sampling before train/test split**
   - **Solution**: Split first, then sample training set only

2. **Using accuracy as primary metric**
   - **Solution**: Focus on F1, AUC, precision/recall

3. **Ignoring data quality issues**
   - **Solution**: Clean data before sampling

4. **Over-sampling without considering computational cost**
   - **Solution**: Balance performance vs. efficiency

#### **✅ Best Practices:**
1. **Always use stratified splitting**
2. **Test multiple methods systematically**
3. **Consider domain expertise in method selection**
4. **Monitor performance on new data**
5. **Document sampling decisions and rationale**

````python
# 🎯 INTERACTIVE BEFORE/AFTER SAMPLING VISUALIZATION

class InteractiveSamplingVisualizer:
    """
    Interactive visualization tool for before/after sampling effects
    """
    
    def __init__(self):
        """Initialize the interactive visualizer"""
        self.methods = {
            'SMOTE': SMOTE(random_state=42),
            'ADASYN': ADASYN(random_state=42),
            'BorderlineSMOTE': BorderlineSMOTE(random_state=42),
            'RandomOverSampler': RandomOverSampler(random_state=42),
            'TomekLinks': TomekLinks(),
            'EditedNearestNeighbours': EditedNearestNeighbours(),
            'NearMiss': NearMiss(version=1),
            'SMOTETomek': SMOTETomek(random_state=42),
            'SMOTEENN': SMOTEENN(random_state=42)
        }
    
    def create_before_after_comparison(self, X, y, method_name, figsize=(16, 6)):
        """
        Create before/after comparison visualization
        """
        
        if method_name not in self.methods:
            print(f"Method {method_name} not available")
            return
        
        # Apply sampling method
        method = self.methods[method_name]
        X_resampled, y_resampled = method.fit_resample(X, y)
        
        # Create subplot
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=figsize)
        
        colors = ['#FF6B6B', '#4ECDC4']  # Red for majority, Teal for minority
        
        # BEFORE - Original dataset
        for class_val, color in zip([0, 1], colors):
            mask = y == class_val
            class_name = 'Majority' if class_val == 0 else 'Minority'
            ax1.scatter(X[mask, 0], X[mask, 1], c=color, alpha=0.7, s=50,
                       label=f'{class_name} (n={np.sum(mask)})')
        
        # Calculate original statistics
        orig_counts = Counter(y)
        orig_ratio = orig_counts[0] / orig_counts[1]
        
        ax1.set_title(f'BEFORE: Original Dataset\\n'
                     f'Total: {len(X)} | Ratio: {orig_ratio:.1f}:1',
                     fontweight='bold', fontsize=12)
        ax1.legend()
        ax1.grid(True, alpha=0.3)
        ax1.set_xlabel('Feature 1')
        ax1.set_ylabel('Feature 2')
        
        # AFTER - Resampled dataset
        for class_val, color in zip([0, 1], colors):
            mask = y_resampled == class_val
            class_name = 'Majority' if class_val == 0 else 'Minority'
            ax2.scatter(X_resampled[mask, 0], X_resampled[mask, 1], c=color, alpha=0.7, s=50,
                       label=f'{class_name} (n={np.sum(mask)})')
        
        # Calculate resampled statistics
        resamp_counts = Counter(y_resampled)
        resamp_ratio = resamp_counts[0] / resamp_counts[1]
        
        ax2.set_title(f'AFTER: {method_name}\\n'
                     f'Total: {len(X_resampled)} | Ratio: {resamp_ratio:.1f}:1',
                     fontweight='bold', fontsize=12)
        ax2.legend()
        ax2.grid(True, alpha=0.3)
        ax2.set_xlabel('Feature 1')
        ax2.set_ylabel('Feature 2')
        
        plt.suptitle(f'🎯 Before/After Comparison: {method_name}', 
                     fontsize=14, fontweight='bold', y=1.02)
        plt.tight_layout()
        plt.show()
        
        # Print detailed statistics
        self._print_sampling_statistics(X, y, X_resampled, y_resampled, method_name)
        
        return X_resampled, y_resampled
    
    def _print_sampling_statistics(self, X_orig, y_orig, X_resamp, y_resamp, method_name):
        """Print detailed sampling statistics"""
        
        orig_counts = Counter(y_orig)
        resamp_counts = Counter(y_resamp)
        
        print(f"\\n📊 DETAILED STATISTICS: {method_name}")
        print("=" * 50)
        print(f"ORIGINAL DATASET:")
        print(f"  Total samples: {len(X_orig)}")
        print(f"  Majority class: {orig_counts[0]} ({orig_counts[0]/len(y_orig)*100:.1f}%)")
        print(f"  Minority class: {orig_counts[1]} ({orig_counts[1]/len(y_orig)*100:.1f}%)")
        print(f"  Imbalance ratio: {orig_counts[0]/orig_counts[1]:.1f}:1")
        
        print(f"\\nRESAMPLED DATASET:")
        print(f"  Total samples: {len(X_resamp)}")
        print(f"  Majority class: {resamp_counts[0]} ({resamp_counts[0]/len(y_resamp)*100:.1f}%)")
        print(f"  Minority class: {resamp_counts[1]} ({resamp_counts[1]/len(y_resamp)*100:.1f}%)")
        print(f"  Imbalance ratio: {resamp_counts[0]/resamp_counts[1]:.1f}:1")
        
        print(f"\\nCHANGES:")
        maj_change = resamp_counts[0] - orig_counts[0]
        min_change = resamp_counts[1] - orig_counts[1]
        total_change = len(X_resamp) - len(X_orig)
        
        print(f"  Majority class: {maj_change:+d} samples ({maj_change/orig_counts[0]*100:+.1f}%)")
        print(f"  Minority class: {min_change:+d} samples ({min_change/orig_counts[1]*100:+.1f}%)")
        print(f"  Total samples: {total_change:+d} samples ({total_change/len(X_orig)*100:+.1f}%)")
    
    def create_method_grid_comparison(self, X, y, selected_methods=None, figsize=(20, 15)):
        """
        Create grid comparison of multiple methods
        """
        
        if selected_methods is None:
            selected_methods = ['SMOTE', 'ADASYN', 'TomekLinks', 'SMOTETomek']
        
        n_methods = len(selected_methods)
        cols = 3
        rows = (n_methods + cols - 1) // cols
        
        fig, axes = plt.subplots(rows, cols, figsize=figsize)
        if rows == 1:
            axes = axes.reshape(1, -1)
        elif cols == 1:
            axes = axes.reshape(-1, 1)
        
        colors = ['#FF6B6B', '#4ECDC4']
        
        for idx, method_name in enumerate(selected_methods):
            row = idx // cols
            col = idx % cols
            ax = axes[row, col]
            
            # Apply sampling method
            method = self.methods[method_name]
            X_resampled, y_resampled = method.fit_resample(X, y)
            
            # Plot resampled data
            for class_val, color in zip([0, 1], colors):
                mask = y_resampled == class_val
                class_name = 'Majority' if class_val == 0 else 'Minority'
                ax.scatter(X_resampled[mask, 0], X_resampled[mask, 1], 
                          c=color, alpha=0.7, s=30,
                          label=f'{class_name} (n={np.sum(mask)})')
            
            # Calculate statistics
            resamp_counts = Counter(y_resampled)
            resamp_ratio = resamp_counts[0] / resamp_counts[1]
            
            ax.set_title(f'{method_name}\\n'
                        f'Total: {len(X_resampled)} | Ratio: {resamp_ratio:.1f}:1',
                        fontweight='bold', fontsize=11)
            ax.legend(fontsize=9)
            ax.grid(True, alpha=0.3)
            ax.set_xlabel('Feature 1')
            ax.set_ylabel('Feature 2')
        
        # Hide unused subplots
        for idx in range(len(selected_methods), rows * cols):
            row = idx // cols
            col = idx % cols
            axes[row, col].set_visible(False)
        
        plt.suptitle('🎯 Multiple Sampling Methods Comparison', 
                     fontsize=16, fontweight='bold', y=0.98)
        plt.tight_layout()
        plt.show()
    
    def create_decision_boundary_comparison(self, X, y, method_name, classifier=None):
        """
        Compare decision boundaries before and after sampling
        """
        
        if classifier is None:
            classifier = LogisticRegression(random_state=42)
        
        # Apply sampling
        method = self.methods[method_name]
        X_resampled, y_resampled = method.fit_resample(X, y)
        
        # Create mesh for decision boundary
        h = 0.02
        x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
        y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
        xx, yy = np.meshgrid(np.arange(x_min, x_max, h),
                           np.arange(y_min, y_max, h))
        
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))
        colors = ['#FF6B6B', '#4ECDC4']
        
        # Train on original data
        clf_orig = classifier.__class__(**classifier.get_params())
        clf_orig.fit(X, y)
        Z_orig = clf_orig.predict(np.c_[xx.ravel(), yy.ravel()])
        Z_orig = Z_orig.reshape(xx.shape)
        
        ax1.contourf(xx, yy, Z_orig, alpha=0.3, colors=['#FFB3B3', '#B3F0E6'])
        ax1.contour(xx, yy, Z_orig, colors='black', linewidths=0.5)
        
        for class_val, color in zip([0, 1], colors):
            mask = y == class_val
            ax1.scatter(X[mask, 0], X[mask, 1], c=color, alpha=0.8, s=50)
        
        ax1.set_title('Original Dataset\\nDecision Boundary', fontweight='bold')
        ax1.grid(True, alpha=0.3)
        
        # Train on resampled data
        clf_resamp = classifier.__class__(**classifier.get_params())
        clf_resamp.fit(X_resampled, y_resampled)
        Z_resamp = clf_resamp.predict(np.c_[xx.ravel(), yy.ravel()])
        Z_resamp = Z_resamp.reshape(xx.shape)
        
        ax2.contourf(xx, yy, Z_resamp, alpha=0.3, colors=['#FFB3B3', '#B3F0E6'])
        ax2.contour(xx, yy, Z_resamp, colors='black', linewidths=0.5)
        
        for class_val, color in zip([0, 1], colors):
            mask = y_resampled == class_val
            ax2.scatter(X_resampled[mask, 0], X_resampled[mask, 1], c=color, alpha=0.6, s=30)
        
        ax2.set_title(f'{method_name}\\nDecision Boundary', fontweight='bold')
        ax2.grid(True, alpha=0.3)
        
        plt.suptitle(f'🎯 Decision Boundary Comparison: {method_name}', 
                     fontsize=14, fontweight='bold')
        plt.tight_layout()
        plt.show()

# Initialize interactive visualizer
visualizer = InteractiveSamplingVisualizer()

print("🎯 INTERACTIVE BEFORE/AFTER SAMPLING VISUALIZATION")
print("=" * 60)

# Demonstrate SMOTE before/after
print("\\n🔄 SMOTE Before/After Comparison:")
X_smote, y_smote = visualizer.create_before_after_comparison(X_demo, y_demo, 'SMOTE')
````

````python
# 📊 COMPREHENSIVE PERFORMANCE EVALUATION FRAMEWORK

class SamplingEvaluationFramework:
    """
    Comprehensive evaluation framework for sampling techniques
    """
    
    def __init__(self):
        """Initialize evaluation framework"""
        self.classifiers = {
            'logistic': LogisticRegression(random_state=42, max_iter=1000),
            'random_forest': RandomForestClassifier(n_estimators=100, random_state=42),
            'svm': SVC(random_state=42, probability=True)
        }
        
        self.metrics = {}
        self.results = {}
    
    def evaluate_sampling_method(self, X_original, y_original, X_sampled, y_sampled, 
                                method_name, test_size=0.3):
        """
        Evaluate a specific sampling method across multiple classifiers
        
        Parameters:
        -----------
        X_original : array, original features
        y_original : array, original labels  
        X_sampled : array, sampled features
        y_sampled : array, sampled labels
        method_name : str, name of sampling method
        test_size : float, test set proportion
        
        Returns:
        --------
        results : dict, evaluation results
        """
        
        # Split original data for testing (consistent test set)
        X_train_orig, X_test, y_train_orig, y_test = train_test_split(
            X_original, y_original, test_size=test_size, 
            random_state=42, stratify=y_original
        )
        
        results = {
            'method': method_name,
            'original_train_size': len(X_train_orig),
            'sampled_train_size': len(X_sampled),
            'test_size': len(X_test),
            'classifiers': {}
        }
        
        # Evaluate each classifier
        for clf_name, clf in self.classifiers.items():
            
            # Train on original data
            clf_orig = clf.__class__(**clf.get_params())
            clf_orig.fit(X_train_orig, y_train_orig)
            y_pred_orig = clf_orig.predict(X_test)
            y_prob_orig = clf_orig.predict_proba(X_test)[:, 1]
            
            # Train on sampled data
            clf_sampled = clf.__class__(**clf.get_params())
            clf_sampled.fit(X_sampled, y_sampled)
            y_pred_sampled = clf_sampled.predict(X_test)
            y_prob_sampled = clf_sampled.predict_proba(X_test)[:, 1]
            
            # Calculate metrics
            results['classifiers'][clf_name] = {
                'original': self._calculate_metrics(y_test, y_pred_orig, y_prob_orig),
                'sampled': self._calculate_metrics(y_test, y_pred_sampled, y_prob_sampled)
            }
        
        return results
    
    def _calculate_metrics(self, y_true, y_pred, y_prob):
        """Calculate comprehensive evaluation metrics"""
        
        from sklearn.metrics import (accuracy_score, precision_score, recall_score, 
                                   f1_score, roc_auc_score, average_precision_score,
                                   balanced_accuracy_score)
        
        metrics = {
            'accuracy': accuracy_score(y_true, y_pred),
            'balanced_accuracy': balanced_accuracy_score(y_true, y_pred),
            'precision': precision_score(y_true, y_pred, zero_division=0),
            'recall': recall_score(y_true, y_pred, zero_division=0),
            'f1_score': f1_score(y_true, y_pred, zero_division=0),
            'roc_auc': roc_auc_score(y_true, y_prob),
            'pr_auc': average_precision_score(y_true, y_prob)
        }
        
        return metrics
    
    def compare_all_methods(self, X, y, sampling_methods_results):
        """
        Compare all sampling methods across all classifiers
        
        Parameters:
        -----------
        X : array, original features
        y : array, original labels
        sampling_methods_results : dict, results from sampling methods
        
        Returns:
        --------
        comparison_results : dict, comprehensive comparison
        """
        
        comparison_results = {}
        
        print("🔄 EVALUATING SAMPLING METHODS...")
        print("=" * 50)
        
        for method_name, data in sampling_methods_results.items():
            if method_name == 'original':
                continue
                
            X_sampled, y_sampled = data['X'], data['y']
            
            print(f"Evaluating {method_name.upper()}...")
            
            results = self.evaluate_sampling_method(
                X, y, X_sampled, y_sampled, method_name
            )
            
            comparison_results[method_name] = results
        
        return comparison_results
    
    def create_performance_summary(self, comparison_results):
        """Create comprehensive performance summary"""
        
        # Create summary dataframe
        summary_data = []
        
        for method_name, method_results in comparison_results.items():
            for clf_name, clf_results in method_results['classifiers'].items():
                
                # Original performance
                orig_metrics = clf_results['original']
                summary_data.append({
                    'Method': f"{method_name}_original",
                    'Classifier': clf_name,
                    'Dataset': 'Original',
                    'Accuracy': orig_metrics['accuracy'],
                    'Balanced_Acc': orig_metrics['balanced_accuracy'],
                    'Precision': orig_metrics['precision'],
                    'Recall': orig_metrics['recall'],
                    'F1_Score': orig_metrics['f1_score'],
                    'ROC_AUC': orig_metrics['roc_auc'],
                    'PR_AUC': orig_metrics['pr_auc']
                })
                
                # Sampled performance
                sampled_metrics = clf_results['sampled']
                summary_data.append({
                    'Method': method_name,
                    'Classifier': clf_name,
                    'Dataset': 'Sampled',
                    'Accuracy': sampled_metrics['accuracy'],
                    'Balanced_Acc': sampled_metrics['balanced_accuracy'],
                    'Precision': sampled_metrics['precision'],
                    'Recall': sampled_metrics['recall'],
                    'F1_Score': sampled_metrics['f1_score'],
                    'ROC_AUC': sampled_metrics['roc_auc'],
                    'PR_AUC': sampled_metrics['pr_auc']
                })
        
        return pd.DataFrame(summary_data)
    
    def visualize_performance_comparison(self, comparison_results, figsize=(20, 15)):
        """Create comprehensive performance visualization"""
        
        # Create performance summary
        summary_df = self.create_performance_summary(comparison_results)
        
        # Filter to show only sampled results (not original baseline)
        sampled_df = summary_df[summary_df['Dataset'] == 'Sampled'].copy()
        
        # Create subplots for different metrics
        fig, axes = plt.subplots(3, 3, figsize=figsize)
        axes = axes.flatten()
        
        metrics = ['Accuracy', 'Balanced_Acc', 'Precision', 'Recall', 
                  'F1_Score', 'ROC_AUC', 'PR_AUC']
        
        colors = plt.cm.Set3(np.linspace(0, 1, len(sampled_df['Method'].unique())))
        
        for idx, metric in enumerate(metrics):
            if idx >= len(axes):
                break
                
            ax = axes[idx]
            
            # Create grouped bar plot
            method_names = sampled_df['Method'].unique()
            classifier_names = sampled_df['Classifier'].unique()
            
            x = np.arange(len(method_names))
            width = 0.25
            
            for i, clf_name in enumerate(classifier_names):
                clf_data = sampled_df[sampled_df['Classifier'] == clf_name]
                values = [clf_data[clf_data['Method'] == method][metric].iloc[0] 
                         if len(clf_data[clf_data['Method'] == method]) > 0 else 0
                         for method in method_names]
                
                ax.bar(x + i * width, values, width, label=clf_name, alpha=0.8)
            
            ax.set_xlabel('Sampling Methods')
            ax.set_ylabel(metric)
            ax.set_title(f'{metric} Comparison', fontweight='bold')
            ax.set_xticks(x + width)
            ax.set_xticklabels(method_names, rotation=45, ha='right')
            ax.legend()
            ax.grid(True, alpha=0.3)
        
        # Hide unused subplots
        for idx in range(len(metrics), len(axes)):
            axes[idx].set_visible(False)
        
        plt.suptitle('📊 Comprehensive Performance Comparison', 
                     fontsize=16, fontweight='bold', y=0.98)
        plt.tight_layout()
        plt.show()
        
        return summary_df
    
    def create_ranking_analysis(self, summary_df):
        """Create ranking analysis of sampling methods"""
        
        print("\n🏆 SAMPLING METHODS RANKING ANALYSIS")
        print("=" * 60)
        
        # Filter sampled results only
        sampled_df = summary_df[summary_df['Dataset'] == 'Sampled'].copy()
        
        # Calculate average performance across classifiers
        avg_performance = sampled_df.groupby('Method')[
            ['Accuracy', 'Balanced_Acc', 'Precision', 'Recall', 'F1_Score', 'ROC_AUC', 'PR_AUC']
        ].mean()
        
        # Create composite score (weighted average)
        weights = {
            'Balanced_Acc': 0.25,
            'F1_Score': 0.25,
            'ROC_AUC': 0.20,
            'PR_AUC': 0.15,
            'Precision': 0.10,
            'Recall': 0.05
        }
        
        avg_performance['Composite_Score'] = sum(
            avg_performance[metric] * weight 
            for metric, weight in weights.items()
        )
        
        # Sort by composite score
        ranking = avg_performance.sort_values('Composite_Score', ascending=False)
        
        print("OVERALL RANKING (by Composite Score):")
        print("-" * 40)
        for idx, (method, scores) in enumerate(ranking.iterrows(), 1):
            print(f"{idx}. {method.upper()}")
            print(f"   Composite Score: {scores['Composite_Score']:.4f}")
            print(f"   F1-Score: {scores['F1_Score']:.4f}")
            print(f"   ROC-AUC: {scores['ROC_AUC']:.4f}")
            print(f"   Balanced Accuracy: {scores['Balanced_Acc']:.4f}")
        
        return ranking

# Initialize evaluation framework
evaluator = SamplingEvaluationFramework()

print("📊 COMPREHENSIVE PERFORMANCE EVALUATION")
print("=" * 60)

# Combine all sampling results for evaluation
all_sampling_results = {}
all_sampling_results.update(oversampling_results)
all_sampling_results.update(undersampling_results)  
all_sampling_results.update(hybrid_results)

# Perform comprehensive evaluation
evaluation_results = evaluator.compare_all_methods(X_demo, y_demo, all_sampling_results)
````

````python
# ⚖️ HYBRID SAMPLING TECHNIQUES: Implementation and Comparison

class HybridSamplingTechniques:
    """
    Comprehensive implementation of hybrid sampling techniques
    """
    
    def __init__(self):
        """Initialize hybrid sampling methods"""
        self.methods = {
            'smote_tomek': SMOTETomek(random_state=42),
            'smote_enn': SMOTEENN(random_state=42)
        }
        
        # Add custom hybrid combinations
        self.custom_methods = {
            'adasyn_tomek': self._adasyn_tomek,
            'borderline_enn': self._borderline_enn
        }
        
        self.results = {}
    
    def _adasyn_tomek(self, X, y):
        """Custom ADASYN + Tomek Links combination"""
        # Apply ADASYN first
        adasyn = ADASYN(random_state=42)
        X_over, y_over = adasyn.fit_resample(X, y)
        
        # Apply Tomek Links
        tomek = TomekLinks()
        X_final, y_final = tomek.fit_resample(X_over, y_over)
        
        return X_final, y_final
    
    def _borderline_enn(self, X, y):
        """Custom BorderlineSMOTE + ENN combination"""
        # Apply BorderlineSMOTE first
        borderline = BorderlineSMOTE(random_state=42)
        X_over, y_over = borderline.fit_resample(X, y)
        
        # Apply ENN
        enn = EditedNearestNeighbours()
        X_final, y_final = enn.fit_resample(X_over, y_over)
        
        return X_final, y_final
    
    def apply_method(self, X, y, method_name):
        """Apply specific hybrid sampling method"""
        
        if method_name in self.methods:
            method = self.methods[method_name]
            X_resampled, y_resampled = method.fit_resample(X, y)
        elif method_name in self.custom_methods:
            method = self.custom_methods[method_name]
            X_resampled, y_resampled = method(X, y)
        else:
            raise ValueError(f"Method {method_name} not available")
        
        return X_resampled, y_resampled
    
    def compare_all_methods(self, X, y):
        """Compare all hybrid sampling methods"""
        
        results = {'original': {'X': X, 'y': y}}
        
        # Standard hybrid methods
        for name, method in self.methods.items():
            try:
                X_resampled, y_resampled = method.fit_resample(X, y)
                results[name] = {'X': X_resampled, 'y': y_resampled}
                print(f"✅ {name.upper()}: {len(X)} → {len(X_resampled)} samples")
            except Exception as e:
                print(f"❌ {name.upper()}: Error - {str(e)}")
        
        # Custom hybrid methods
        for name, method in self.custom_methods.items():
            try:
                X_resampled, y_resampled = method(X, y)
                results[name] = {'X': X_resampled, 'y': y_resampled}
                print(f"✅ {name.upper()}: {len(X)} → {len(X_resampled)} samples")
            except Exception as e:
                print(f"❌ {name.upper()}: Error - {str(e)}")
                
        return results
    
    def visualize_comparison(self, X, y, figsize=(20, 12)):
        """Visualize all hybrid sampling methods side by side"""
        
        results = self.compare_all_methods(X, y)
        n_methods = len(results)
        
        # Create subplots
        fig, axes = plt.subplots(2, 3, figsize=figsize)
        axes = axes.flatten()
        
        colors = ['#FF6B6B', '#4ECDC4']  # Red for majority, Teal for minority
        
        for idx, (method_name, data) in enumerate(results.items()):
            if idx >= len(axes):
                break
                
            ax = axes[idx]
            X_plot, y_plot = data['X'], data['y']
            
            # Plot each class
            for class_val, color in zip([0, 1], colors):
                mask = y_plot == class_val
                class_name = 'Majority' if class_val == 0 else 'Minority'
                ax.scatter(X_plot[mask, 0], X_plot[mask, 1], 
                          c=color, alpha=0.7, s=35, 
                          label=f'{class_name} (n={np.sum(mask)})')
            
            # Calculate statistics  
            class_counts = Counter(y_plot)
            ratio = class_counts[0] / class_counts[1] if class_counts[1] > 0 else float('inf')
            
            ax.set_title(f'{method_name.upper()}\n'
                        f'Total: {len(X_plot)} | Ratio: {ratio:.1f}:1',
                        fontweight='bold', fontsize=12)
            ax.legend(fontsize=10)
            ax.grid(True, alpha=0.3)
            ax.set_xlabel('Feature 1')
            ax.set_ylabel('Feature 2')
        
        # Hide unused subplots
        for idx in range(len(results), len(axes)):
            axes[idx].set_visible(False)
        
        plt.suptitle('⚖️ Hybrid Sampling Techniques Comparison', 
                     fontsize=16, fontweight='bold', y=0.98)
        plt.tight_layout()
        plt.show()
        
        return results
    
    def detailed_analysis(self, X, y, method_name):
        """Provide detailed analysis of a specific hybrid method"""
        
        print(f"\n🔬 DETAILED ANALYSIS: {method_name.upper()}")
        print("=" * 60)
        
        # Original dataset stats
        original_counts = Counter(y)
        print(f"Original dataset:")
        print(f"  Majority: {original_counts[0]} samples")
        print(f"  Minority: {original_counts[1]} samples")
        print(f"  Ratio: {original_counts[0]/original_counts[1]:.1f}:1")
        
        # Apply method and analyze
        X_resampled, y_resampled = self.apply_method(X, y, method_name)
        final_counts = Counter(y_resampled)
        
        print(f"\nAfter {method_name.upper()}:")
        print(f"  Majority: {final_counts[0]} samples")
        print(f"  Minority: {final_counts[1]} samples")
        print(f"  Ratio: {final_counts[0]/final_counts[1]:.1f}:1")
        print(f"  Total samples: {len(X)} → {len(X_resampled)}")
        
        # Calculate changes
        maj_change = final_counts[0] - original_counts[0]
        min_change = final_counts[1] - original_counts[1]
        
        print(f"\nChanges:")
        print(f"  Majority class: {maj_change:+d} samples ({maj_change/original_counts[0]*100:+.1f}%)")
        print(f"  Minority class: {min_change:+d} samples ({min_change/original_counts[1]*100:+.1f}%)")
        
        return X_resampled, y_resampled

# Initialize hybrid sampling techniques
hybrid_sampler = HybridSamplingTechniques()

print("\n⚖️ HYBRID SAMPLING TECHNIQUES DEMONSTRATION")
print("=" * 60)

# Apply and visualize all hybrid sampling methods
hybrid_results = hybrid_sampler.visualize_comparison(X_demo, y_demo)
````

# 🎯 Over/Under Sampling Techniques: Comprehensive Mathematical Guide

> **"In the realm of imbalanced data, the art of rebalancing transforms the impossible into the inevitable."** - Anonymous

Class imbalance is one of the most prevalent challenges in machine learning, where one or more classes have significantly fewer samples than others. This comprehensive guide explores the mathematical foundations, advanced algorithms, and practical implementations of sampling techniques to address class imbalance.

---

## 🧠 **Understanding Class Imbalance**

### **Mathematical Definition**

Class imbalance occurs when the distribution of classes is not uniform:

$$\text{Imbalance Ratio} = \frac{|C_{\text{majority}}|}{|C_{\text{minority}}|}$$

Where:
- $|C_{\text{majority}}|$ = Number of samples in majority class
- $|C_{\text{minority}}|$ = Number of samples in minority class

**Severity Levels:**
- **Mild Imbalance**: $1:2$ to $1:4$ ratio
- **Moderate Imbalance**: $1:5$ to $1:20$ ratio  
- **Severe Imbalance**: $1:20$ to $1:100$ ratio
- **Extreme Imbalance**: $>1:100$ ratio

### **Why Class Imbalance Matters**

**Mathematical Impact on Learning:**

1. **Decision Boundary Bias**: Classifiers tend to favor the majority class
   $$P(y = c_{\text{maj}} | \mathbf{x}) > P(y = c_{\text{min}} | \mathbf{x}))$$

2. **Loss Function Dominance**: Majority class dominates the loss
   $$\mathcal{L} = \frac{1}{n} \sum_{i=1}^{n} \ell(y_i, \hat{y}_i) \approx \frac{1}{n} \sum_{i \in C_{\text{maj}}} \ell(y_i, \hat{y}_i)$$

3. **Evaluation Metric Misleading**: High accuracy with poor minority class performance
   $$\text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN} \approx \frac{TN}{TN + FP}$$

---

## 🔄 **Sampling Techniques Taxonomy**

### **1. Oversampling Methods** 📈
- **Goal**: Increase minority class samples
- **Approach**: Generate synthetic or duplicate existing samples
- **Mathematical Principle**: $|D'_{\text{min}}| > |D_{\text{min}}|$

### **2. Undersampling Methods** 📉
- **Goal**: Reduce majority class samples  
- **Approach**: Remove redundant or noisy majority samples
- **Mathematical Principle**: $|D'_{\text{maj}}| < |D_{\text{maj}}|$

### **3. Hybrid Methods** ⚖️
- **Goal**: Combine over and undersampling
- **Approach**: Balance both classes simultaneously
- **Mathematical Principle**: $|D'_{\text{min}}| \uparrow$ and $|D'_{\text{maj}}| \downarrow$

---

## 📊 **Key Mathematical Concepts**

### **Distance Metrics**

**Euclidean Distance:**
$$d_E(\mathbf{x}_i, \mathbf{x}_j) = \sqrt{\sum_{k=1}^{d} (x_{ik} - x_{jk})^2}$$

**Manhattan Distance:**
$$d_M(\mathbf{x}_i, \mathbf{x}_j) = \sum_{k=1}^{d} |x_{ik} - x_{jk}|$$

**Minkowski Distance:**
$$d_p(\mathbf{x}_i, \mathbf{x}_j) = \left(\sum_{k=1}^{d} |x_{ik} - x_{jk}|^p\right)^{1/p}$$

### **Neighborhood Concepts**

**k-Nearest Neighbors:**
$$N_k(\mathbf{x}_i) = \{\mathbf{x}_{j_1}, \mathbf{x}_{j_2}, \ldots, \mathbf{x}_{j_k}\}$$

Where $d(\mathbf{x}_i, \mathbf{x}_{j_1}) \leq d(\mathbf{x}_i, \mathbf{x}_{j_2}) \leq \ldots \leq d(\mathbf{x}_i, \mathbf{x}_{j_k})$

**Local Density:**
$$\rho_i = \frac{1}{k} \sum_{j \in N_k(\mathbf{x}_i)} \frac{1}{d(\mathbf{x}_i, \mathbf{x}_j) + \epsilon}$$

# 📈 1. OVERSAMPLING TECHNIQUES: Mathematical Theory and Algorithms

Oversampling techniques address class imbalance by generating additional samples for the minority class. These methods range from simple duplication to sophisticated synthetic sample generation.

---

## 🎯 **1. SMOTE (Synthetic Minority Oversampling Technique)**

### **Mathematical Foundation**

SMOTE generates synthetic samples by interpolating between existing minority class samples and their nearest neighbors.

**Algorithm Steps:**

1. **Select minority sample**: $\mathbf{x}_i \in C_{\text{minority}}$
2. **Find k-nearest neighbors**: $N_k(\mathbf{x}_i) = \{\mathbf{x}_{j_1}, \mathbf{x}_{j_2}, \ldots, \mathbf{x}_{j_k}\}$
3. **Random selection**: Choose $\mathbf{x}_j$ randomly from $N_k(\mathbf{x}_i)$
4. **Generate synthetic sample**:

$$\mathbf{x}_{\text{new}} = \mathbf{x}_i + \lambda \cdot (\mathbf{x}_j - \mathbf{x}_i)$$

Where $\lambda \sim \text{Uniform}(0, 1)$ is a random interpolation factor.

**Vectorized Form:**
$$\mathbf{x}_{\text{synthetic}} = \mathbf{x}_i + \boldsymbol{\lambda} \odot (\mathbf{X}_{\text{neighbors}} - \mathbf{x}_i)$$

**Key Properties:**
- **Convex Combination**: New samples lie on line segments between existing samples
- **Local Density Preservation**: Respects local data distribution
- **Feature Space Continuity**: Maintains feature relationships

### **Advantages and Limitations**

**✅ Advantages:**
- Simple and effective
- Preserves minority class characteristics
- Works well with continuous features
- Computationally efficient

**❌ Limitations:**
- May generate noisy samples in overlapping regions
- Doesn't consider majority class distribution
- Can amplify noise if original data is noisy
- May not work well with categorical features

---

## 🎯 **2. ADASYN (Adaptive Synthetic Sampling)**

### **Mathematical Foundation**

ADASYN improves upon SMOTE by adaptively generating more synthetic samples in regions where minority class samples are scarce.

**Density Calculation:**
For each minority sample $\mathbf{x}_i$:

$$r_i = \frac{\Delta_i}{k}$$

Where $\Delta_i$ is the number of majority class samples among the k-nearest neighbors of $\mathbf{x}_i$.

**Normalized Density:**
$$\hat{r}_i = \frac{r_i}{\sum_{j=1}^{m} r_j}$$

**Number of synthetic samples for $\mathbf{x}_i$:**
$$g_i = \hat{r}_i \times G$$

Where $G$ is the total number of synthetic samples to generate.

**Synthetic Sample Generation:**
$$\mathbf{x}_{\text{new}} = \mathbf{x}_i + \lambda \cdot (\mathbf{x}_{\text{neighbor}} - \mathbf{x}_i)$$

Where $\mathbf{x}_{\text{neighbor}}$ is randomly chosen from minority class neighbors only.

**Key Innovation:**
- **Adaptive Distribution**: More samples generated in difficult regions
- **Density-Based**: Considers local class distribution
- **Noise Reduction**: Focuses on important boundary regions

---

## 🎯 **3. BorderlineSMOTE**

### **Mathematical Foundation**

BorderlineSMOTE focuses on minority samples near the decision boundary, as these are most critical for classification.

**Borderline Detection:**
For each minority sample $\mathbf{x}_i$, classify based on k-nearest neighbors:

1. **Safe**: $m \leq \frac{k}{2}$ (few majority neighbors)
2. **Danger**: $\frac{k}{2} < m < k$ (borderline region)
3. **Noise**: $m = k$ (surrounded by majority)

Where $m$ is the number of majority class samples among k-nearest neighbors.

**Variants:**

**BorderlineSMOTE-1**: Generate synthetic samples only for "Danger" samples
$$\mathbf{x}_{\text{new}} = \mathbf{x}_i + \lambda \cdot (\mathbf{x}_j - \mathbf{x}_i)$$
Where $\mathbf{x}_j$ is a minority class neighbor.

**BorderlineSMOTE-2**: Generate synthetic samples using both minority and majority neighbors
$$\mathbf{x}_{\text{new}} = \mathbf{x}_i + \lambda \cdot (\mathbf{x}_j - \mathbf{x}_i)$$
Where $\mathbf{x}_j$ can be from either class, but with different $\lambda$ ranges.

---

## 🎯 **4. SVMSMOTE**

### **Mathematical Foundation**

SVMSMOTE uses Support Vector Machine to identify support vectors and generates synthetic samples around them.

**Algorithm Steps:**

1. **Train SVM** on original imbalanced data
2. **Identify support vectors** from minority class
3. **Generate synthetic samples** around support vectors using SMOTE

**Support Vector Selection:**
$$\mathbf{x}_{\text{sv}} \in \{\mathbf{x}_i : \alpha_i > 0, y_i = -1\}$$

Where $\alpha_i$ are Lagrange multipliers and $y_i = -1$ represents minority class.

**Advantages:**
- Focuses on most informative samples
- Better decision boundary definition
- Reduces noise in synthetic generation

---

## 🎯 **5. Advanced Oversampling Techniques**

### **SMOTE-NC (Nominal-Continuous)**

Handles mixed-type features by treating categorical and continuous features differently.

**Distance Metric:**
$$d(\mathbf{x}_i, \mathbf{x}_j) = \sqrt{\sum_{k \in C} (x_{ik} - x_{jk})^2 + \sum_{k \in N} \delta(x_{ik}, x_{jk})}$$

Where:
- $C$ = continuous features
- $N$ = nominal features  
- $\delta(a, b) = 0$ if $a = b$, else $1$

**Synthetic Generation:**
- **Continuous features**: Standard SMOTE interpolation
- **Categorical features**: Mode of k-nearest neighbors

### **Geometric SMOTE**

Uses geometric transformations instead of linear interpolation.

**Geometric Mean:**
$$\mathbf{x}_{\text{new}} = \sqrt[k]{\prod_{i=1}^{k} \mathbf{x}_{n_i}}$$

**Harmonic Mean:**
$$\mathbf{x}_{\text{new}} = \frac{k}{\sum_{i=1}^{k} \frac{1}{\mathbf{x}_{n_i}}}$$

---

## 🎯 **6. Random Oversampling**

### **Mathematical Foundation**

The simplest oversampling technique that randomly duplicates minority class samples.

**Selection Probability:**
$$P(\mathbf{x}_i) = \frac{1}{|C_{\text{minority}}|}$$

**Duplication Process:**
$$D' = D \cup \{\mathbf{x}_{i_1}, \mathbf{x}_{i_2}, \ldots, \mathbf{x}_{i_n}\}$$

Where $\mathbf{x}_{i_j}$ are randomly selected minority samples.

**Advantages:**
- Simple implementation
- No information loss
- Fast execution

**Disadvantages:**
- Overfitting risk
- No new information
- May amplify noise

# ⚖️ 3. HYBRID SAMPLING TECHNIQUES: Combined Approaches

Hybrid sampling techniques combine the strengths of both oversampling and undersampling methods to achieve better class balance and data quality simultaneously.

---

## 🎯 **1. SMOTE + Tomek Links (SMOTETomek)**

### **Mathematical Foundation**

SMOTETomek applies SMOTE oversampling followed by Tomek Links undersampling to both increase minority samples and clean the dataset.

**Two-Stage Process:**

**Stage 1: SMOTE Oversampling**
Generate synthetic minority samples using SMOTE:
$$\mathbf{x}_{\text{new}} = \mathbf{x}_i + \lambda \cdot (\mathbf{x}_j - \mathbf{x}_i)$$

**Stage 2: Tomek Links Cleaning**
Remove samples forming Tomek Links:
$$\text{Remove } (\mathbf{x}_i, \mathbf{x}_j) \text{ if } d(\mathbf{x}_i, \mathbf{x}_j) = \min_{k \neq i} d(\mathbf{x}_i, \mathbf{x}_k) = \min_{l \neq j} d(\mathbf{x}_j, \mathbf{x}_l)$$

**Mathematical Benefits:**
- **Synthetic Generation**: SMOTE increases minority representation
- **Boundary Cleaning**: Tomek Links removes noisy borderline samples
- **Class Separation**: Improves separation between classes

**Algorithm:**
```
1. Apply SMOTE to generate synthetic minority samples
2. Compute distance matrix for all samples
3. Identify Tomek Links pairs
4. Remove majority class samples from Tomek Links
5. Return cleaned balanced dataset
```

---

## 🎯 **2. SMOTE + ENN (SMOTEENN)**

### **Mathematical Foundation**

SMOTEENN combines SMOTE oversampling with Edited Nearest Neighbors undersampling for comprehensive data cleaning.

**Two-Stage Process:**

**Stage 1: SMOTE Application**
Standard SMOTE synthetic generation.

**Stage 2: ENN Cleaning**
Remove samples whose class differs from majority of k-neighbors:
$$\text{Remove } \mathbf{x}_i \text{ if } \text{mode}(\{y_j : \mathbf{x}_j \in N_k(\mathbf{x}_i)\}) \neq y_i$$

**Key Advantages:**
- **Noise Reduction**: ENN removes misclassified samples
- **Consistency**: Ensures neighborhood consistency
- **Robustness**: More aggressive cleaning than Tomek Links

**Mathematical Properties:**
- **Local Consistency**: $\forall \mathbf{x}_i$, neighborhood vote matches sample class
- **Boundary Smoothing**: Removes samples in overlapping regions
- **Quality Improvement**: Higher quality synthetic samples remain

---

## 🎯 **3. ADASYN + Tomek Links**

### **Mathematical Foundation**

Combines adaptive synthetic sampling with Tomek Links cleaning.

**Process:**
1. **ADASYN Generation**: Adaptive density-based synthetic generation
2. **Tomek Cleaning**: Remove noisy borderline samples

**Density-Weighted Generation:**
$$g_i = \hat{r}_i \times G \text{ where } \hat{r}_i = \frac{r_i}{\sum_{j=1}^{m} r_j}$$

**Followed by Tomek Links removal for final cleaning.**

---

## 🎯 **4. BorderlineSMOTE + Undersampling**

### **Mathematical Foundation**

Focuses synthetic generation on borderline samples while cleaning majority class.

**Borderline Identification:**
Classify minority samples as Safe, Danger, or Noise based on k-NN:
- **Danger Region**: $\frac{k}{2} < m < k$ majority neighbors
- Generate synthetic samples only for "Danger" samples

**Combined with undersampling methods:**
- Random Undersampling
- NearMiss variants
- ENN or Tomek Links

---

## 🎯 **5. Advanced Hybrid Approaches**

### **ROSE (Random Over-Sampling Examples)**

Generates synthetic samples using kernel density estimation.

**Probability Density Function:**
$$f(\mathbf{x}) = \frac{1}{n} \sum_{i=1}^{n} K\left(\frac{\mathbf{x} - \mathbf{x}_i}{h}\right)$$

**Synthetic Generation:**
$$\mathbf{x}_{\text{new}} = \mathbf{x}_i + h \cdot \boldsymbol{\epsilon}$$

Where $\boldsymbol{\epsilon} \sim \mathcal{N}(0, I)$ and $h$ is bandwidth parameter.

### **SPIDER (Selective Preprocessing of Imbalanced Data)**

Three-stage approach:
1. **Amplification**: Generate synthetic minority samples
2. **Filtering**: Remove noisy samples
3. **Relabeling**: Correct mislabeled samples

**Relabeling Rule:**
Change label of $\mathbf{x}_i$ if:
$$\frac{|\{j : \mathbf{x}_j \in N_k(\mathbf{x}_i), y_j \neq y_i\}|}{k} > \theta$$

## 🎯 **Mathematical Analysis of Hybrid Effects**

$$\Delta I_{\text{net}} = \Delta I_{\text{over}} - \Delta I_{\text{under}}$$

## 🎯 **Ensemble Hybrid Methods**

**Mathematical Foundation:**
$$P(H(\mathbf{x}) = y) = \frac{1}{B} \sum_{i=1}^{B} P(h_i(\mathbf{x}) = y)$$

**Modified AdaBoost:**

2. Apply SMOTE to current weighted dataset
3. Train weak learner $h_t$
4. Update sample weights based on performance
5. Combine weak learners: $H(\mathbf{x}) = \sum_{t=1}^{T} \alpha_t h_t(\mathbf{x})$

# 📉 2. UNDERSAMPLING TECHNIQUES: Mathematical Theory and Algorithms

Undersampling techniques address class imbalance by intelligently removing samples from the majority class. These methods aim to preserve the most informative samples while reducing computational overhead.

---

## 🎯 **1. Random Undersampling**

### **Mathematical Foundation**

The simplest undersampling technique that randomly removes samples from the majority class.

**Selection Process:**
$$D'_{\text{maj}} = \{\mathbf{x}_{i_1}, \mathbf{x}_{i_2}, \ldots, \mathbf{x}_{i_m}\} \subset D_{\text{maj}}$$

Where $m < |D_{\text{maj}}|$ and samples are selected with uniform probability:
$$P(\mathbf{x}_i) = \frac{1}{|D_{\text{maj}}|}$$

**Sampling Ratio:**
$$\text{ratio} = \frac{|D'_{\text{maj}}|}{|D_{\text{maj}}|} \in (0, 1)$$

**Advantages:**
- Simple and fast
- No assumptions about data distribution
- Reduces computational cost

**Disadvantages:**
- Information loss
- May remove important boundary samples
- No consideration of sample importance

---

## 🎯 **2. Tomek Links**

### **Mathematical Foundation**

Tomek Links identify pairs of samples from different classes that are each other's nearest neighbors, indicating potential class overlap or noise.

**Definition:**
A pair $(\mathbf{x}_i, \mathbf{x}_j)$ forms a Tomek Link if:
1. $\mathbf{x}_i$ and $\mathbf{x}_j$ belong to different classes
2. $\mathbf{x}_j$ is the nearest neighbor of $\mathbf{x}_i$
3. $\mathbf{x}_i$ is the nearest neighbor of $\mathbf{x}_j$

**Mathematical Condition:**
$$d(\mathbf{x}_i, \mathbf{x}_j) < d(\mathbf{x}_i, \mathbf{x}_k) \quad \forall k \neq j$$
$$d(\mathbf{x}_j, \mathbf{x}_i) < d(\mathbf{x}_j, \mathbf{x}_l) \quad \forall l \neq i$$

**Removal Strategy:**
- Remove only the majority class sample from each Tomek Link
- Preserve minority class samples to maintain information

**Algorithm:**
1. Compute distance matrix $D_{n \times n}$
2. For each sample, find nearest neighbor
3. Identify reciprocal nearest neighbor pairs
4. Remove majority class samples from identified pairs

**Properties:**
- **Boundary Cleaning**: Removes samples near decision boundary
- **Noise Reduction**: Eliminates potentially mislabeled samples
- **Class Separation**: Improves separation between classes

---

## 🎯 **3. Edited Nearest Neighbors (ENN)**

### **Mathematical Foundation**

ENN removes majority class samples whose class differs from the majority class of their k-nearest neighbors.

**Decision Rule:**
Remove sample $\mathbf{x}_i$ if:
$$\text{mode}(\{y_j : \mathbf{x}_j \in N_k(\mathbf{x}_i)\}) \neq y_i$$

**Majority Vote:**
$$\hat{y}_i = \arg\max_{c} \sum_{j \in N_k(\mathbf{x}_i)} \mathbf{1}(y_j = c)$$

**Variants:**

**All-KNN**: Applies ENN with varying values of k
$$k \in \{1, 2, 3, \ldots, k_{\max}\}$$

**Repeated ENN**: Iteratively applies ENN until no more samples are removed

**Conservative ENN**: Only removes samples if all k neighbors have different class

**Mathematical Properties:**
- **Noise Tolerance**: Removes outliers and noisy samples
- **Boundary Refinement**: Smooths decision boundaries
- **Local Consistency**: Ensures local neighborhood consistency

---

## 🎯 **4. NearMiss Methods**

### **Mathematical Foundation**

NearMiss methods select majority class samples based on their distance to minority class samples.

### **NearMiss-1**
Selects majority samples with smallest average distance to k nearest minority samples.

**Selection Criterion:**
$$S_1(\mathbf{x}_i) = \frac{1}{k} \sum_{j=1}^{k} d(\mathbf{x}_i, \mathbf{x}_{n_j})$$

Where $\mathbf{x}_{n_j}$ are k nearest minority class samples to $\mathbf{x}_i$.

### **NearMiss-2**
Selects majority samples with smallest average distance to k farthest minority samples.

**Selection Criterion:**
$$S_2(\mathbf{x}_i) = \frac{1}{k} \sum_{j=1}^{k} d(\mathbf{x}_i, \mathbf{x}_{f_j})$$

Where $\mathbf{x}_{f_j}$ are k farthest minority class samples from $\mathbf{x}_i$.

### **NearMiss-3**
For each minority sample, selects m majority samples with smallest distances.

**Selection Process:**
1. For each minority sample $\mathbf{x}_{\text{min}}$
2. Find m closest majority samples
3. Union of all selected samples

**Mathematical Formulation:**
$$D'_{\text{maj}} = \bigcup_{i \in C_{\text{min}}} \{m \text{ closest majority samples to } \mathbf{x}_i\}$$

---

## 🎯 **5. Condensed Nearest Neighbor (CNN)**

### **Mathematical Foundation**

CNN finds a consistent subset of the training set such that every sample in the original set is correctly classified by its nearest neighbor in the subset.

**Consistency Condition:**
A subset $S \subseteq D$ is consistent if:
$$\forall \mathbf{x}_i \in D: \arg\min_{\mathbf{x}_j \in S} d(\mathbf{x}_i, \mathbf{x}_j) \text{ has same class as } \mathbf{x}_i$$

**Algorithm:**
1. Initialize $S$ with one sample from each class
2. For each sample $\mathbf{x}_i \in D \setminus S$:
   - If 1-NN classification using $S$ is wrong, add $\mathbf{x}_i$ to $S$
3. Repeat until no more samples are added

**Properties:**
- **Minimal Consistent Set**: Finds smallest consistent subset
- **Boundary Preservation**: Keeps important boundary samples
- **Classification Preservation**: Maintains classification accuracy

---

## 🎯 **6. One-Sided Selection (OSS)**

### **Mathematical Foundation**

OSS combines Tomek Links and CNN to remove both noisy and redundant samples.

**Two-Stage Process:**

**Stage 1: Noise Removal**
Apply Tomek Links to remove borderline and noisy samples.

**Stage 2: Redundancy Removal**
Apply CNN to remove redundant majority samples.

**Mathematical Framework:**
$$D'_{\text{maj}} = \text{CNN}(\text{TomekLinks}(D_{\text{maj}}))$$

**Advantages:**
- **Comprehensive Cleaning**: Addresses both noise and redundancy
- **Boundary Preservation**: Maintains important decision boundary information
- **Balanced Reduction**: Achieves good balance between information preservation and size reduction

---

## 🎯 **7. Cluster Centroids**

### **Mathematical Foundation**

Replaces majority class samples with cluster centroids, effectively reducing the majority class size while preserving overall distribution.

**Clustering Process:**
1. Apply clustering algorithm (e.g., K-means) to majority class
2. Replace all samples in each cluster with centroid
3. Number of clusters = desired number of majority samples

**K-Means Centroid:**
$$\boldsymbol{\mu}_k = \frac{1}{|C_k|} \sum_{\mathbf{x}_i \in C_k} \mathbf{x}_i$$

**Optimization Objective:**
$$\min_{\boldsymbol{\mu}_1, \ldots, \boldsymbol{\mu}_K} \sum_{k=1}^{K} \sum_{\mathbf{x}_i \in C_k} ||\mathbf{x}_i - \boldsymbol{\mu}_k||^2$$

**Advantages:**
- **Distribution Preservation**: Maintains overall data distribution
- **Interpretable**: Centroids represent cluster characteristics
- **Flexible**: Can control exact number of samples

**Disadvantages:**
- **Information Loss**: Loses individual sample characteristics
- **Clustering Dependency**: Results depend on clustering quality
- **Parameter Sensitivity**: Requires choosing number of clusters

---
