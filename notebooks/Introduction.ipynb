# 🤖 Welcome to Machine Learning: A Comprehensive Introduction

## 🎯 What is Machine Learning?

**Machine learning** is a revolutionary subset of artificial intelligence (AI) that enables computers to learn and improve their performance on tasks through experience, without being explicitly programmed for every scenario [1]. Think of it as teaching a computer to recognize patterns and make decisions, much like how humans learn from examples.

---

## 🧠 Core Concepts and Mathematical Foundations

### **The Learning Process**
Machine learning follows a systematic process that can be mathematically represented:

$$f: X \rightarrow Y$$

Where:
- $X$ represents the **input space** (features)
- $Y$ represents the **output space** (predictions/labels)  
- $f$ is the **function** our algorithm learns

### **The Learning Paradigm**
Given a dataset $D = \{(x_1, y_1), (x_2, y_2), ..., (x_n, y_n)\}$, our goal is to find a function $\hat{f}$ that:

$$\hat{f} = \arg\min_{f \in \mathcal{F}} \mathcal{L}(f, D)$$

Where $\mathcal{L}$ is a **loss function** measuring prediction errors, and $\mathcal{F}$ is our **hypothesis space**.

---

## 📚 Types of Machine Learning

### 🎯 **1. Supervised Learning**
- **Definition**: Learning with labeled examples (input-output pairs)
- **Mathematical Goal**: Minimize prediction error on new, unseen data
- **Loss Function**: $\mathcal{L} = \frac{1}{n}\sum_{i=1}^{n} \ell(y_i, \hat{f}(x_i))$

**Subcategories:**
- **🔢 Regression**: Predicting continuous values
  - *Examples*: House prices, stock values, temperature
  - *Common Loss*: Mean Squared Error (MSE) = $\frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2$

- **🏷️ Classification**: Predicting discrete categories  
  - *Examples*: Email spam detection, medical diagnosis, image recognition
  - *Common Loss*: Cross-entropy = $-\frac{1}{n}\sum_{i=1}^{n}\sum_{c=1}^{C} y_{ic} \log(\hat{p}_{ic})$

### 🔍 **2. Unsupervised Learning**
- **Definition**: Finding hidden patterns in unlabeled data
- **Mathematical Goal**: Discover structure in data distribution $P(X)$

**Subcategories:**
- **🎯 Clustering**: Grouping similar data points
  - *Goal*: Minimize within-cluster variation: $\sum_{i=1}^{k}\sum_{x \in C_i} \|x - \mu_i\|^2$
  
- **📉 Dimensionality Reduction**: Simplifying data while preserving information
  - *Goal*: Find projection that maximizes variance: $\max_w \text{Var}(Xw)$

### 🎮 **3. Reinforcement Learning**
- **Definition**: Learning through interaction with an environment
- **Mathematical Goal**: Maximize cumulative reward: $\mathbb{E}\left[\sum_{t=0}^{\infty} \gamma^t R_t\right]$

Where $\gamma$ is the discount factor and $R_t$ is the reward at time $t$.

## 🛠️ Popular Algorithms and Their Applications

### **🏠 Linear Models**
| Algorithm | Type | Mathematical Form | Best For |
|-----------|------|-------------------|----------|
| **Linear Regression** | Regression | $y = \beta_0 + \beta_1x_1 + ... + \beta_px_p$ | Simple relationships |
| **Logistic Regression** | Classification | $P(y=1) = \frac{1}{1+e^{-(\beta_0 + \beta_1x_1 + ... + \beta_px_p)}}$ | Binary classification |
| **Ridge Regression** | Regression | $\min \|y - X\beta\|^2 + \lambda\|\beta\|^2$ | Preventing overfitting |

### **🌳 Tree-Based Models**
| Algorithm | Type | Key Concept | Best For |
|-----------|------|-------------|----------|
| **Decision Trees** | Both | Recursive binary splits | Interpretable rules |
| **Random Forest** | Both | Ensemble of trees | High accuracy, robustness |
| **Gradient Boosting** | Both | Sequential error correction | Complex patterns |

### **📊 Distance-Based Models**
| Algorithm | Type | Distance Metric | Best For |
|-----------|------|----------------|----------|
| **K-Nearest Neighbors** | Both | $d(x_i, x_j) = \sqrt{\sum_{k=1}^{p}(x_{ik} - x_{jk})^2}$ | Local patterns |
| **K-Means Clustering** | Clustering | Euclidean distance to centroids | Spherical clusters |
| **Support Vector Machines** | Classification | Margin maximization | High-dimensional data |

## 📧 Classification: Email Spam Detection

**Problem**: Categorizing emails as spam or legitimate (ham) based on content features.

**Toy Example**:
```python
# Sample email features (simplified)
emails = [
    {"words": ["free", "money", "win", "click"], "spam": 1},
    {"words": ["meeting", "schedule", "work", "project"], "spam": 0},
    {"words": ["urgent", "prize", "congratulations", "winner"], "spam": 1},
    {"words": ["family", "dinner", "weekend", "love"], "spam": 0}
]

# Feature extraction: Count of spam-indicating words
spam_words = ["free", "money", "win", "urgent", "prize", "click"]

def extract_features(email):
    return sum(1 for word in email["words"] if word in spam_words)

# Simple classification rule
def classify_email(email):
    spam_score = extract_features(email)
    return "SPAM" if spam_score >= 2 else "HAM"

# Test on new email

prediction = classify_email(new_email)
print(f"Email classification: {prediction}")
```

**Mathematical Foundation**:
$$P(\text{spam}|\text{features}) = \frac{P(\text{features}|\text{spam}) \cdot P(\text{spam})}{P(\text{features})}$$

## 📊 Ordinal Regression: Movie Rating Prediction

**Problem**: Predicting ordinal ratings (1-5 stars) where order matters but distances between ratings aren't necessarily equal.

**Toy Example**:
```python
# Movie features and ratings
movies = [
    {"budget": 50, "cast_score": 8, "genre_score": 7, "rating": 4},  # Good movie
    {"budget": 200, "cast_score": 9, "genre_score": 9, "rating": 5}, # Excellent
    {"budget": 10, "cast_score": 4, "genre_score": 3, "rating": 2},  # Poor
    {"budget": 80, "cast_score": 6, "genre_score": 6, "rating": 3}   # Average
]

# Ordinal regression considers the order: 1 < 2 < 3 < 4 < 5
def predict_rating(budget, cast_score, genre_score):
    # Weighted score combining all features
    score = (budget/50) + cast_score + genre_score
    
    # Map continuous score to ordinal rating
    if score < 8: return 1
    elif score < 12: return 2
    elif score < 16: return 3
    elif score < 20: return 4
    else: return 5

# Predict new movie rating
new_rating = predict_rating(budget=100, cast_score=7, genre_score=8)
print(f"Predicted movie rating: {new_rating} stars")
```

**Mathematical Foundation**:
$$P(Y \leq j) = F(\theta_j - \mathbf{x}^T\boldsymbol{\beta})$$
where $F$ is the cumulative distribution function and $\theta_1 < \theta_2 < ... < \theta_{k-1}$ are threshold parameters.

## 🎯 Clustering: Customer Segmentation

**Problem**: Grouping customers with similar behavior patterns without predefined labels.

**Toy Example**:
```python
# Customer data: [annual_spending, purchase_frequency]
customers = [
    {"id": 1, "spending": 5000, "frequency": 24, "segment": "High-Value"},
    {"id": 2, "spending": 1200, "frequency": 8, "segment": "Regular"},
    {"id": 3, "spending": 4800, "frequency": 20, "segment": "High-Value"},
    {"id": 4, "spending": 800, "frequency": 4, "segment": "Occasional"},
    {"id": 5, "spending": 1500, "frequency": 12, "segment": "Regular"}
]

# Simple K-means clustering (K=3)
def simple_clustering(customers):
    # Define cluster centers (centroids)
    centroids = {
        "High-Value": {"spending": 4900, "frequency": 22},
        "Regular": {"spending": 1350, "frequency": 10},
        "Occasional": {"spending": 800, "frequency": 4}
    }
    
    def distance(customer, centroid):
        return ((customer["spending"] - centroid["spending"])**2 + 
                (customer["frequency"] - centroid["frequency"])**2)**0.5
    
    # Assign each customer to nearest cluster
    for customer in customers:
        distances = {name: distance(customer, centroid) 
                    for name, centroid in centroids.items()}
        customer["predicted_segment"] = min(distances, key=distances.get)
    
    return customers

# Cluster customers
clustered = simple_clustering(customers)
for customer in clustered:
    print(f"Customer {customer['id']}: {customer['predicted_segment']} "
          f"(actual: {customer['segment']})")
```

**Mathematical Foundation**:
$$\arg\min_{C} \sum_{i=1}^{k} \sum_{x \in C_i} \|x - \mu_i\|^2$$
where $C_i$ are clusters and $\mu_i$ are cluster centroids.

````python
# 🎮 INTERACTIVE TOY EXAMPLES IMPLEMENTATION

print("🎮 Interactive Toy Examples: Core ML Tasks")
print("=" * 50)

# --- 📧 CLASSIFICATION: EMAIL SPAM DETECTION ---
print("\n📧 1. CLASSIFICATION: Email Spam Detection")
print("-" * 40)

# Sample email data
emails_data = [
    {"words": ["free", "money", "win", "click"], "label": "SPAM"},
    {"words": ["meeting", "schedule", "work", "project"], "label": "HAM"},
    {"words": ["urgent", "prize", "congratulations", "winner"], "label": "SPAM"},
    {"words": ["family", "dinner", "weekend", "love"], "label": "HAM"},
    {"words": ["discount", "offer", "limited", "time"], "label": "SPAM"},
    {"words": ["conference", "presentation", "team", "update"], "label": "HAM"}
]

# Define spam indicators
spam_words = ["free", "money", "win", "urgent", "prize", "click", "discount", "offer", "limited"]

def extract_spam_features(email_words):
    """Extract features from email for spam detection"""
    return sum(1 for word in email_words if word in spam_words)

def classify_email(email_words, threshold=2):
    """Simple rule-based email classifier"""
    spam_score = extract_spam_features(email_words)
    return "SPAM" if spam_score >= threshold else "HAM"

# Test the classifier
print("Training Data Performance:")
correct_predictions = 0
for i, email in enumerate(emails_data):
    prediction = classify_email(email["words"])
    is_correct = prediction == email["label"]
    correct_predictions += is_correct
    
    print(f"  Email {i+1}: {email['words'][:3]}... -> {prediction} "
          f"(actual: {email['label']}) {'✓' if is_correct else '✗'}")

accuracy = correct_predictions / len(emails_data)
print(f"\nClassification Accuracy: {accuracy:.1%}")

# Test on new emails
test_emails = [
    ["urgent", "free", "money", "click"],
    ["meeting", "tomorrow", "office"],
    ["win", "prize", "congratulations"]
]

print(f"\nTesting on New Emails:")
for i, email in enumerate(test_emails):
    prediction = classify_email(email)
    spam_score = extract_spam_features(email)
    print(f"  Test {i+1}: {email} -> {prediction} (spam score: {spam_score})")

# --- 🏠 REGRESSION: HOUSE PRICE PREDICTION ---
print("\n\n🏠 2. REGRESSION: House Price Prediction")
print("-" * 40)

# Sample house data
houses_data = [
    {"size": 1200, "bedrooms": 2, "location": 7, "price": 250000},
    {"size": 1800, "bedrooms": 3, "location": 8, "price": 320000},
    {"size": 2500, "bedrooms": 4, "location": 9, "price": 450000},
    {"size": 900, "bedrooms": 1, "location": 6, "price": 180000},
    {"size": 2000, "bedrooms": 3, "location": 7, "price": 360000},
    {"size": 1500, "bedrooms": 2, "location": 8, "price": 290000}
]

def predict_house_price(size, bedrooms, location_score):
    """Simple linear model for house price prediction"""
    # Coefficients based on market analysis
    size_coef = 120      # $120 per sq ft
    bedroom_coef = 15000 # $15k per bedroom
    location_coef = 8000 # $8k per location point
    base_price = 50000   # Base price
    
    return size * size_coef + bedrooms * bedroom_coef + location_score * location_coef + base_price

# Evaluate model performance
print("Training Data Performance:")
total_error = 0
for i, house in enumerate(houses_data):
    predicted = predict_house_price(house["size"], house["bedrooms"], house["location"])
    actual = house["price"]
    error = abs(predicted - actual)
    error_pct = (error / actual) * 100
    total_error += error
    
    print(f"  House {i+1}: {house['size']}sqft, {house['bedrooms']}BR, Loc:{house['location']} "
          f"-> ${predicted:,.0f} (actual: ${actual:,.0f}, error: {error_pct:.1f}%)")

avg_error = total_error / len(houses_data)
print(f"\nAverage Prediction Error: ${avg_error:,.0f}")

# Predict new house prices
test_houses = [
    {"size": 1600, "bedrooms": 3, "location": 7.5},
    {"size": 2200, "bedrooms": 4, "location": 8.5},
    {"size": 1000, "bedrooms": 2, "location": 6}
]

print(f"\nPredicting New House Prices:")
for i, house in enumerate(test_houses):
    prediction = predict_house_price(house["size"], house["bedrooms"], house["location"])
    print(f"  House {i+1}: {house['size']}sqft, {house['bedrooms']}BR, Loc:{house['location']} "
          f"-> ${prediction:,.0f}")

# --- 📊 ORDINAL REGRESSION: MOVIE RATING PREDICTION ---
print("\n\n📊 3. ORDINAL REGRESSION: Movie Rating Prediction")
print("-" * 40)

# Sample movie data
movies_data = [
    {"budget": 50, "cast": 8, "genre": 7, "rating": 4},   # Good movie
    {"budget": 200, "cast": 9, "genre": 9, "rating": 5},  # Excellent
    {"budget": 10, "cast": 4, "genre": 3, "rating": 2},   # Poor
    {"budget": 80, "cast": 6, "genre": 6, "rating": 3},   # Average
    {"budget": 120, "cast": 8, "genre": 8, "rating": 4},  # Good
    {"budget": 15, "cast": 3, "genre": 4, "rating": 2}    # Poor
]

def predict_movie_rating(budget, cast_score, genre_score):
    """Ordinal regression for movie ratings (1-5 stars)"""
    # Calculate weighted score
    score = (budget / 50) * 2 + cast_score + genre_score
    
    # Map to ordinal scale (thresholds learned from data)
    if score < 8:
        return 1
    elif score < 12:
        return 2
    elif score < 16:
        return 3
    elif score < 20:
        return 4
    else:
        return 5

# Evaluate ordinal regression
print("Training Data Performance:")
correct_predictions = 0
for i, movie in enumerate(movies_data):
    predicted = predict_movie_rating(movie["budget"], movie["cast"], movie["genre"])
    actual = movie["rating"]
    is_correct = predicted == actual
    correct_predictions += is_correct
    
    print(f"  Movie {i+1}: Budget:${movie['budget']}M, Cast:{movie['cast']}, Genre:{movie['genre']} "
          f"-> {predicted} stars (actual: {actual}) {'✓' if is_correct else '✗'}")

accuracy = correct_predictions / len(movies_data)
print(f"\nRating Prediction Accuracy: {accuracy:.1%}")

# Test on new movies
test_movies = [
    {"budget": 90, "cast": 7, "genre": 8},
    {"budget": 25, "cast": 5, "genre": 4},
    {"budget": 150, "cast": 9, "genre": 9}
]

print(f"\nPredicting New Movie Ratings:")
for i, movie in enumerate(test_movies):
    prediction = predict_movie_rating(movie["budget"], movie["cast"], movie["genre"])
    print(f"  Movie {i+1}: Budget:${movie['budget']}M, Cast:{movie['cast']}, Genre:{movie['genre']} "
          f"-> {prediction} stars")

# --- 🎯 CLUSTERING: CUSTOMER SEGMENTATION ---
print("\n\n🎯 4. CLUSTERING: Customer Segmentation")
print("-" * 40)

# Sample customer data
customers_data = [
    {"id": 1, "spending": 5000, "frequency": 24, "actual_segment": "High-Value"},
    {"id": 2, "spending": 1200, "frequency": 8, "actual_segment": "Regular"},
    {"id": 3, "spending": 4800, "frequency": 20, "actual_segment": "High-Value"},
    {"id": 4, "spending": 800, "frequency": 4, "actual_segment": "Occasional"},
    {"id": 5, "spending": 1500, "frequency": 12, "actual_segment": "Regular"},
    {"id": 6, "spending": 300, "frequency": 2, "actual_segment": "Occasional"},
    {"id": 7, "spending": 5200, "frequency": 26, "actual_segment": "High-Value"},
    {"id": 8, "spending": 1400, "frequency": 10, "actual_segment": "Regular"}
]

def euclidean_distance(point1, point2):
    """Calculate Euclidean distance between two points"""
    return ((point1["spending"] - point2["spending"])**2 + 
            (point1["frequency"] - point2["frequency"])**2)**0.5

def simple_kmeans_clustering(customers, k=3):
    """Simple K-means clustering implementation"""
    # Initialize centroids based on domain knowledge
    centroids = {
        "High-Value": {"spending": 4900, "frequency": 22},
        "Regular": {"spending": 1350, "frequency": 10},
        "Occasional": {"spending": 600, "frequency": 3}
    }
    
    # Assign each customer to nearest cluster
    for customer in customers:
        distances = {}
        for segment_name, centroid in centroids.items():
            distance = euclidean_distance(customer, centroid)
            distances[segment_name] = distance
        
        # Assign to cluster with minimum distance
        predicted_segment = min(distances, key=distances.get)
        customer["predicted_segment"] = predicted_segment
        customer["distance_to_centroid"] = distances[predicted_segment]
    
    return customers, centroids

# Perform clustering
clustered_customers, cluster_centroids = simple_kmeans_clustering(customers_data.copy())

print("Customer Clustering Results:")
correct_clusters = 0
for customer in clustered_customers:
    is_correct = customer["predicted_segment"] == customer["actual_segment"]
    correct_clusters += is_correct
    
    print(f"  Customer {customer['id']}: ${customer['spending']}/year, {customer['frequency']} purchases "
          f"-> {customer['predicted_segment']} (actual: {customer['actual_segment']}) "
          f"{'✓' if is_correct else '✗'}")

clustering_accuracy = correct_clusters / len(clustered_customers)
print(f"\nClustering Accuracy: {clustering_accuracy:.1%}")

print(f"\nCluster Centroids:")
for segment, centroid in cluster_centroids.items():
    print(f"  {segment}: ${centroid['spending']}/year, {centroid['frequency']} purchases")

# Cluster new customers
new_customers = [
    {"id": "A", "spending": 3000, "frequency": 15},
    {"id": "B", "spending": 500, "frequency": 3},
    {"id": "C", "spending": 6000, "frequency": 30}
]

print(f"\nClustering New Customers:")
for customer in new_customers:
    distances = {}
    for segment_name, centroid in cluster_centroids.items():
        distance = euclidean_distance(customer, centroid)
        distances[segment_name] = distance
    
    predicted_segment = min(distances, key=distances.get)
    print(f"  Customer {customer['id']}: ${customer['spending']}/year, {customer['frequency']} purchases "
          f"-> {predicted_segment}")

print("\n" + "="*60)
print("🎓 TOY EXAMPLES SUMMARY:")
print("• Classification: Discrete category prediction (spam/ham)")
print("• Regression: Continuous value prediction (house prices)")  
print("• Ordinal Regression: Ordered category prediction (1-5 stars)")
print("• Clustering: Pattern discovery without labels (customer segments)")
print("• Each approach has different mathematical foundations and use cases")
print("="*60)
````

Using this data, create a simple linear regression model to:

## Solution

### Step 1: Calculate the slope (m)

m = (y2 - y1) / (x2 - x1)
m = (400,000 - 200,000) / (3000 - 1000)
m = 200,000 / 2000
m = 100

### Step 2: Calculate the y-intercept (b)

Using the point (1000, 200,000):
200,000 = 100 * 1000 + b
b = 200,000 - 100,000
b = 100,000

### Step 3: Form the linear equation

Price = 100 * Size + 100,000

### Step 4: Predict the price of a 1800 sq ft house

Price = 100 * 1800 + 100,000
Price = 180,000 + 100,000
Price = $280,000

### Step 5: Determine the average price increase per square foot

The slope (m) represents the average price increase per square foot.
Average price increase per square foot = $100

## Conclusion

1. The predicted price of a 1800 sq ft house is $280,000

## Machine learning algorithms can be classified into several categories based on different criteria. Here are some common ways to classify machine learning algorithms:
### Based on Learning Style

1. **Supervised Learning**:
   - **Definition**: Algorithms are trained on labeled data, which means that each training example is paired with an output label.
   - **Examples**:
     - Classification: Logistic Regression, Support Vector Machines (SVM), k-Nearest Neighbors (k-NN), Decision Trees, Random Forest, Naive Bayes.
     - Regression: Linear Regression, Ridge Regression, Lasso Regression, Polynomial Regression.

2. **Unsupervised Learning**:
   - **Definition**: Algorithms are used to find patterns or intrinsic structures in input data without labeled responses.
   - **Examples**:
     - Clustering: k-Means, Hierarchical Clustering, DBSCAN, Gaussian Mixture Models.
     - Association: Apriori, Eclat.

3. **Semi-Supervised Learning**:
   - **Definition**: Algorithms use a small amount of labeled data and a large amount of unlabeled data for training.
   - **Examples**: Semi-Supervised Support Vector Machines, Semi-Supervised k-Means.

4. **Reinforcement Learning**:
   - **Definition**: Algorithms learn to make decisions by taking actions in an environment to maximize some notion of cumulative reward.
   - **Examples**: Q-Learning, Deep Q-Network (DQN), Policy Gradient methods, SARSA (State-Action-Reward-State-Action).

### Based on Similarity and Distance Measures

1. **Instance-Based Learning**:
   - **Definition**: Algorithms compare new problem instances with instances seen in training, which have been stored in memory.
   - **Examples**: k-Nearest Neighbors, Locally Weighted Learning.

2. **Model-Based Learning**:
   - **Definition**: Algorithms build a model of the input data and make predictions based on the model.
   - **Examples**: Linear Regression, Logistic Regression, Neural Networks.

### Based on Model Representation

1. **Linear Models**:
   - **Definition**: Algorithms that assume a linear relationship between input variables and the output.
   - **Examples**: Linear Regression, Logistic Regression, Linear Discriminant Analysis (LDA).

2. **Non-Linear Models**:
   - **Definition**: Algorithms that can model non-linear relationships between input variables and the output.
   - **Examples**: Decision Trees, Neural Networks, Support Vector Machines with non-linear kernels.

### Based on Training Approach

1. **Batch Learning**:
   - **Definition**: Algorithms that are trained using the entire dataset at once.
   - **Examples**: Linear Regression, Neural Networks trained with gradient descent.

2. **Online Learning**:
   - **Definition**: Algorithms that are trained incrementally by feeding them data instances sequentially.
   - **Examples**: Stochastic Gradient Descent, Perceptron.

### Based on Type of Output

1. **Classification**:
   - **Definition**: Algorithms that predict a discrete label.
   - **Examples**: Logistic Regression, Decision Trees, SVM, k-NN, Naive Bayes.

2. **Regression**:
   - **Definition**: Algorithms that predict a continuous value.
   - **Examples**: Linear Regression, Ridge Regression, Lasso Regression, Support Vector Regression (SVR).

3. **Clustering**:
   - **Definition**: Algorithms that group data points into clusters based on similarity.
   - **Examples**: k-Means, DBSCAN, Hierarchical Clustering.

4. **Dimensionality Reduction**:
   - **Definition**: Algorithms that reduce the number of random variables under consideration.
   - **Examples**: Principal Component Analysis (PCA), t-Distributed Stochastic Neighbor Embedding (t-SNE), Linear Discriminant Analysis (LDA).

### Based on Neural Network Architecture

1. **Feedforward Neural Networks**:
   - **Definition**: Neural networks where connections between the nodes do not form a cycle.
   - **Examples**: Multilayer Perceptron (MLP).

2. **Convolutional Neural Networks (CNNs)**:
   - **Definition**: Neural networks that use convolutional layers to process grid-like data such as images.
   - **Examples**: LeNet, AlexNet, VGG, ResNet.

3. **Recurrent Neural Networks (RNNs)**:
   - **Definition**: Neural networks that have connections forming directed cycles, which allow them to maintain a memory of previous inputs.
   - **Examples**: LSTM, GRU.

### Based on Ensemble Methods

1. **Bagging**:
   - **Definition**: Algorithms that create multiple versions of a predictor and use them to get an aggregated predictor.
   - **Examples**: Random Forest, Bagged Decision Trees.

2. **Boosting**:
   - **Definition**: Algorithms that combine weak learners sequentially to create a strong learner.
   - **Examples**: AdaBoost, Gradient Boosting Machines (GBM), XGBoost, LightGBM, CatBoost.

3. **Stacking**:
   - **Definition**: Algorithms that combine multiple classification or regression models via a meta-classifier or meta-regressor.
   - **Examples**: Stacked Generalization.

````python
# 🔍 2. UNSUPERVISED LEARNING: DISCOVERING HIDDEN PATTERNS

print("🔍 2. UNSUPERVISED LEARNING DEMONSTRATIONS")
print("-" * 50)

fig, axes = plt.subplots(2, 3, figsize=(18, 12))

# --- CLUSTERING ANALYSIS: COMPREHENSIVE COMPARISON ---
print("🎯 Clustering Analysis: Multiple Algorithm Comparison")

# Generate diverse clustering datasets
clustering_datasets = {
    'Blob Clusters': make_blobs(n_samples=300, centers=4, cluster_std=1.0, random_state=42),
    'Varied Densities': make_blobs(n_samples=300, centers=3, cluster_std=[0.5, 1.5, 0.8], random_state=42),
    'Non-spherical': make_moons(n_samples=300, noise=0.1, random_state=42)
}

# Clustering algorithms to compare
clustering_algorithms = {
    'K-Means': KMeans(n_clusters=3, random_state=42, n_init=10),
    'DBSCAN': DBSCAN(eps=0.3, min_samples=10),
}

print("🔍 Clustering Algorithm Comparison:")

for i, (dataset_name, (X, y_true)) in enumerate(clustering_datasets.items()):
    if i >= 2:  # Limit to 2 datasets for space
        break
        
    for j, (alg_name, algorithm) in enumerate(clustering_algorithms.items()):
        ax = axes[i, j]
        
        # Apply clustering
        if hasattr(algorithm, 'fit_predict'):
            y_pred = algorithm.fit_predict(X)
        else:
            y_pred = algorithm.fit(X).labels_
        
        # Calculate number of clusters (excluding noise for DBSCAN)
        n_clusters = len(set(y_pred)) - (1 if -1 in y_pred else 0)
        n_noise = list(y_pred).count(-1) if -1 in y_pred else 0
        
        # Plot results
        unique_labels = set(y_pred)
        colors = plt.cm.Spectral(np.linspace(0, 1, len(unique_labels)))
        
        for k, color in zip(unique_labels, colors):
            if k == -1:  # Noise points (DBSCAN)
                ax.scatter(X[y_pred == k, 0], X[y_pred == k, 1], 
                          c='black', marker='x', s=50, alpha=0.6, label='Noise')
            else:
                ax.scatter(X[y_pred == k, 0], X[y_pred == k, 1], 
                          c=[color], s=50, alpha=0.8, label=f'Cluster {k}')
        
        # Add cluster centers for K-Means
        if alg_name == 'K-Means' and hasattr(algorithm, 'cluster_centers_'):
            ax.scatter(algorithm.cluster_centers_[:, 0], algorithm.cluster_centers_[:, 1],
                      c='red', marker='x', s=200, linewidths=3, label='Centroids')
        
        ax.set_title(f'{alg_name} on {dataset_name}\nClusters: {n_clusters}' + 
                    (f', Noise: {n_noise}' if n_noise > 0 else ''), fontweight='bold')
        ax.set_xlabel('Feature 1')
        ax.set_ylabel('Feature 2')
        ax.grid(True, alpha=0.3)
        if i == 0 and j == 0:
            ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
        
        print(f"  • {alg_name} on {dataset_name}: {n_clusters} clusters found")

# --- DIMENSIONALITY REDUCTION: PCA DEMONSTRATION ---
print("\n📉 Dimensionality Reduction: PCA Analysis")

# Generate high-dimensional data
np.random.seed(42)
n_samples = 200

# Create 4D data with inherent 2D structure
angle = np.linspace(0, 4*np.pi, n_samples)
radius = np.linspace(1, 3, n_samples)

# Original 2D spiral
spiral_2d = np.column_stack([
    radius * np.cos(angle),
    radius * np.sin(angle)
])

# Embed in 4D space with correlated features
X_4d = np.column_stack([
    spiral_2d[:, 0],                                    # First principal component
    spiral_2d[:, 1],                                    # Second principal component  
    spiral_2d[:, 0] * 0.3 + np.random.normal(0, 0.1, n_samples),  # Correlated feature
    spiral_2d[:, 1] * 0.2 + np.random.normal(0, 0.15, n_samples)  # Another correlated feature
])

# Apply PCA
pca = PCA()
X_pca = pca.fit_transform(X_4d)

# Plot original 4D data (first 2 dimensions) vs PCA projection
ax = axes[0, 2]
colors = plt.cm.viridis(np.linspace(0, 1, n_samples))
scatter = ax.scatter(X_4d[:, 0], X_4d[:, 1], c=angle, cmap='viridis', alpha=0.7)
ax.set_title('Original Data (First 2 Dims)', fontweight='bold')
ax.set_xlabel('Original Feature 1')
ax.set_ylabel('Original Feature 2')
ax.grid(True, alpha=0.3)
plt.colorbar(scatter, ax=ax, label='Spiral Progress')

ax = axes[1, 2]
scatter2 = ax.scatter(X_pca[:, 0], X_pca[:, 1], c=angle, cmap='viridis', alpha=0.7)
ax.set_title('PCA Projection (2D)', fontweight='bold')
ax.set_xlabel('First Principal Component')
ax.set_ylabel('Second Principal Component')
ax.grid(True, alpha=0.3)
plt.colorbar(scatter2, ax=ax, label='Spiral Progress')

# Explained variance analysis
explained_variance_ratio = pca.explained_variance_ratio_
cumulative_variance = np.cumsum(explained_variance_ratio)

print(f"  • PCA Results:")
print(f"    - Original dimensions: 4")
print(f"    - Components explaining 95% variance: {np.argmax(cumulative_variance >= 0.95) + 1}")
print(f"    - First 2 components explain: {cumulative_variance[1]:.1%} of variance")

for i, var_ratio in enumerate(explained_variance_ratio):
    print(f"    - PC{i+1}: {var_ratio:.1%} variance")

plt.tight_layout()
plt.show()

# --- ADVANCED CLUSTERING: CUSTOMER SEGMENTATION CASE STUDY ---
print("\n🛍️ Real-World Case Study: Customer Segmentation")

# Generate realistic customer data
np.random.seed(42)
n_customers = 500

# Customer segments with realistic characteristics
# Segment 1: High-value customers (20%)
high_value = int(0.2 * n_customers)
annual_spending_hv = np.random.normal(5000, 1000, high_value)
frequency_hv = np.random.normal(24, 4, high_value)
avg_order_hv = np.random.normal(200, 50, high_value)

# Segment 2: Regular customers (50%)  
regular = int(0.5 * n_customers)
annual_spending_reg = np.random.normal(1500, 400, regular)
frequency_reg = np.random.normal(12, 3, regular)
avg_order_reg = np.random.normal(80, 20, regular)

# Segment 3: Occasional customers (30%)
occasional = n_customers - high_value - regular
annual_spending_occ = np.random.normal(400, 150, occasional)
frequency_occ = np.random.normal(4, 2, occasional)
avg_order_occ = np.random.normal(60, 15, occasional)

# Combine all segments
customer_data = pd.DataFrame({
    'Annual_Spending': np.concatenate([annual_spending_hv, annual_spending_reg, annual_spending_occ]),
    'Purchase_Frequency': np.concatenate([frequency_hv, frequency_reg, frequency_occ]),
    'Avg_Order_Value': np.concatenate([avg_order_hv, avg_order_reg, avg_order_occ]),
    'True_Segment': np.concatenate([
        ['High-Value'] * high_value,
        ['Regular'] * regular, 
        ['Occasional'] * occasional
    ])
})

# Apply K-means clustering
scaler = StandardScaler()
X_customers = scaler.fit_transform(customer_data[['Annual_Spending', 'Purchase_Frequency', 'Avg_Order_Value']])

kmeans_customers = KMeans(n_clusters=3, random_state=42, n_init=10)
customer_clusters = kmeans_customers.fit_predict(X_customers)

# Visualize customer segmentation
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# True segments
ax = axes[0]
segment_colors = {'High-Value': 'red', 'Regular': 'blue', 'Occasional': 'green'}
for segment, color in segment_colors.items():
    mask = customer_data['True_Segment'] == segment
    ax.scatter(customer_data.loc[mask, 'Annual_Spending'], 
              customer_data.loc[mask, 'Purchase_Frequency'],
              c=color, alpha=0.6, label=segment, s=50)

ax.set_xlabel('Annual Spending ($)')
ax.set_ylabel('Purchase Frequency')
ax.set_title('True Customer Segments', fontweight='bold')
ax.legend()
ax.grid(True, alpha=0.3)

# K-means clusters
ax = axes[1]  
cluster_colors = ['red', 'blue', 'green']
for cluster_id in range(3):
    mask = customer_clusters == cluster_id
    ax.scatter(customer_data.loc[mask, 'Annual_Spending'],
              customer_data.loc[mask, 'Purchase_Frequency'], 
              c=cluster_colors[cluster_id], alpha=0.6, 
              label=f'Cluster {cluster_id}', s=50)

# Add cluster centers (transform back to original scale)
centers_original = scaler.inverse_transform(kmeans_customers.cluster_centers_)
ax.scatter(centers_original[:, 0], centers_original[:, 1], 
          c='black', marker='x', s=200, linewidths=3, label='Centroids')

ax.set_xlabel('Annual Spending ($)')
ax.set_ylabel('Purchase Frequency') 
ax.set_title('K-Means Customer Clusters', fontweight='bold')
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Analyze cluster characteristics
print("\\n📊 Customer Cluster Analysis:")
customer_data['Cluster'] = customer_clusters
cluster_summary = customer_data.groupby('Cluster').agg({
    'Annual_Spending': ['mean', 'std'],
    'Purchase_Frequency': ['mean', 'std'], 
    'Avg_Order_Value': ['mean', 'std']
}).round(2)

print(cluster_summary)

print("\\n🔍 UNSUPERVISED LEARNING INSIGHTS:")
print("=" * 45)
print("🎯 Clustering Key Findings:")
print("• K-Means works well for spherical, similar-sized clusters")
print("• DBSCAN can find arbitrary shaped clusters and identify noise")
print("• Customer segmentation reveals actionable business insights")

print("\\n🎯 Dimensionality Reduction Key Findings:")  
print("• PCA reveals the intrinsic dimensionality of data")
print("• First few components often capture most variance")
print("• Useful for visualization and noise reduction")
````

````python
# 🎯 1. SUPERVISED LEARNING: COMPREHENSIVE EXAMPLES

print("🎯 1. SUPERVISED LEARNING DEMONSTRATIONS")
print("-" * 50)

# Create figure with subplots for comprehensive visualization
fig = plt.figure(figsize=(20, 15))

# --- REGRESSION EXAMPLE: HOUSE PRICE PREDICTION ---
print("🏠 Regression Example: Advanced House Price Prediction")

# Generate realistic house data with multiple features
np.random.seed(42)
n_houses = 300

# Create correlated features that affect house price
house_size = np.random.normal(2000, 600, n_houses)
bedrooms = np.clip(np.round(house_size/500 + np.random.normal(0, 0.5, n_houses)), 1, 6)
age = np.random.exponential(15, n_houses)
location_score = np.random.uniform(1, 10, n_houses)

# Price formula with realistic relationships
house_prices = (house_size * 120 +           # $120 per sq ft
                bedrooms * 15000 +           # $15k per bedroom
                -age * 2000 +                # Decrease with age
                location_score * 8000 +      # Location premium
                np.random.normal(0, 30000, n_houses))  # Random noise

# Create feature matrix
X_houses = np.column_stack([house_size, bedrooms, age, location_score])
feature_names = ['Size (sq ft)', 'Bedrooms', 'Age (years)', 'Location Score']

# Split data
X_train, X_test, y_train, y_test = train_test_split(X_houses, house_prices, test_size=0.3, random_state=42)

# Compare different regression models
regression_models = {
    'Linear Regression': LinearRegression(),
    'Ridge Regression': Ridge(alpha=100),
    'Gradient Boosting': GradientBoostingRegressor(random_state=42)
}

# Train and evaluate models
reg_results = {}
subplot_idx = 1

for name, model in regression_models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    
    r2 = r2_score(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    
    reg_results[name] = {'r2': r2, 'mse': mse, 'predictions': y_pred}
    
    # Plot actual vs predicted
    ax = plt.subplot(3, 4, subplot_idx)
    plt.scatter(y_test, y_pred, alpha=0.6, color=f'C{subplot_idx-1}')
    
    # Perfect prediction line
    min_val, max_val = min(y_test.min(), y_pred.min()), max(y_test.max(), y_pred.max())
    plt.plot([min_val, max_val], [min_val, max_val], 'r--', linewidth=2, alpha=0.8)
    
    plt.xlabel('Actual Price ($)')
    plt.ylabel('Predicted Price ($)')
    plt.title(f'{name}\nR² = {r2:.3f}', fontweight='bold')
    plt.grid(True, alpha=0.3)
    
    subplot_idx += 1

# Feature importance visualization (using Gradient Boosting)
ax = plt.subplot(3, 4, 4)
gb_model = regression_models['Gradient Boosting']
importances = gb_model.feature_importances_
indices = np.argsort(importances)[::-1]

plt.bar(range(len(importances)), importances[indices], alpha=0.8)
plt.xlabel('Features')
plt.ylabel('Importance')
plt.title('Feature Importance\n(Gradient Boosting)', fontweight='bold')
plt.xticks(range(len(importances)), [feature_names[i] for i in indices], rotation=45)
plt.grid(True, alpha=0.3)

print(f"📊 Best Regression Model: {max(reg_results.items(), key=lambda x: x[1]['r2'])[0]}")

# --- CLASSIFICATION EXAMPLE: MULTI-CLASS PROBLEM ---
print("\n🎯 Classification Example: Advanced Pattern Recognition")

# Create multiple datasets with different complexity
datasets = {
    'Linear Separable': make_classification(n_samples=300, n_features=2, n_redundant=0, 
                                          n_informative=2, n_clusters_per_class=1, 
                                          class_sep=2.0, random_state=42),
    'Non-linear (Moons)': make_moons(n_samples=300, noise=0.15, random_state=42),
    'Complex (Circles)': make_circles(n_samples=300, noise=0.1, factor=0.3, random_state=42)
}

# Classification algorithms to compare
classifiers = {
    'Logistic Regression': LogisticRegression(random_state=42),
    'Random Forest': RandomForestClassifier(n_estimators=100, random_state=42),
    'SVM (RBF)': SVC(kernel='rbf', random_state=42),
    'K-Nearest Neighbors': KNeighborsClassifier(n_neighbors=5)
}

# Process each dataset
for dataset_idx, (dataset_name, (X, y)) in enumerate(datasets.items()):
    print(f"\\n🔍 Analyzing {dataset_name} dataset...")
    
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
    
    best_accuracy = 0
    best_classifier = None
    
    for clf_idx, (clf_name, classifier) in enumerate(classifiers.items()):
        # Train classifier
        classifier.fit(X_train, y_train)
        y_pred = classifier.predict(X_test)
        accuracy = accuracy_score(y_test, y_pred)
        
        if accuracy > best_accuracy:
            best_accuracy = accuracy
            best_classifier = clf_name
        
        # Create decision boundary plot
        ax = plt.subplot(3, 4, 5 + dataset_idx * 4 + clf_idx)
        
        # Create mesh for decision boundary
        h = 0.02
        x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
        y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
        xx, yy = np.meshgrid(np.arange(x_min, x_max, h),
                           np.arange(y_min, y_max, h))
        
        try:
            if hasattr(classifier, 'predict_proba'):
                Z = classifier.predict_proba(np.c_[xx.ravel(), yy.ravel()])[:, 1]
            else:
                Z = classifier.decision_function(np.c_[xx.ravel(), yy.ravel()])
        except:
            Z = classifier.predict(np.c_[xx.ravel(), yy.ravel()])
        
        Z = Z.reshape(xx.shape)
        
        # Plot decision boundary
        plt.contourf(xx, yy, Z, levels=50, alpha=0.4, cmap='RdYlBu')
        
        # Plot data points
        scatter = plt.scatter(X[:, 0], X[:, 1], c=y, cmap='RdYlBu', 
                            edgecolors='black', alpha=0.8)
        
        plt.title(f'{clf_name}\\n{dataset_name}\\nAcc: {accuracy:.3f}', 
                 fontweight='bold', fontsize=10)
        plt.xlabel('Feature 1')
        plt.ylabel('Feature 2')
        
        if dataset_idx == 0 and clf_idx == 0:
            plt.legend(*scatter.legend_elements(), title="Classes")
    
    print(f"🏆 Best classifier for {dataset_name}: {best_classifier} ({best_accuracy:.3f})")

plt.tight_layout()
plt.show()

# Performance summary
print("\\n📈 SUPERVISED LEARNING INSIGHTS:")
print("=" * 40)
print("🔍 Regression Key Findings:")
print("• Complex models (Gradient Boosting) often perform better")
print("• Feature engineering significantly impacts performance")
print("• R² score shows variance explained by the model")

print("\\n🔍 Classification Key Findings:")
print("• Algorithm performance depends heavily on data characteristics")
print("• Linear models work well for linearly separable data")
print("• Non-linear models (SVM, Random Forest) handle complex patterns")
print("• Decision boundaries reveal algorithm behavior")
````
