## 🧠 **Mathematical Foundations of GPT-3**

### 🏗️ **Transformer Architecture Deep Dive**

#### 🔤 **1. Input Representation & Embeddings**

GPT transforms discrete tokens into continuous representations through **learned embeddings**:

$$\text{Input}(x_i) = \text{TokenEmb}(x_i) + \text{PositionEmb}(i)$$

Where:
- **Token Embedding**: $E \in \mathbb{R}^{V \times d}$ maps vocabulary tokens to dense vectors
- **Positional Embedding**: $P \in \mathbb{R}^{T \times d}$ encodes sequence position information
- **Vocabulary Size**: $V = 50,257$ (GPT-3's tokenizer)
- **Model Dimension**: $d = 12,288$ (for GPT-3 175B)
- **Context Length**: $T = 2,048$ tokens

#### 🔍 **Key Insight: Why Positional Embeddings?**
Unlike RNNs, transformers have **no inherent notion of sequence order**. Positional embeddings inject positional information, allowing the model to understand token relationships and dependencies.

#### 🎯 **2. Causal Self-Attention Mechanism**

GPT uses **causal (masked) self-attention** to ensure autoregressive generation:

$$\text{CausalAttention}(Q, K, V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}} + M\right)V$$

Where:
- **Queries**: $Q = XW_Q \in \mathbb{R}^{T \times d}$
- **Keys**: $K = XW_K \in \mathbb{R}^{T \times d}$  
- **Values**: $V = XW_V \in \mathbb{R}^{T \times d}$
- **Causal Mask**: $M_{ij} = \begin{cases} 0 & \text{if } i \leq j \\ -\infty & \text{if } i > j \end{cases}$

#### 🚫 **Causal Masking: The Key to Autoregression**
The causal mask ensures that when predicting token $i$, the model **cannot look ahead** to future tokens $j > i$. This preserves the **left-to-right** generation property essential for language modeling.

#### 📊 **Attention Score Computation:**
```
Attention Weights = softmax((Q @ K.T) / sqrt(d_k) + causal_mask)
Output = Attention Weights @ V
```

#### 🧠 **3. Multi-Head Attention: Parallel Information Processing**

Multi-head attention allows the model to **attend to different types of relationships simultaneously**:

$$\text{MultiHead}(Q, K, V) = \text{Concat}(\text{head}_1, \ldots, \text{head}_h)W_O$$

Where each head focuses on different aspects:
$$\text{head}_i = \text{CausalAttention}(QW_Q^{(i)}, KW_K^{(i)}, VW_V^{(i)})$$

#### 🎭 **What Different Heads Learn:**
- **Syntactic Heads**: Grammar, POS tags, dependency relations
- **Semantic Heads**: Word meanings, entity relationships
- **Positional Heads**: Relative positions, discourse structure
- **Coreference Heads**: Pronoun resolution, entity tracking

#### 📐 **Dimension Calculations:**
```
GPT-3 Configuration:
- Total Heads: h = 96
- Head Dimension: d_head = d_model / h = 12,288 / 96 = 128
- Each head processes 128-dimensional representations
```

#### ⚡ **4. Feed-Forward Network: Position-wise Processing**

The FFN processes each position independently, providing **non-linear transformations**:

$$\text{FFN}(x) = \text{GELU}(xW_1 + b_1)W_2 + b_2$$

#### 🔥 **Key Architectural Choices:**
- **Activation**: GELU (Gaussian Error Linear Unit) instead of ReLU
- **Expansion Factor**: $4 \times$ dimension expansion (768 → 3072 → 768)
- **Position-wise**: Same transformation applied to each sequence position

#### 🧮 **Dimension Flow:**
```
Input:  [batch_size, seq_len, d_model]     # e.g., [32, 1024, 768]
   ↓
W1:     [d_model, 4*d_model]             # [768, 3072]
   ↓
GELU:   [batch_size, seq_len, 4*d_model] # [32, 1024, 3072]
   ↓
W2:     [4*d_model, d_model]             # [3072, 768]
   ↓
Output: [batch_size, seq_len, d_model]    # [32, 1024, 768]
```

#### 🎯 **Why GELU over ReLU?**
GELU provides **smoother gradients** and better performance on language tasks compared to ReLU's hard cutoff.

#### 🔧 **5. Layer Normalization & Residual Connections**

GPT uses **pre-normalization** architecture for training stability:

$$\text{LayerNorm}(x) = \frac{x - \mu}{\sigma + \epsilon} \odot \gamma + \beta$$

Where:
- $\mu, \sigma$: Mean and standard deviation across the feature dimension
- $\gamma, \beta$: Learnable scale and shift parameters
- $\epsilon = 1e^{-5}$: Small constant for numerical stability

#### 🔄 **Transformer Block Architecture:**
```python
def transformer_block(x):
    # Pre-normalization pattern (GPT-3 style)
    x = x + multi_head_attention(layer_norm(x))
    x = x + feed_forward(layer_norm(x))
    return x
```

#### 📈 **Why Pre-Normalization?**
- **Training Stability**: Gradients flow more smoothly
- **Faster Convergence**: Reduces training time
- **Better Performance**: Enables training of very deep models

### 🎓 **Training Objective: Next Token Prediction**

GPT is trained to **predict the next token** in a sequence using maximum likelihood estimation:

$$\mathcal{L} = -\sum_{t=1}^{T} \log P(x_t | x_1, x_2, \ldots, x_{t-1}; \theta)$$

Where:
- $x_t$: Token at position $t$
- $\theta$: Model parameters
- $P(x_t | x_{<t})$: Conditional probability of next token

#### 🎯 **Autoregressive Generation:**
```
Input:  "The cat sat on the"
Target: "cat sat on the mat"

Model learns: P(cat|The) × P(sat|The,cat) × P(on|The,cat,sat) × ...
```

````python
# 🏗️ Complete GPT Implementation from Scratch

import torch
import torch.nn as nn
import torch.nn.functional as F
import math
import numpy as np
from dataclasses import dataclass
from typing import Optional, Tuple
import matplotlib.pyplot as plt
import seaborn as sns
from torchviz import make_dot
from IPython.display import Image, display

# Set random seeds for reproducibility
torch.manual_seed(42)
np.random.seed(42)

@dataclass
class GPTConfig:
    """Configuration class for GPT model"""
    vocab_size: int = 50257        # GPT-3 tokenizer vocabulary size
    block_size: int = 1024         # Maximum sequence length
    n_layer: int = 12              # Number of transformer blocks
    n_head: int = 12               # Number of attention heads
    n_embd: int = 768              # Embedding dimension
    dropout: float = 0.1           # Dropout probability
    bias: bool = True              # Use bias in linear layers
    
    def __post_init__(self):
        # Ensure n_embd is divisible by n_head
        assert self.n_embd % self.n_head == 0, f"n_embd ({self.n_embd}) must be divisible by n_head ({self.n_head})"

class LayerNorm(nn.Module):
    """Layer normalization with optional bias"""
    
    def __init__(self, ndim: int, bias: bool = True):
        super().__init__()
        self.weight = nn.Parameter(torch.ones(ndim))
        self.bias = nn.Parameter(torch.zeros(ndim)) if bias else None

    def forward(self, input: torch.Tensor) -> torch.Tensor:
        return F.layer_norm(input, self.weight.shape, self.weight, self.bias, eps=1e-5)

class CausalSelfAttention(nn.Module):
    """
    Multi-head causal self-attention mechanism.
    
    Implements the core attention mechanism with causal masking to ensure
    autoregressive generation (no looking into the future).
    """
    
    def __init__(self, config: GPTConfig):
        super().__init__()
        assert config.n_embd % config.n_head == 0
        
        # Key, Query, Value projections for all heads (batched)
        self.c_attn = nn.Linear(config.n_embd, 3 * config.n_embd, bias=config.bias)
        # Output projection
        self.c_proj = nn.Linear(config.n_embd, config.n_embd, bias=config.bias)
        
        # Regularization
        self.attn_dropout = nn.Dropout(config.dropout)
        self.resid_dropout = nn.Dropout(config.dropout)
        
        # Configuration
        self.n_head = config.n_head
        self.n_embd = config.n_embd
        self.dropout = config.dropout
        
        # Causal mask (lower triangular matrix)
        self.register_buffer("bias", torch.tril(torch.ones(config.block_size, config.block_size))
                            .view(1, 1, config.block_size, config.block_size))

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        B, T, C = x.size()  # batch size, sequence length, embedding dimensionality
        
        # Calculate query, key, values for all heads in batch and move head forward to be the batch dim
        q, k, v = self.c_attn(x).split(self.n_embd, dim=2)
        
        # Reshape for multi-head attention
        # (B, T, C) -> (B, T, n_head, head_dim) -> (B, n_head, T, head_dim)
        k = k.view(B, T, self.n_head, C // self.n_head).transpose(1, 2)
        q = q.view(B, T, self.n_head, C // self.n_head).transpose(1, 2)
        v = v.view(B, T, self.n_head, C // self.n_head).transpose(1, 2)

        # Causal self-attention: Self-attend (B, n_head, T, head_dim) x (B, n_head, head_dim, T) -> (B, n_head, T, T)
        att = (q @ k.transpose(-2, -1)) * (1.0 / math.sqrt(k.size(-1)))
        
        # Apply causal mask (prevent looking into future)
        att = att.masked_fill(self.bias[:,:,:T,:T] == 0, float('-inf'))
        att = F.softmax(att, dim=-1)
        att = self.attn_dropout(att)
        
        # Apply attention to values
        y = att @ v  # (B, n_head, T, T) x (B, n_head, T, head_dim) -> (B, n_head, T, head_dim)
        
        # Re-assemble all head outputs side by side
        y = y.transpose(1, 2).contiguous().view(B, T, C)

        # Output projection
        y = self.resid_dropout(self.c_proj(y))
        return y

class MLP(nn.Module):
    """
    Position-wise feed-forward network.
    
    Implements the MLP component of each transformer block with GELU activation.
    """
    
    def __init__(self, config: GPTConfig):
        super().__init__()
        self.c_fc    = nn.Linear(config.n_embd, 4 * config.n_embd, bias=config.bias)
        self.gelu    = nn.GELU()
        self.c_proj  = nn.Linear(4 * config.n_embd, config.n_embd, bias=config.bias)
        self.dropout = nn.Dropout(config.dropout)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        x = self.c_fc(x)
        x = self.gelu(x)
        x = self.c_proj(x)
        x = self.dropout(x)
        return x

class Block(nn.Module):
    """
    Transformer block: communication (attention) followed by computation (MLP).
    
    Uses pre-normalization (LayerNorm before attention/MLP) for better training dynamics.
    """
    
    def __init__(self, config: GPTConfig):
        super().__init__()
        self.ln_1 = LayerNorm(config.n_embd, bias=config.bias)
        self.attn = CausalSelfAttention(config)
        self.ln_2 = LayerNorm(config.n_embd, bias=config.bias)
        self.mlp = MLP(config)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        # Pre-normalization pattern
        x = x + self.attn(self.ln_1(x))  # Self-attention with residual connection
        x = x + self.mlp(self.ln_2(x))   # MLP with residual connection
        return x

class GPT(nn.Module):
    """
    GPT (Generative Pre-trained Transformer) model.
    
    A decoder-only transformer for autoregressive language modeling.
    """
    
    def __init__(self, config: GPTConfig):
        super().__init__()
        assert config.vocab_size is not None
        assert config.block_size is not None
        self.config = config

        self.transformer = nn.ModuleDict(dict(
            wte = nn.Embedding(config.vocab_size, config.n_embd),        # Token embeddings
            wpe = nn.Embedding(config.block_size, config.n_embd),        # Position embeddings
            drop = nn.Dropout(config.dropout),
            h = nn.ModuleList([Block(config) for _ in range(config.n_layer)]),  # Transformer blocks
            ln_f = LayerNorm(config.n_embd, bias=config.bias),           # Final layer norm
        ))
        self.lm_head = nn.Linear(config.n_embd, config.vocab_size, bias=False)

        # Weight tying: share weights between token embeddings and language model head
        # This reduces parameters and often improves performance
        self.transformer.wte.weight = self.lm_head.weight

        # Initialize weights
        self.apply(self._init_weights)
        
        # Apply special scaled initialization to projection layers
        for pn, p in self.named_parameters():
            if pn.endswith('c_proj.weight'):
                torch.nn.init.normal_(p, mean=0.0, std=0.02/math.sqrt(2 * config.n_layer))

    def get_num_params(self, non_embedding=True):
        """Return the number of parameters in the model"""
        n_params = sum(p.numel() for p in self.parameters())
        if non_embedding:
            n_params -= self.transformer.wpe.weight.numel()
        return n_params

    def _init_weights(self, module):
        """Initialize weights following GPT-2 paper"""
        if isinstance(module, nn.Linear):
            torch.nn.init.normal_(module.weight, mean=0.0, std=0.02)
            if module.bias is not None:
                torch.nn.init.zeros_(module.bias)
        elif isinstance(module, nn.Embedding):
            torch.nn.init.normal_(module.weight, mean=0.0, std=0.02)

    def forward(self, idx: torch.Tensor, targets: Optional[torch.Tensor] = None) -> Tuple[torch.Tensor, Optional[torch.Tensor]]:
        """
        Forward pass through the model.
        
        Args:
            idx: Input token indices of shape (B, T)
            targets: Target token indices of shape (B, T), optional for training
            
        Returns:
            logits: Output logits of shape (B, T, vocab_size)
            loss: Cross-entropy loss if targets provided, else None
        """
        device = idx.device
        b, t = idx.size()
        assert t <= self.config.block_size, f"Cannot forward sequence of length {t}, block size is only {self.config.block_size}"
        
        # Generate position indices
        pos = torch.arange(0, t, dtype=torch.long, device=device).unsqueeze(0)  # shape (1, t)

        # Token embeddings + positional embeddings
        tok_emb = self.transformer.wte(idx)  # token embeddings of shape (b, t, n_embd)
        pos_emb = self.transformer.wpe(pos)  # position embeddings of shape (1, t, n_embd)
        x = self.transformer.drop(tok_emb + pos_emb)
        
        # Forward through transformer blocks
        for block in self.transformer.h:
            x = block(x)
        
        # Final layer normalization
        x = self.transformer.ln_f(x)
        
        # Language modeling head
        logits = self.lm_head(x)

        # Calculate loss if targets provided
        loss = None
        if targets is not None:
            # Flatten for cross-entropy calculation
            loss = F.cross_entropy(logits.view(-1, logits.size(-1)), targets.view(-1), ignore_index=-1)

        return logits, loss

    @torch.no_grad()
    def generate(self, idx: torch.Tensor, max_new_tokens: int, temperature: float = 1.0, top_k: Optional[int] = None) -> torch.Tensor:
        """
        Generate new tokens autoregressively.
        
        Args:
            idx: Conditioning sequence of indices (B, T)
            max_new_tokens: Number of new tokens to generate
            temperature: Sampling temperature (higher = more random)
            top_k: If set, only sample from top k most likely tokens
            
        Returns:
            Generated sequence including original context
        """
        for _ in range(max_new_tokens):
            # Crop idx to last block_size tokens if needed
            idx_cond = idx if idx.size(1) <= self.config.block_size else idx[:, -self.config.block_size:]
            
            # Forward pass
            logits, _ = self(idx_cond)
            
            # Focus only on the last time step
            logits = logits[:, -1, :] / temperature  # (B, vocab_size)
            
            # Optionally crop probabilities to only the top k options
            if top_k is not None:
                v, _ = torch.topk(logits, min(top_k, logits.size(-1)))
                logits[logits < v[:, [-1]]] = -float('Inf')
            
            # Apply softmax and sample
            probs = F.softmax(logits, dim=-1)  # (B, vocab_size)
            idx_next = torch.multinomial(probs, num_samples=1)  # (B, 1)
            
            # Append to sequence
            idx = torch.cat((idx, idx_next), dim=1)  # (B, T+1)
            
        return idx

# Create model configurations for different sizes
def get_gpt_config(model_size: str = "small") -> GPTConfig:
    """Get GPT configuration for different model sizes"""
    configs = {
        "nano": GPTConfig(n_layer=4, n_head=4, n_embd=128, block_size=256),
        "micro": GPTConfig(n_layer=6, n_head=6, n_embd=192, block_size=512),
        "small": GPTConfig(n_layer=12, n_head=12, n_embd=768, block_size=1024),
        "medium": GPTConfig(n_layer=24, n_head=16, n_embd=1024, block_size=1024),
        "large": GPTConfig(n_layer=24, n_head=16, n_embd=1536, block_size=1024),
    }
    return configs.get(model_size, configs["small"])

# Initialize a smaller GPT model for demonstration
print("🏗️ Building GPT model...")
config = get_gpt_config("nano")  # Start with nano size for quick training
model = GPT(config)

print(f"📊 Model Configuration:")
print(f"   • Layers: {config.n_layer}")
print(f"   • Heads: {config.n_head}")
print(f"   • Embedding dim: {config.n_embd}")
print(f"   • Block size: {config.block_size}")
print(f"   • Parameters: {model.get_num_params():,}")
print(f"   • Parameters (non-embedding): {model.get_num_params(non_embedding=True):,}")

# Create sample input
batch_size, seq_len = 2, 32
sample_input = torch.randint(0, config.vocab_size, (batch_size, seq_len))

print(f"\n🔮 Testing forward pass...")
print(f"   • Input shape: {sample_input.shape}")

# Forward pass
with torch.no_grad():
    logits, _ = model(sample_input)
    print(f"   • Output shape: {logits.shape}")
    print(f"   • Memory usage: {torch.cuda.memory_allocated() / 1024**2:.1f} MB" if torch.cuda.is_available() else "   • Running on CPU")

print("\n✅ GPT model built successfully!")

# Visualize the model architecture
print("\n🎨 Creating model visualization...")
try:
    # Create a smaller input for visualization
    viz_input = torch.randint(0, config.vocab_size, (1, 8))
    viz_logits, _ = model(viz_input)
    
    # Create computation graph
    dot = make_dot(viz_logits, params=dict(model.named_parameters()))
    dot.format = 'png'
    dot.render("gpt_architecture", cleanup=True)
    
    print("   ✅ Architecture diagram saved as 'gpt_architecture.png'")
    
    # Display the image if possible
    try:
        display(Image("gpt_architecture.png"))
    except:
        print("   ℹ️ To view the architecture diagram, open 'gpt_architecture.png'")
        
except Exception as e:
    print(f"   ⚠️ Visualization failed: {e}")

def print_model_summary(model: GPT):
    """Print detailed model summary"""
    print("\n📋 Detailed Model Summary:")
    print("=" * 60)
    
    total_params = 0
    for name, module in model.named_modules():
        if len(list(module.parameters())) > 0:
            module_params = sum(p.numel() for p in module.parameters())
            total_params += module_params
            if hasattr(module, 'weight'):
                weight_shape = tuple(module.weight.shape)
                print(f"{name:30} {str(weight_shape):20} {module_params:>10,}")
    
    print("=" * 60)
    print(f"{'Total Parameters:':<30} {'':<20} {total_params:>10,}")
    print(f"{'Model Size (MB):':<30} {'':<20} {total_params * 4 / 1024**2:>10.1f}")

# Print detailed summary
print_model_summary(model)
````

````python
# 🚀 Training and Text Generation Pipeline

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm.auto import tqdm
import time
from typing import List, Dict, Tuple
import json

# Set style for plots
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")

class TextDataset(Dataset):
    """
    Dataset for autoregressive language modeling.
    
    Creates input-target pairs where target is input shifted by one position.
    """
    
    def __init__(self, data: torch.Tensor, block_size: int):
        self.data = data
        self.block_size = block_size
        
    def __len__(self):
        return len(self.data) - self.block_size
    
    def __getitem__(self, idx):
        # Get a chunk of data starting at idx
        chunk = self.data[idx:idx + self.block_size + 1]
        x = chunk[:-1].clone()  # Input sequence
        y = chunk[1:].clone()   # Target sequence (shifted by 1)
        return x, y

class SimpleTokenizer:
    """
    Simple character-level tokenizer for demonstration purposes.
    
    In practice, you would use more sophisticated tokenizers like BPE.
    """
    
    def __init__(self, text: str):
        # Get unique characters and create mappings
        self.chars = sorted(list(set(text)))
        self.vocab_size = len(self.chars)
        
        # Create character to integer mappings
        self.stoi = {ch: i for i, ch in enumerate(self.chars)}
        self.itos = {i: ch for i, ch in enumerate(self.chars)}
        
        print(f"📝 Tokenizer initialized:")
        print(f"   • Vocabulary size: {self.vocab_size}")
        print(f"   • Characters: {''.join(self.chars[:20])}{'...' if len(self.chars) > 20 else ''}")
    
    def encode(self, text: str) -> List[int]:
        """Convert text to list of integers"""
        return [self.stoi[ch] for ch in text]
    
    def decode(self, tokens: List[int]) -> str:
        """Convert list of integers back to text"""
        return ''.join([self.itos[i] for i in tokens])

class GPTTrainer:
    """
    Comprehensive training pipeline for GPT models.
    """
    
    def __init__(self, model: GPT, tokenizer: SimpleTokenizer, device: str = 'cpu'):
        self.model = model.to(device)
        self.tokenizer = tokenizer
        self.device = device
        
        # Training history
        self.train_losses = []
        self.val_losses = []
        self.learning_rates = []
        
    def create_datasets(self, text: str, train_split: float = 0.9) -> Tuple[DataLoader, DataLoader]:
        """Create training and validation datasets from text"""
        
        # Tokenize the entire text
        tokens = torch.tensor(self.tokenizer.encode(text), dtype=torch.long)
        print(f"📊 Dataset statistics:")
        print(f"   • Total tokens: {len(tokens):,}")
        print(f"   • Total characters: {len(text):,}")
        
        # Split into train and validation
        split_idx = int(train_split * len(tokens))
        train_data = tokens[:split_idx]
        val_data = tokens[split_idx:]
        
        # Create datasets
        train_dataset = TextDataset(train_data, self.model.config.block_size)
        val_dataset = TextDataset(val_data, self.model.config.block_size)
        
        print(f"   • Training samples: {len(train_dataset):,}")
        print(f"   • Validation samples: {len(val_dataset):,}")
        
        return train_dataset, val_dataset
    
    def train(self, train_dataset: TextDataset, val_dataset: TextDataset, 
              epochs: int = 10, batch_size: int = 32, learning_rate: float = 3e-4,
              warmup_steps: int = 100, max_lr: float = 6e-4, eval_interval: int = 100):
        """
        Train the GPT model with modern training techniques.
        """
        
        # Create data loaders
        train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True, 
                                num_workers=0, pin_memory=True)
        val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False, 
                               num_workers=0, pin_memory=True)
        
        # Setup optimizer and scheduler
        optimizer = optim.AdamW(self.model.parameters(), lr=learning_rate, 
                              weight_decay=0.1, betas=(0.9, 0.95), eps=1e-8)
        
        # Learning rate scheduler with warmup
        def lr_lambda(step):
            if step < warmup_steps:
                return step / warmup_steps
            else:
                # Cosine decay after warmup
                progress = (step - warmup_steps) / (len(train_loader) * epochs - warmup_steps)
                return 0.5 * (1 + np.cos(np.pi * progress))
        
        scheduler = optim.lr_scheduler.LambdaLR(optimizer, lr_lambda)
        
        print(f"🏋️ Starting training...")
        print(f"   • Epochs: {epochs}")
        print(f"   • Batch size: {batch_size}")
        print(f"   • Learning rate: {learning_rate}")
        print(f"   • Warmup steps: {warmup_steps}")
        print(f"   • Device: {self.device}")
        
        self.model.train()
        step = 0
        best_val_loss = float('inf')
        
        for epoch in range(epochs):
            epoch_loss = 0
            num_batches = 0
            
            # Training loop
            pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{epochs}")
            for batch_idx, (x, y) in enumerate(pbar):
                x, y = x.to(self.device), y.to(self.device)
                
                # Forward pass
                logits, loss = self.model(x, y)
                
                # Backward pass
                optimizer.zero_grad()
                loss.backward()
                
                # Gradient clipping
                torch.nn.utils.clip_grad_norm_(self.model.parameters(), 1.0)
                
                optimizer.step()
                scheduler.step()
                
                # Track metrics
                epoch_loss += loss.item()
                num_batches += 1
                step += 1
                
                # Update progress bar
                current_lr = scheduler.get_last_lr()[0]
                pbar.set_postfix({
                    'loss': f'{loss.item():.4f}',
                    'lr': f'{current_lr:.6f}'
                })
                
                # Evaluation
                if step % eval_interval == 0:
                    val_loss = self.evaluate(val_loader)
                    self.train_losses.append(loss.item())
                    self.val_losses.append(val_loss)
                    self.learning_rates.append(current_lr)
                    
                    print(f"\nStep {step:,} | Train Loss: {loss.item():.4f} | Val Loss: {val_loss:.4f} | LR: {current_lr:.6f}")
                    
                    # Save best model
                    if val_loss < best_val_loss:
                        best_val_loss = val_loss
                        print(f"💾 New best validation loss: {val_loss:.4f}")
                    
                    self.model.train()
            
            avg_epoch_loss = epoch_loss / num_batches
            print(f"\n📊 Epoch {epoch+1} completed | Average Loss: {avg_epoch_loss:.4f}")
        
        print(f"\n🎉 Training completed!")
        print(f"   • Best validation loss: {best_val_loss:.4f}")
    
    @torch.no_grad()
    def evaluate(self, val_loader: DataLoader) -> float:
        """Evaluate the model on validation set"""
        self.model.eval()
        
        total_loss = 0
        num_batches = 0
        
        for x, y in val_loader:
            x, y = x.to(self.device), y.to(self.device)
            logits, loss = self.model(x, y)
            total_loss += loss.item()
            num_batches += 1
        
        return total_loss / num_batches
    
    def plot_training_curves(self):
        """Plot training and validation curves"""
        if not self.train_losses:
            print("No training data to plot!")
            return
        
        fig, axes = plt.subplots(1, 2, figsize=(15, 5))
        
        # Loss curves
        steps = range(len(self.train_losses))
        axes[0].plot(steps, self.train_losses, label='Training Loss', alpha=0.8)
        axes[0].plot(steps, self.val_losses, label='Validation Loss', alpha=0.8)
        axes[0].set_xlabel('Evaluation Step')
        axes[0].set_ylabel('Loss')
        axes[0].set_title('Training and Validation Loss')
        axes[0].legend()
        axes[0].grid(True)
        
        # Learning rate curve
        axes[1].plot(steps, self.learning_rates, label='Learning Rate', color='green', alpha=0.8)
        axes[1].set_xlabel('Evaluation Step')
        axes[1].set_ylabel('Learning Rate')
        axes[1].set_title('Learning Rate Schedule')
        axes[1].legend()
        axes[1].grid(True)
        
        plt.tight_layout()
        plt.show()
    
    @torch.no_grad()
    def generate_text(self, prompt: str = "", max_new_tokens: int = 100, 
                     temperature: float = 1.0, top_k: int = None) -> str:
        """Generate text from the trained model"""
        self.model.eval()
        
        if prompt:
            # Encode prompt
            prompt_tokens = torch.tensor(self.tokenizer.encode(prompt), dtype=torch.long, device=self.device).unsqueeze(0)
        else:
            # Start with a random token
            prompt_tokens = torch.randint(0, self.tokenizer.vocab_size, (1, 1), device=self.device)
        
        # Generate tokens
        generated = self.model.generate(prompt_tokens, max_new_tokens, temperature, top_k)
        
        # Decode to text
        generated_text = self.tokenizer.decode(generated[0].cpu().tolist())
        
        return generated_text

# Create sample dataset from Shakespeare text
shakespeare_text = """
To be, or not to be, that is the question:
Whether 'tis nobler in the mind to suffer
The slings and arrows of outrageous fortune,
Or to take arms against a sea of troubles
And by opposing end them. To die—to sleep,
No more; and by a sleep to say we end
The heart-ache and the thousand natural shocks
That flesh is heir to: 'tis a consummation
Devoutly to be wish'd. To die, to sleep;
To sleep, perchance to dream—ay, there's the rub:
For in that sleep of death what dreams may come,
When we have shuffled off this mortal coil,
Must give us pause—there's the respect
That makes calamity of so long life.

First Citizen:
Before we proceed any further, hear me speak.

All:
Speak, speak.

First Citizen:
You are all resolved rather to die than to famish?

All:
Resolved, resolved.

First Citizen:
First, you know Caius Marcius is chief enemy to the people.

All:
We know't, we know't.

First Citizen:
Let us kill him, and we'll have corn at our own price.
Is't a verdict?

All:
No more talking on't; let it be done. Away, away!

Second Citizen:
One word, good citizens.

First Citizen:
We are accounted poor citizens, the patricians good.
What authority surfeits on would relieve us. If they
would yield us but the superfluity while it were
wholesome, we might guess they relieved us humanely;
but they think we are too dear. The leanness that
afflicts us, the object of our misery, is as an
inventory to particularize their abundance; our
sufferance is a gain to them. Let us revenge this with
our pikes ere we become rakes; for the gods know I
speak this in hunger for bread, not in thirst for revenge.
""" * 3  # Repeat to have more training data

# Create tokenizer
print("🔤 Creating tokenizer...")
tokenizer = SimpleTokenizer(shakespeare_text)

# Update model config to match tokenizer
config.vocab_size = tokenizer.vocab_size
model = GPT(config)  # Recreate model with correct vocab size

# Initialize trainer
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
trainer = GPTTrainer(model, tokenizer, device)

print(f"\n🎭 Shakespeare Dataset Created!")
print(f"   • Text length: {len(shakespeare_text):,} characters")
print(f"   • Vocabulary size: {tokenizer.vocab_size}")
print(f"   • Sample text: {shakespeare_text[:100]}...")

# Create datasets
train_dataset, val_dataset = trainer.create_datasets(shakespeare_text, train_split=0.9)

# Train the model (smaller parameters for quick demonstration)
print(f"\n🚀 Starting training...")
trainer.train(
    train_dataset=train_dataset,
    val_dataset=val_dataset,
    epochs=5,           # Few epochs for demo
    batch_size=16,      # Small batch size
    learning_rate=1e-3, # Higher LR for quick learning
    warmup_steps=50,
    eval_interval=50
)

# Plot training curves
trainer.plot_training_curves()

# Generate some text samples
print(f"\n🎨 Text Generation Examples:")
print("=" * 60)

# Example 1: Start with prompt
prompt = "To be or not to"
generated = trainer.generate_text(prompt, max_new_tokens=50, temperature=0.8, top_k=40)
print(f"\n🎭 Prompt: '{prompt}'")
print(f"Generated: {generated}")

# Example 2: Different temperature
generated_creative = trainer.generate_text(prompt, max_new_tokens=50, temperature=1.2, top_k=40)
print(f"\n🌟 More creative (temp=1.2): {generated_creative}")

# Example 3: More conservative
generated_conservative = trainer.generate_text(prompt, max_new_tokens=50, temperature=0.3, top_k=10)
print(f"\n📚 More conservative (temp=0.3): {generated_conservative}")

print(f"\n✅ GPT Training and Generation Demo Complete!")

# Analyze attention patterns (bonus visualization)
def visualize_attention(model, text, layer_idx=0, head_idx=0):
    """Visualize attention patterns for a given text"""
    model.eval()
    
    # Tokenize input
    tokens = torch.tensor(tokenizer.encode(text[:50]), dtype=torch.long, device=device).unsqueeze(0)
    
    # Forward pass with hooks to capture attention
    with torch.no_grad():
        # Get embeddings
        tok_emb = model.transformer.wte(tokens)
        pos_emb = model.transformer.wpe(torch.arange(tokens.size(1), device=device).unsqueeze(0))
        x = model.transformer.drop(tok_emb + pos_emb)
        
        # Forward through blocks until target layer
        for i, block in enumerate(model.transformer.h):
            if i == layer_idx:
                # Get attention weights from this layer
                B, T, C = x.size()
                q, k, v = block.attn.c_attn(block.ln_1(x)).split(model.config.n_embd, dim=2)
                
                # Reshape for multi-head attention
                k = k.view(B, T, model.config.n_head, C // model.config.n_head).transpose(1, 2)
                q = q.view(B, T, model.config.n_head, C // model.config.n_head).transpose(1, 2)
                
                # Calculate attention
                att = (q @ k.transpose(-2, -1)) * (1.0 / math.sqrt(k.size(-1)))
                att = att.masked_fill(block.attn.bias[:,:,:T,:T] == 0, float('-inf'))
                att = torch.softmax(att, dim=-1)
                
                # Extract specific head
                attention_matrix = att[0, head_idx, :, :].cpu().numpy()
                
                # Create visualization
                plt.figure(figsize=(10, 8))
                plt.imshow(attention_matrix, cmap='Blues', aspect='auto')
                plt.colorbar(label='Attention Weight')
                
                # Add token labels
                token_chars = [tokenizer.itos[t.item()] for t in tokens[0]]
                plt.xticks(range(len(token_chars)), token_chars, rotation=45)
                plt.yticks(range(len(token_chars)), token_chars)
                
                plt.title(f'Attention Patterns - Layer {layer_idx}, Head {head_idx}')
                plt.xlabel('Key Position')
                plt.ylabel('Query Position')
                plt.tight_layout()
                plt.show()
                break
            x = block(x)

# Visualize attention for a sample
print(f"\n🔍 Attention Visualization:")
sample_text = "To be or not to be"
visualize_attention(model, sample_text, layer_idx=1, head_idx=0)
````
