Linear regression is a statistical method used to model the relationship between a dependent variable (response variable) and one or more independent variables (predictors). The algorithm seeks to find the best-fitting line through the data points that minimizes the sum of the squared differences between the observed values and the values predicted by the line. Here's a step-by-step description of the algorithm:

### 1. Data Preparation
- **Collect Data**: Gather the data points, where each data point consists of the independent variable(s) $(X)$ and the dependent variable $(Y)$.
- **Format Data**: Organize the data into a format suitable for analysis, typically a matrix for $(X)$ and a vector for $(Y)$.

### 2. Hypothesis Representation
- **Model Assumption**: Assume the relationship between the dependent variable $(Y)$ and the independent variable(s) $(X)$ is linear.
- **Equation of the Line**: For simple linear regression (one predictor), the hypothesis $h_\theta(x)$ is represented as:
  $h_\theta(x) = \theta_0 + \theta_1 x$
  For multiple linear regression (multiple predictors), the hypothesis is:
  $h_\theta(x) = \theta_0 + \theta_1 x_1 + \theta_2 x_2 + \ldots + \theta_n x_n$

### 3. Cost Function
- **Mean Squared Error (MSE)**: The cost function $J(\theta)$ measures the accuracy of the hypothesis function by averaging the squared differences between the predicted values and the actual values:
  $J(\theta) = \frac{1}{2m} \sum_{i=1}^{m} (h_\theta(x^{(i)}) - y^{(i)})^2$
  where $m$ is the number of training examples.

### 4. Optimization Algorithm
- **Gradient Descent**: An iterative optimization algorithm used to minimize the cost function $J(\theta)$. The update rule for gradient descent is:
  $\theta_j := \theta_j - \alpha \frac{1}{m} \sum_{i=1}^{m} (h_\theta(x^{(i)}) - y^{(i)}) x_j^{(i)}$
  where $\alpha$ is the learning rate, and $\theta_j$ is the parameter being updated.

### 5. Convergence
- **Iterate Until Convergence**: Repeat the gradient descent update rule until the cost function $J(\theta)$ converges to a minimum value, or the change in $J(\theta)$ is below a predefined threshold.

### 6. Model Evaluation
- **Prediction**: Use the optimized parameters $\theta$ to predict the value of $Y$ for new input data $X$.
- **Performance Metrics**: Evaluate the model using metrics such as Mean Absolute Error (MAE), Root Mean Squared Error (RMSE), $R^2$ score, etc.

### 7. Assumptions and Diagnostics
- **Linearity**: The relationship between the independent and dependent variables is linear.
- **Independence**: Observations are independent of each other.
- **Homoscedasticity**: The residuals (errors) have constant variance.
- **Normality**: The residuals of the model are normally distributed.

### Example of Simple Linear Regression
Given a dataset with one predictor $x$ and a response variable $y$:
1. **Data**:
   $\{(x^{(1)}, y^{(1)}), (x^{(2)}, y^{(2)}), \ldots, (x^{(m)}, y^{(m)})\}$
2. **Hypothesis**:
   $h_\theta(x) = \theta_0 + \theta_1 x$
3. **Cost Function**:
   $J(\theta) = \frac{1}{2m} \sum_{i=1}^{m} (h_\theta(x^{(i)}) - y^{(i)})^2$
4. **Gradient Descent Update Rule**:
   $\theta_0 := \theta_0 - \alpha \frac{1}{m} \sum_{i=1}^{m} (h_\theta(x^{(i)}) - y^{(i)})$
   $\theta_1 := \theta_1 - \alpha \frac{1}{m} \sum_{i=1}^{m} (h_\theta(x^{(i)}) - y^{(i)}) x^{(i)}$
5. **Repeat** until convergence.

By following these steps, you can implement linear regression to find the best-fitting line for your data, allowing you to make predictions based on new input values.

```python
import numpy as np
import pandas as pd

# Load dataset
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"
data = pd.read_csv(url, delimiter=";")

# Select features (independent variables) and target (dependent variable)
X = data.iloc[:, :-1].values  # All columns except the last one (features)
y = data.iloc[:, -1].values   # The last column (target)

# Normalize features
X_mean = X.mean(axis=0)
X_std = X.std(axis=0)
X = (X - X_mean) / X_std

# Add a column of ones to X to account for the intercept term (θ_0)
X = np.hstack((np.ones((X.shape[0], 1)), X))

# Initialize theta (parameters) with zeros
theta = np.zeros(X.shape[1])

# Define the cost function J(θ)
def compute_cost(X, y, theta):
    m = len(y)
    predictions = X @ theta  # h_θ(X) = Xθ
    errors = predictions - y
    cost = (1 / (2 * m)) * np.sum(errors ** 2)  # J(θ) = (1 / 2m) Σ (h_θ(x^(i)) - y^(i))^2
    return cost

# Define the gradient descent function
def gradient_descent(X, y, theta, alpha, num_iterations):
    m = len(y)
    cost_history = np.zeros(num_iterations)

    for i in range(num_iterations):
        predictions = X @ theta  # h_θ(X) = Xθ
        errors = predictions - y
        gradient = (1 / m) * (X.T @ errors)  # ∇J(θ) = (1 / m) X^T (h_θ(X) - y)
        theta -= alpha * gradient  # θ_j := θ_j - α * ∇J(θ)
        cost_history[i] = compute_cost(X, y, theta)  # Store the cost at each iteration

    return theta, cost_history

# Set hyperparameters for gradient descent
alpha = 0.01  # Learning rate
num_iterations = 3000  # Number of iterations

# Perform gradient descent to learn theta
theta, cost_history = gradient_descent(X, y, theta, alpha, num_iterations)

# Print the final parameters
print("Theta parameters:", theta)
print("Final cost:", cost_history[-1])

# Prediction function
def predict(X, theta):
    return X @ theta  # h_θ(X) = Xθ

# Example prediction using the first data point
prediction = predict(X[0, :], theta)
print("Predicted quality:", prediction)
print("Actual quality:", y[0])
```
