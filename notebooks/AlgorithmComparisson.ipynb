# Statistical Tests for Comparing Algorithm Accuracy

## 1. ANOVA (Analysis of Variance)
**Purpose**: ANOVA tests for significant differences between the means of three or more groups.

**When to Use**:
- Comparing the performance of multiple algorithms on different datasets.
- Assumes the data is normally distributed and variances are equal (homogeneity of variances).

**How It Works**:
- **One-Way ANOVA**: Tests if there is a difference in the mean accuracy across different algorithms or datasets.
- **Two-Way ANOVA**: Tests the effect of two independent variables (e.g., algorithms and datasets) on accuracy, and also examines interaction effects.

**Steps**:
1. Calculate the mean accuracy for each group.
2. Compute the overall mean accuracy.
3. Determine the variance within each group and between groups.
4. Use the F-ratio to test the null hypothesis that all group means are equal.

**Interpretation**:
- **F-ratio**: A large F-ratio indicates that there is more variance between groups than within groups, suggesting a significant difference.
- **p-value**: If the p-value is less than the significance level (e.g., 0.05), reject the null hypothesis.

**Post-Hoc Tests** (if ANOVA is significant):
- **Tukey's HSD (Honest Significant Difference)**: Compares all possible pairs of means.
- **Bonferroni Correction**: Adjusts the significance level when multiple comparisons are made to control the family-wise error rate.

## 2. Paired t-test
**Purpose**: Compares the means of two related groups to determine if there is a significant difference between them.

**When to Use**:
- Comparing the accuracy of two algorithms on the same set of datasets (repeated measures).

**How It Works**:
1. Compute the difference between each pair of measurements (e.g., accuracy scores of two algorithms on the same dataset).
2. Calculate the mean and standard deviation of these differences.
3. Use the t-statistic to determine if the mean difference is significantly different from zero.

**Steps**:
1. Calculate the differences between paired observations.
2. Compute the mean and standard deviation of these differences.
3. Use the t-statistic formula to test the null hypothesis that the mean difference is zero.

**Interpretation**:
- **t-value**: A large t-value indicates a greater difference between the paired means.
- **p-value**: If the p-value is less than the significance level, reject the null hypothesis.

## 3. Kruskal-Wallis Test
**Purpose**: A non-parametric test used to compare the medians of three or more independent groups.

**When to Use**:
- When the data does not meet the assumptions of normality or when using ordinal data.

**How It Works**:
1. Rank all data points from all groups together.
2. Calculate the sum of ranks for each group.
3. Compute the test statistic using the ranks and group sizes.

**Steps**:
1. Rank the combined data from all groups.
2. Calculate the sum of ranks for each group.
3. Use the Kruskal-Wallis H statistic to test the null hypothesis that all group medians are equal.

**Interpretation**:
- **H-statistic**: A large H-statistic indicates significant differences between groups.
- **p-value**: If the p-value is less than the significance level, reject the null hypothesis.

## 4. Wilcoxon Signed-Rank Test
**Purpose**: A non-parametric test for comparing the medians of two related groups.

**When to Use**:
- Comparing the accuracy of two algorithms on the same set of datasets when data does not meet the assumptions of normality.

**How It Works**:
1. Calculate the differences between paired observations.
2. Rank the absolute values of the differences.
3. Sum the ranks for positive and negative differences separately.

**Steps**:
1. Find the differences between paired observations and rank the absolute values.
2. Calculate the sum of ranks for positive and negative differences.
3. Use the Wilcoxon signed-rank statistic to test the null hypothesis that the median difference is zero.

**Interpretation**:
- **W-statistic**: A large deviation from zero indicates a significant difference between the medians.
- **p-value**: If the p-value is less than the significance level, reject the null hypothesis.

## 5. Friedman Test
**Purpose**: A non-parametric test used to compare more than two related groups.

**When to Use**:
- Comparing multiple algorithms across the same set of datasets (repeated measures).

**How It Works**:
1. Rank each set of related observations.
2. Calculate the test statistic based on the sum of ranks and the number of groups.

**Steps**:
1. Rank the data for each group.
2. Compute the sum of ranks for each group.
3. Calculate the Friedman chi-square statistic to test the null hypothesis that all group medians are equal.

**Interpretation**:
- **Chi-Square Statistic**: A large value indicates significant differences between groups.
- **p-value**: If the p-value is less than the significance level, reject the null hypothesis.

## Bonferroni-Dunn Correction
**Purpose**: Adjusts the significance level to control the family-wise error rate when performing multiple comparisons.

**When to Use**:
- When performing multiple pairwise comparisons after finding a significant result with ANOVA or Friedman test.

**How It Works**:
1. Divide the desired significance level (e.g., 0.05) by the number of comparisons.
2. Use the adjusted significance level to determine if each pairwise comparison is statistically significant.

**Steps**:
1. Calculate the number of comparisons $ \frac{k(k-1)}{2} $ for $ k $ groups.
2. Adjust the significance level: $ \alpha_{adjusted} = \frac{\alpha}{\text{number of comparisons}} $.
3. Compare the p-values of the pairwise comparisons to the adjusted significance level.

**Interpretation**:
- If the p-value of a pairwise comparison is less than the adjusted significance level, the difference is considered statistically significant.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from scipy.stats import ttest_rel, kruskal, wilcoxon, friedmanchisquare, ranksums
from sklearn.model_selection import cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, AdaBoostClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.neural_network import MLPClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis, QuadraticDiscriminantAnalysis
from sklearn.gaussian_process import GaussianProcessClassifier
from sklearn.gaussian_process.kernels import RBF
from statsmodels.stats.multicomp import pairwise_tukeyhsd
from statsmodels.stats.multitest import multipletests
from sklearn.datasets import make_classification, make_circles, make_moons, make_blobs
from itertools import combinations
import warnings

# Suppress warnings for cleaner output
warnings.filterwarnings('ignore')

# Set random seed for reproducibility
np.random.seed(42)

# Configure plotting
plt.style.use('default')
sns.set_palette("husl")
plt.rcParams['figure.figsize'] = (12, 8)

def create_datasets():
    """Create diverse datasets for algorithm comparison"""
    datasets = []
    
    # Linear separable dataset
    X_linear, y_linear = make_classification(
        n_samples=1000, n_features=2, n_redundant=0, n_informative=2,
        n_clusters_per_class=1, random_state=42
    )
    datasets.append(("Linear", (X_linear, y_linear)))
    
    # Circles dataset
    X_circles, y_circles = make_circles(
        n_samples=1000, noise=0.1, factor=0.5, random_state=42
    )
    datasets.append(("Circles", (X_circles, y_circles)))
    
    # Moons dataset
    X_moons, y_moons = make_moons(
        n_samples=1000, noise=0.1, random_state=42
    )
    datasets.append(("Moons", (X_moons, y_moons)))
    
    # Blobs dataset
    X_blobs, y_blobs = make_blobs(
        n_samples=1000, centers=3, cluster_std=1.5, random_state=42
    )
    datasets.append(("Blobs", (X_blobs, y_blobs)))
    
    return datasets

# Define classifiers to compare
classifiers = {
    'Naive Bayes': GaussianNB(),
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'Random Forest': RandomForestClassifier(n_estimators=100, random_state=42),
    'Gradient Boosting': GradientBoostingClassifier(random_state=42),
    'AdaBoost': AdaBoostClassifier(random_state=42),
    'K-NN': KNeighborsClassifier(n_neighbors=5),
    'SVM': make_pipeline(StandardScaler(), SVC(random_state=42)),
    'Neural Network': make_pipeline(StandardScaler(), MLPClassifier(random_state=42, max_iter=500)),
    'Logistic Regression': make_pipeline(StandardScaler(), LogisticRegression(random_state=42, max_iter=1000)),
    'LDA': LinearDiscriminantAnalysis(),
    'QDA': QuadraticDiscriminantAnalysis(),
    'Gaussian Process': make_pipeline(StandardScaler(), GaussianProcessClassifier(RBF(), random_state=42))
}

def evaluate_classifiers(X, y, classifiers, cv=5):
    """Evaluate classifiers using cross-validation"""
    results = {}
    
    for name, clf in classifiers.items():
        try:
            scores = cross_val_score(clf, X, y, cv=cv, scoring='accuracy')
            results[name] = scores
        except Exception as e:
            print(f"Error with {name}: {e}")
            continue
    
    return results

def visualize_results(results, dataset_name):
    """Visualize algorithm performance comparison"""
    # Create DataFrame for easier plotting
    data_for_plot = []
    for name, scores in results.items():
        for score in scores:
            data_for_plot.append({'Algorithm': name, 'Accuracy': score})
    
    df = pd.DataFrame(data_for_plot)
    
    # Create box plot
    plt.figure(figsize=(15, 8))
    sns.boxplot(data=df, x='Algorithm', y='Accuracy')
    plt.title(f'Algorithm Performance Comparison - {dataset_name} Dataset', fontsize=16, fontweight='bold')
    plt.xlabel('Algorithm', fontsize=12)
    plt.ylabel('Accuracy', fontsize=12)
    plt.xticks(rotation=45, ha='right')
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

def create_comparison_table(all_results):
    """Create a comprehensive comparison table"""
    # Calculate mean and std for each algorithm across all datasets
    algorithm_stats = {}
    
    # Get all unique algorithms
    all_algorithms = set()
    for results in all_results.values():
        all_algorithms.update(results.keys())
    
    # Initialize stats dictionary
    for alg in all_algorithms:
        algorithm_stats[alg] = {'scores': []}
    
    # Collect all scores
    for dataset_name, results in all_results.items():
        for alg_name, scores in results.items():
            algorithm_stats[alg_name]['scores'].extend(scores)
    
    # Calculate statistics
    comparison_data = []
    for alg_name, data in algorithm_stats.items():
        if data['scores']:  # Only if we have scores
            scores = np.array(data['scores'])
            comparison_data.append({
                'Algorithm': alg_name,
                'Mean Accuracy': scores.mean(),
                'Std Accuracy': scores.std(),
                'Min Accuracy': scores.min(),
                'Max Accuracy': scores.max(),
                'Median Accuracy': np.median(scores)
            })
    
    # Create DataFrame and sort by mean accuracy
    df = pd.DataFrame(comparison_data)
    df = df.sort_values('Mean Accuracy', ascending=False)
    
    return df

def perform_statistical_tests(results):
    classifier_names = list(results.keys())
    data = [results[name] for name in classifier_names]

    print("Summary Statistics:")
    summary = pd.DataFrame({name: results[name] for name in classifier_names}).describe()
    print(summary)
    print()

    if len(set(map(tuple, data))) == 1:
        print("All classifiers performed identically on this dataset.")
        return

    print("1. ANOVA Test")
    try:
        with warnings.catch_warnings():
            warnings.filterwarnings('ignore', category=stats.ConstantInputWarning)
            f_statistic, p_value = stats.f_oneway(*data)
        print(f"F-statistic: {f_statistic}")
        print(f"p-value: {p_value}")
    except ValueError as e:
        print(f"ANOVA test could not be performed: {str(e)}")

    print("\n2. Paired t-tests")
    t_test_results = []
    for i, clf1 in enumerate(classifier_names):
        for clf2 in classifier_names[i+1:]:
            if np.array_equal(results[clf1], results[clf2]):
                print(f"{clf1} vs {clf2}: Identical performance, skipping test")
            else:
                t_statistic, p_value = ttest_rel(results[clf1], results[clf2])
                t_test_results.append((clf1, clf2, t_statistic, p_value))

    if t_test_results:
        _, p_values_corrected, _, _ = multipletests([r[3] for r in t_test_results], method='bonferroni')

        for (clf1, clf2, t_statistic, p_value), p_corrected in zip(t_test_results, p_values_corrected):
            print(f"{clf1} vs {clf2}:")
            print(f"  t-statistic: {t_statistic}")
            print(f"  p-value: {p_value}")
            print(f"  Bonferroni-corrected p-value: {p_corrected}")
            print(f"  Significant: {'Yes' if p_corrected < 0.05 else 'No'}")
            print()

    print("\n3. Kruskal-Wallis Test")
    try:
        h_statistic, kw_p_value = kruskal(*data)
        print(f"H-statistic: {h_statistic}")
        print(f"p-value: {kw_p_value}")
    except ValueError as e:
        print(f"Kruskal-Wallis test could not be performed: {str(e)}")

    print("\n4. Wilcoxon Signed-Rank Tests / Mann-Whitney U-tests")
    wilcoxon_results = []
    for i, clf1 in enumerate(classifier_names):
        for clf2 in classifier_names[i+1:]:
            if np.array_equal(results[clf1], results[clf2]):
                print(f"{clf1} vs {clf2}: Identical performance, skipping test")
                wilcoxon_results.append((clf1, clf2, None, 1.0))
            else:
                try:
                    statistic, p_value = wilcoxon(results[clf1], results[clf2], zero_method='zsplit')
                except ValueError:
                    statistic, p_value = ranksums(results[clf1], results[clf2])
                wilcoxon_results.append((clf1, clf2, statistic, p_value))

    if wilcoxon_results:
        _, p_values_corrected, _, _ = multipletests([r[3] for r in wilcoxon_results], method='bonferroni')

        for (clf1, clf2, statistic, p_value), p_corrected in zip(wilcoxon_results, p_values_corrected):
            print(f"{clf1} vs {clf2}:")
            if statistic is not None:
                print(f"  Statistic: {statistic}")
                print(f"  p-value: {p_value}")
                print(f"  Bonferroni-corrected p-value: {p_corrected}")
                print(f"  Significant: {'Yes' if p_corrected < 0.05 else 'No'}")
            else:
                print("  No significant difference (identical performance)")
            print()

    print("\n5. Friedman Test")
    try:
        friedman_statistic, fr_p_value = friedmanchisquare(*data)
        print(f"Chi-square statistic: {friedman_statistic}")
        print(f"p-value: {fr_p_value}")
    except ValueError as e:
        print(f"Friedman test could not be performed: {str(e)}")

# Main execution
all_results = {}
datasets = create_datasets()

for dataset_name, (X, y) in datasets:
    print(f"Processing dataset: {dataset_name}")
    results = evaluate_classifiers(X, y, classifiers)
    all_results[dataset_name] = results
    visualize_results(results, dataset_name)
    perform_statistical_tests(results)
    print("\n" + "="*50 + "\n")

# Create and print the algorithm vs algorithm comparison table
comparison_table = create_comparison_table(all_results)
print("\nAlgorithm vs Algorithm Comparison Table:")
print(comparison_table)
```
