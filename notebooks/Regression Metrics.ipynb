# Regression Metrics

In regression problems, the performance of a model is often evaluated using a variety of metrics. Here is a list of common regression metrics along with their formulas:

## 1. Mean Absolute Error (MAE)
$ \text{MAE} = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i| $

where $ y_i $ is the actual value, $ \hat{y}_i $ is the predicted value, and $ n $ is the number of data points.

## 2. Mean Squared Error (MSE)
$ \text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 $

## 3. Root Mean Squared Error (RMSE)
$ \text{RMSE} = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2} = \sqrt{\text{MSE}} $

## 4. Mean Squared Logarithmic Error (MSLE)
$ \text{MSLE} = \frac{1}{n} \sum_{i=1}^{n} \left( \log(1 + y_i) - \log(1 + \hat{y}_i) \right)^2 $

## 5. Root Mean Squared Logarithmic Error (RMSLE)
$ \text{RMSLE} = \sqrt{\frac{1}{n} \sum_{i=1}^{n} \left( \log(1 + y_i) - \log(1 + \hat{y}_i) \right)^2} = \sqrt{\text{MSLE}} $

## 6. Mean Absolute Percentage Error (MAPE)
$ \text{MAPE} = \frac{100\%}{n} \sum_{i=1}^{n} \left| \frac{y_i - \hat{y}_i}{y_i} \right| $

## 7. Symmetric Mean Absolute Percentage Error (SMAPE)
$ \text{SMAPE} = \frac{100\%}{n} \sum_{i=1}^{n} \frac{|y_i - \hat{y}_i|}{(|y_i| + |\hat{y}_i|)/2} $

## 8. R-squared (Coefficient of Determination)
$ R^2 = 1 - \frac{\sum_{i=1}^{n} (y_i - \hat{y}_i)^2}{\sum_{i=1}^{n} (y_i - \bar{y})^2} $

where $ \bar{y} $ is the mean of the actual values.

## 9. Adjusted R-squared
$ \bar{R}^2 = 1 - (1 - R^2) \frac{n - 1}{n - p - 1} $

where $ p $ is the number of predictors.

## 10. Explained Variance Score
$ \text{Explained Variance} = 1 - \frac{\text{Var}(y - \hat{y})}{\text{Var}(y)} $

These metrics provide different perspectives on how well a regression model fits the data and help to identify areas where the model can be improved.
