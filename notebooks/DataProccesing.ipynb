# Data Preprocessing and Cleaning: Algorithms and Mathematical Foundations

## 1. Handling Missing Data

### a) Deletion Methods
- Listwise Deletion: Remove entire rows with any missing values.
- Pairwise Deletion: Remove rows only for analyses involving variables with missing values [1].

### b) Imputation Methods
- Mean/Median/Mode Imputation: Replace missing values with the mean, median, or mode of the column.
- K-Nearest Neighbors (KNN) Imputation: Fill missing values based on similar data points [2].
- Regression Imputation: Use other variables to predict missing values.
- Multiple Imputation: Create multiple plausible imputed datasets and combine results [3].

### c) Advanced Methods
- Expectation-Maximization (EM) Algorithm: Iterative method to find maximum likelihood estimates [4].

The EM algorithm alternates between two steps:

1. Expectation (E) step: Calculate the expected value of the log-likelihood function:
   Q(θ|θ(t)) = E[log L(θ; X, Z) | X, θ(t)]

2. Maximization (M) step: Find the parameter that maximizes this expected log-likelihood:
   θ(t+1) = argmax Q(θ|θ(t))

- Multiple Imputation:
This method creates m complete datasets, analyzes each, and pools the results using Rubin's rules [3]:

For an estimated parameter Q, the overall estimate is:
Q̄ = (1/m) Σ Q̂i

The total variance is:
T = W + (1 + 1/m)B

Where W = (1/m) Σ Ŵi and B = (1/(m-1)) Σ (Q̂i - Q̄)^2

## 2. Data Transformation

### a) Logarithmic Transformation: log(x) or log(x+1) for skewed data.

### b) Box-Cox Transformation [5]:
For λ ≠ 0: y(λ) = (x^λ - 1) / λ
For λ = 0: y(λ) = ln(x)

The optimal λ is chosen to maximize the log-likelihood function:
l(λ) = -n/2 * ln[(1/n) Σ (y(λ)i - ȳ(λ))^2] + (λ-1) Σ ln(xi)

### c) Yeo-Johnson Transformation [6]:
For x ≥ 0, λ ≠ 0: ((x+1)^λ - 1) / λ
For x ≥ 0, λ = 0: ln(x + 1)
For x < 0, λ ≠ 2: -((-x+1)^(2-λ) - 1) / (2-λ)
For x < 0, λ = 2: -ln(-x + 1)

### d) Square Root Transformation: Useful for count data.
### e) Exponential Transformation: Can be used to compress the range of large values.
### f) Binning: Convert continuous variables into discrete bins.

## 3. Feature Scaling

### a) Normalization (Min-Max Scaling) [7]:
x_normalized = (x - min(x)) / (max(x) - min(x))
Scales features to a fixed range, usually [0, 1].

### b) Standardization (Z-score Normalization) [7]:
x_standardized = (x - mean(x)) / std(x)
Transforms data to have zero mean and unit variance.

### c) Robust Scaling [8]:
x_robust = (x - median(x)) / IQR(x)
Where IQR is the interquartile range: IQR = Q3 - Q1

### d) Max Abs Scaling:
x_scaled = x / max(abs(x))
Scales each feature by its maximum absolute value.

### e) Quantile Transformation [9]:
For a uniform distribution:
F(x) = rank(x) / (n + 1)

For a normal distribution:
x_transformed = Φ^(-1)(F(x))

Where Φ^(-1) is the inverse CDF of the standard normal distribution.

### f) Power Transformer:
Applies the Box-Cox or Yeo-Johnson transformation to each feature independently.

## 4. Advanced Imputation Methods

### a) K-Nearest Neighbors (KNN) Imputation [2]:
For a data point with a missing value, find the k nearest neighbors based on other features. The missing value is imputed as the average (for continuous variables) or mode (for categorical variables) of the corresponding values from these neighbors.

### b) Regression Imputation:
Build a regression model for each variable with missing values:
Y = β0 + β1X1 + β2X2 + ... + βpXp + ε

Where X1, ..., Xp are other variables, β0, ..., βp are coefficients, and ε is the error term.
