## 📊 1. Handling Missing Data: Mathematical Approaches

Missing data is ubiquitous in real-world datasets and can significantly impact analysis results. Understanding the mathematical foundations of imputation methods is crucial for making informed preprocessing decisions.

### 🎯 **Types of Missing Data Mechanisms**

#### **1. Missing Completely at Random (MCAR)**
$$P(\text{Missing} | \mathbf{X}_{obs}, \mathbf{X}_{miss}) = P(\text{Missing})$$
- Missing data is independent of both observed and unobserved values
- Example: Random equipment failure during data collection

#### **2. Missing at Random (MAR)**
$$P(\text{Missing} | \mathbf{X}_{obs}, \mathbf{X}_{miss}) = P(\text{Missing} | \mathbf{X}_{obs})$$
- Missing data depends on observed values but not on unobserved values
- Example: Survey non-response correlated with respondent's age

#### **3. Missing Not at Random (MNAR)**
$$P(\text{Missing} | \mathbf{X}_{obs}, \mathbf{X}_{miss}) = P(\text{Missing} | \mathbf{X}_{obs}, \mathbf{X}_{miss})$$
- Missing data depends on the unobserved values themselves
- Example: High-income individuals not reporting their income

---

### 🔄 **Deletion Methods**

#### **a) Listwise Deletion (Complete Case Analysis)**
Remove entire observations with any missing values:

$$\mathbf{D}_{complete} = \{(\mathbf{x}_i, y_i) : \mathbf{x}_i \text{ and } y_i \text{ fully observed}\}$$

**Mathematical Properties:**
- Unbiased estimates under MCAR assumption
- Preserves joint distribution structure
- **Loss of Power:** Reduces sample size from $n$ to $n_{complete}$

#### **b) Pairwise Deletion**
Use all available data for each analysis:

$$\text{Corr}(X_j, X_k) = \frac{\sum_{i \in S_{jk}} (x_{ij} - \bar{x}_j)(x_{ik} - \bar{x}_k)}{\sqrt{\sum_{i \in S_{jk}} (x_{ij} - \bar{x}_j)^2} \sqrt{\sum_{i \in S_{jk}} (x_{ik} - \bar{x}_k)^2}}$$

Where $S_{jk} = \{i : x_{ij} \text{ and } x_{ik} \text{ both observed}\}$

---

### 🎯 **Simple Imputation Methods**

#### **a) Mean Imputation**
$$\hat{x}_{ij} = \begin{cases} 
x_{ij} & \text{if } x_{ij} \text{ is observed} \\
\bar{x}_j & \text{if } x_{ij} \text{ is missing}
\end{cases}$$

Where the **sample mean** is:
$$\bar{x}_j = \frac{1}{n_{obs,j}} \sum_{i: x_{ij} \text{ observed}} x_{ij}$$

**Properties:**
- ✅ Preserves sample mean: $E[\hat{X}_j] = \bar{x}_j$
- ❌ Reduces variance: $\text{Var}(\hat{X}_j) < \text{Var}(X_j)$
- ❌ Distorts correlations between variables

#### **b) Median Imputation**
$$\hat{x}_{ij} = \begin{cases} 
x_{ij} & \text{if } x_{ij} \text{ is observed} \\
\text{median}(X_j) & \text{if } x_{ij} \text{ is missing}
\end{cases}$$

**Median Definition:**
$$\text{median}(X_j) = \begin{cases}
x_{(n/2)} & \text{if } n \text{ is even} \\
\frac{x_{((n+1)/2)} + x_{((n-1)/2)}}{2} & \text{if } n \text{ is odd}
\end{cases}$$

Where $x_{(k)}$ denotes the $k$-th order statistic.

#### **c) Mode Imputation (Categorical Data)**
$$\hat{x}_{ij} = \arg\max_{c} P(X_j = c | X_j \text{ observed})$$

---

### 🤖 **Advanced Imputation Methods**

#### **a) K-Nearest Neighbors (KNN) Imputation**

**Algorithm:**
1. **Distance Calculation:** For sample $i$ with missing value in feature $j$:
   $$d(i, i') = \sqrt{\sum_{j' \neq j} w_{j'} (x_{ij'} - x_{i'j'})^2}$$
   
   Where $w_{j'}$ are feature weights (typically $w_{j'} = 1$).

2. **Neighbor Selection:** Find $k$ nearest neighbors:
   $$\mathcal{N}_k(i) = \{i_1, i_2, \ldots, i_k\} \text{ such that } d(i, i_1) \leq d(i, i_2) \leq \ldots \leq d(i, i_k)$$

3. **Imputation:** 
   - **Continuous:** $\hat{x}_{ij} = \frac{1}{k} \sum_{i' \in \mathcal{N}_k(i)} x_{i'j}$
   - **Categorical:** $\hat{x}_{ij} = \arg\max_{c} \sum_{i' \in \mathcal{N}_k(i)} \mathbf{1}(x_{i'j} = c)$

**Distance Weighting (Optional):**
$$\hat{x}_{ij} = \frac{\sum_{i' \in \mathcal{N}_k(i)} w_{i'} x_{i'j}}{\sum_{i' \in \mathcal{N}_k(i)} w_{i'}}$$

Where $w_{i'} = \frac{1}{d(i, i') + \epsilon}$ for small $\epsilon > 0$.

#### **b) Regression Imputation**

Build a regression model for each variable with missing values:

$$X_j = \beta_{j0} + \sum_{k \neq j} \beta_{jk} X_k + \epsilon_j$$

**Parameter Estimation (Least Squares):**
$$\hat{\boldsymbol{\beta}}_j = (\mathbf{X}_{-j}^T \mathbf{X}_{-j})^{-1} \mathbf{X}_{-j}^T \mathbf{x}_j$$

**Imputation:**
$$\hat{x}_{ij} = \hat{\beta}_{j0} + \sum_{k \neq j} \hat{\beta}_{jk} x_{ik}$$

**Stochastic Regression Imputation:**
$$\hat{x}_{ij} = \hat{\beta}_{j0} + \sum_{k \neq j} \hat{\beta}_{jk} x_{ik} + \epsilon_{ij}$$

Where $\epsilon_{ij} \sim \mathcal{N}(0, \hat{\sigma}_j^2)$ and $\hat{\sigma}_j^2 = \frac{1}{n-p-1} \sum (\hat{\epsilon}_{ij})^2$.

---

### 🔄 **Expectation-Maximization (EM) Algorithm**

The EM algorithm iteratively estimates parameters and missing values through maximum likelihood estimation.

#### **Mathematical Framework:**

Given data $\mathbf{X} = (\mathbf{X}_{obs}, \mathbf{X}_{miss})$ with parameters $\boldsymbol{\theta}$:

**E-Step:** Calculate the expected value of the log-likelihood:
$$Q(\boldsymbol{\theta}|\boldsymbol{\theta}^{(t)}) = E_{\mathbf{X}_{miss}|\mathbf{X}_{obs}, \boldsymbol{\theta}^{(t)}}[\log L(\boldsymbol{\theta}; \mathbf{X}_{obs}, \mathbf{X}_{miss})]$$

**M-Step:** Maximize the expected log-likelihood:
$$\boldsymbol{\theta}^{(t+1)} = \arg\max_{\boldsymbol{\theta}} Q(\boldsymbol{\theta}|\boldsymbol{\theta}^{(t)})$$

#### **For Multivariate Normal Data:**

**E-Step:** Update sufficient statistics:
- $\mathbf{S}^{(t+1)}_{11} = \sum_{i=1}^{n} E[\mathbf{X}_{i,obs} \mathbf{X}_{i,obs}^T | \mathbf{X}_{i,obs}, \boldsymbol{\theta}^{(t)}]$
- $\mathbf{S}^{(t+1)}_{22} = \sum_{i=1}^{n} E[\mathbf{X}_{i,miss} \mathbf{X}_{i,miss}^T | \mathbf{X}_{i,obs}, \boldsymbol{\theta}^{(t)}]$
- $\mathbf{S}^{(t+1)}_{12} = \sum_{i=1}^{n} E[\mathbf{X}_{i,obs} \mathbf{X}_{i,miss}^T | \mathbf{X}_{i,obs}, \boldsymbol{\theta}^{(t)}]$

**M-Step:** Update parameters:
$$\boldsymbol{\mu}^{(t+1)} = \frac{1}{n} \sum_{i=1}^{n} E[\mathbf{X}_i | \mathbf{X}_{i,obs}, \boldsymbol{\theta}^{(t)}]$$

$$\boldsymbol{\Sigma}^{(t+1)} = \frac{1}{n} \sum_{i=1}^{n} E[\mathbf{X}_i \mathbf{X}_i^T | \mathbf{X}_{i,obs}, \boldsymbol{\theta}^{(t)}] - \boldsymbol{\mu}^{(t+1)} (\boldsymbol{\mu}^{(t+1)})^T$$

---

### 📈 **Multiple Imputation by Chained Equations (MICE)**

MICE creates multiple plausible imputed datasets and combines results using Rubin's rules.

#### **Algorithm Steps:**

1. **Initialize:** Set initial imputations $\mathbf{X}^{(0)}$
2. **Iterate:** For each incomplete variable $j$:
   $$\mathbf{x}_j^{(t+1)} \sim P(\mathbf{x}_j | \mathbf{x}_{-j}^{(t)}, \boldsymbol{\theta}_j^{(t)})$$
3. **Repeat** until convergence

#### **Rubin's Rules for Combining Results:**

For parameter estimate $Q$ across $m$ imputations:

**Point Estimate:**
$$\bar{Q} = \frac{1}{m} \sum_{i=1}^{m} \hat{Q}_i$$

**Within-Imputation Variance:**
$$\bar{U} = \frac{1}{m} \sum_{i=1}^{m} U_i$$

**Between-Imputation Variance:**
$$B = \frac{1}{m-1} \sum_{i=1}^{m} (\hat{Q}_i - \bar{Q})^2$$

**Total Variance:**
$$T = \bar{U} + \left(1 + \frac{1}{m}\right)B$$

**Confidence Intervals:**
$$\bar{Q} \pm t_{df} \sqrt{T}$$

Where $df = (m-1)\left(1 + \frac{\bar{U}}{(1 + 1/m)B}\right)^2$ (degrees of freedom)

## 🎯 2. Feature Scaling: Mathematical Transformations

### **a) Min-Max Normalization**
Scale features to range $[0, 1]$:
$$x_{norm} = \frac{x - \min(X)}{\max(X) - \min(X)}$$

**Generalized to range $[a, b]$:**
$$x_{norm} = a + \frac{(x - \min(X))(b - a)}{\max(X) - \min(X)}$$

### **b) Standardization (Z-Score)**
Transform to zero mean and unit variance:
$$x_{std} = \frac{x - \mu}{\sigma}$$

Where:
- $\mu = \frac{1}{n}\sum_{i=1}^{n} x_i$ (sample mean)
- $\sigma = \sqrt{\frac{1}{n-1}\sum_{i=1}^{n} (x_i - \mu)^2}$ (sample standard deviation)

### **c) Robust Scaling**
Use median and IQR to reduce outlier impact:
$$x_{robust} = \frac{x - \text{median}(X)}{\text{IQR}(X)}$$

Where $\text{IQR} = Q_3 - Q_1$ (interquartile range)

## 🔄 3. Data Transformations

### **a) Box-Cox Transformation**
$$y^{(\lambda)} = \begin{cases}
\frac{x^{\lambda} - 1}{\lambda} & \text{if } \lambda \neq 0 \\
\ln(x) & \text{if } \lambda = 0
\end{cases}$$

**Optimal $\lambda$ via Maximum Likelihood:**
$$\ell(\lambda) = -\frac{n}{2}\ln\left[\frac{1}{n}\sum_{i=1}^{n}(y_i^{(\lambda)} - \bar{y}^{(\lambda)})^2\right] + (\lambda-1)\sum_{i=1}^{n}\ln(x_i)$$

### **b) Yeo-Johnson Transformation**
Extends Box-Cox to handle negative values:

$$y^{(\lambda)} = \begin{cases}
\frac{(x+1)^{\lambda} - 1}{\lambda} & \text{if } x \geq 0, \lambda \neq 0 \\
\ln(x + 1) & \text{if } x \geq 0, \lambda = 0 \\
-\frac{((-x)+1)^{2-\lambda} - 1}{2-\lambda} & \text{if } x < 0, \lambda \neq 2 \\
-\ln((-x) + 1) & \text{if } x < 0, \lambda = 2
\end{cases}$$

## 🔍 4. Outlier Detection Methods

### **a) Z-Score Method**
$$z_i = \frac{x_i - \mu}{\sigma}$$
**Outlier if:** $|z_i| > k$ (typically $k = 2$ or $3$)

### **b) Modified Z-Score (MAD-based)**
$$M_i = \frac{0.6745(x_i - \tilde{x})}{\text{MAD}}$$

Where:
- $\tilde{x} = \text{median}(X)$
- $\text{MAD} = \text{median}(|x_i - \tilde{x}|)$

**Outlier if:** $|M_i| > 3.5$

### **c) Interquartile Range (IQR) Method**
$$\text{Lower Bound} = Q_1 - 1.5 \times \text{IQR}$$
$$\text{Upper Bound} = Q_3 + 1.5 \times \text{IQR}$$

```python
# 🎯 4. OUTLIER DETECTION: MATHEMATICAL METHODS & COMPREHENSIVE ANALYSIS
print("\n🎯 4. Outlier Detection: Mathematical Implementation & Visual Analysis")
print("-" * 80)

# Create dataset with different types of outliers
np.random.seed(42)
n_samples = 500

# Generate base data
normal_data = np.random.normal(50, 10, n_samples)
exponential_data = np.random.exponential(5, n_samples)
uniform_data = np.random.uniform(20, 80, n_samples)

# Add different types of outliers
# Type 1: Extreme outliers (far from distribution)
normal_with_outliers = normal_data.copy()
outlier_indices = np.random.choice(n_samples, 25, replace=False)
normal_with_outliers[outlier_indices] += np.random.choice([-1, 1], 25) * np.random.uniform(50, 100, 25)

# Type 2: Mild outliers
exponential_with_outliers = exponential_data.copy()
mild_outlier_indices = np.random.choice(n_samples, 20, replace=False)
exponential_with_outliers[mild_outlier_indices] += np.random.uniform(30, 50, 20)

# Type 3: Clustered outliers
uniform_with_clusters = uniform_data.copy()
cluster_indices = np.random.choice(n_samples, 15, replace=False)
uniform_with_clusters[cluster_indices] = np.random.normal(120, 5, 15)  # Outlier cluster

outlier_data = pd.DataFrame({
    'Normal_Extreme': normal_with_outliers,
    'Exponential_Mild': exponential_with_outliers,
    'Uniform_Clustered': uniform_with_clusters
})

print("📊 Dataset with Outliers Statistics:")
print(outlier_data.describe())

class MathematicalOutlierDetector:
    def __init__(self, method='zscore', threshold=3):
        self.method = method
        self.threshold = threshold
        self.params_ = {}
        
    def fit(self, X):
        """
        Calculate parameters for outlier detection:
        
        Z-Score: μ = (1/n)Σxi, σ = √[(1/(n-1))Σ(xi-μ)²]
        IQR: Q₁ = 25th percentile, Q₃ = 75th percentile
        Modified Z-Score: median, MAD = median(|xi - median|)
        """
        for col in X.columns:
            data = X[col].values
            
            if self.method == 'zscore':
                # Standard Z-score method
                mean = np.mean(data)
                std = np.std(data, ddof=1)  # Sample standard deviation
                self.params_[col] = {'mean': mean, 'std': std}
                
            elif self.method == 'iqr':
                # Interquartile Range method
                Q1 = np.percentile(data, 25)
                Q3 = np.percentile(data, 75)
                IQR = Q3 - Q1
                lower_bound = Q1 - 1.5 * IQR
                upper_bound = Q3 + 1.5 * IQR
                self.params_[col] = {'Q1': Q1, 'Q3': Q3, 'IQR': IQR, 
                                   'lower': lower_bound, 'upper': upper_bound}
                
            elif self.method == 'modified_zscore':
                # Modified Z-score using MAD
                median = np.median(data)
                mad = np.median(np.abs(data - median))
                self.params_[col] = {'median': median, 'mad': mad}
        
        return self
    
    def detect_outliers(self, X):
        """
        Detect outliers using mathematical formulas:
        
        Z-Score: |zi| = |(xi - μ) / σ| > threshold
        IQR: xi < Q₁ - 1.5×IQR or xi > Q₃ + 1.5×IQR  
        Modified Z-Score: |Mi| = |0.6745(xi - median) / MAD| > 3.5
        """
        outliers = {}
        
        for col in X.columns:
            data = X[col].values
            col_outliers = np.zeros(len(data), dtype=bool)
            
            if self.method == 'zscore':
                # |z| = |(x - μ) / σ| > threshold
                mean = self.params_[col]['mean']
                std = self.params_[col]['std']
                z_scores = np.abs((data - mean) / std)
                col_outliers = z_scores > self.threshold
                
            elif self.method == 'iqr':
                # x < Q₁ - 1.5×IQR or x > Q₃ + 1.5×IQR
                lower = self.params_[col]['lower']
                upper = self.params_[col]['upper']
                col_outliers = (data < lower) | (data > upper)
                
            elif self.method == 'modified_zscore':
                # |M| = |0.6745(x - median) / MAD| > 3.5
                median = self.params_[col]['median']
                mad = self.params_[col]['mad']
                if mad != 0:  # Avoid division by zero
                    modified_z_scores = np.abs(0.6745 * (data - median) / mad)
                    col_outliers = modified_z_scores > 3.5
                
            outliers[col] = col_outliers
            
        return outliers
    
    def fit_detect(self, X):
        return self.fit(X).detect_outliers(X)

# Apply different outlier detection methods
outlier_methods = {
    'Z-Score (|z| > 3)': MathematicalOutlierDetector('zscore', 3),
    'Z-Score (|z| > 2)': MathematicalOutlierDetector('zscore', 2),
    'IQR Method': MathematicalOutlierDetector('iqr'),
    'Modified Z-Score': MathematicalOutlierDetector('modified_zscore')
}

# Detect outliers with each method
detection_results = {}
for method_name, detector in outlier_methods.items():
    outliers = detector.fit_detect(outlier_data)
    detection_results[method_name] = outliers

# Comprehensive visualization
fig, axes = plt.subplots(len(outlier_data.columns), len(outlier_methods) + 1, 
                        figsize=(25, 15))

colors = ['blue', 'red', 'green']

for i, col in enumerate(outlier_data.columns):
    # Plot 1: Original data with box plot
    ax = axes[i, 0]
    data = outlier_data[col].values
    
    # Box plot
    bp = ax.boxplot([data], patch_artist=True)
    bp['boxes'][0].set_facecolor(colors[i])
    bp['boxes'][0].set_alpha(0.7)
    
    # Scatter plot overlay
    y_pos = np.ones(len(data))
    ax.scatter(y_pos, data, alpha=0.6, color=colors[i], s=20)
    
    ax.set_title(f'{col}\\nOriginal Data', fontweight='bold')
    ax.set_ylabel('Value')
    ax.grid(True, alpha=0.3)
    
    # Plot outlier detection results for each method
    for j, (method_name, outliers) in enumerate(detection_results.items(), 1):
        ax = axes[i, j]
        
        col_outliers = outliers[col]
        normal_points = data[~col_outliers]
        outlier_points = data[col_outliers]
        
        # Plot normal points
        if len(normal_points) > 0:
            ax.scatter(np.ones(len(normal_points)), normal_points, 
                      alpha=0.6, color='blue', s=20, label='Normal')
        
        # Plot outliers
        if len(outlier_points) > 0:
            ax.scatter(np.ones(len(outlier_points)), outlier_points, 
                      alpha=0.8, color='red', s=30, marker='x', 
                      label='Outliers')
        
        # Add statistics
        n_outliers = np.sum(col_outliers)
        outlier_pct = (n_outliers / len(data)) * 100
        
        ax.set_title(f'{method_name}\\n{col}\\nOutliers: {n_outliers} ({outlier_pct:.1f}%)', 
                    fontweight='bold')
        ax.set_ylabel('Value')
        ax.legend()
        ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Detailed comparison of outlier detection methods
print("\\n📊 OUTLIER DETECTION COMPARISON")
print("=" * 60)

comparison_table = []
for col in outlier_data.columns:
    for method_name, outliers in detection_results.items():
        col_outliers = outliers[col]
        n_outliers = np.sum(col_outliers)
        outlier_pct = (n_outliers / len(outlier_data)) * 100
        
        # Calculate outlier statistics
        if n_outliers > 0:
            outlier_values = outlier_data[col][col_outliers]
            min_outlier = outlier_values.min()
            max_outlier = outlier_values.max()
            mean_outlier = outlier_values.mean()
        else:
            min_outlier = max_outlier = mean_outlier = 'N/A'
        
        comparison_table.append({
            'Feature': col,
            'Method': method_name,
            'Count': n_outliers,
            'Percentage': f'{outlier_pct:.1f}%',
            'Min_Outlier': f'{min_outlier:.2f}' if min_outlier != 'N/A' else 'N/A',
            'Max_Outlier': f'{max_outlier:.2f}' if max_outlier != 'N/A' else 'N/A',
            'Mean_Outlier': f'{mean_outlier:.2f}' if mean_outlier != 'N/A' else 'N/A'
        })

df_comparison = pd.DataFrame(comparison_table)
print(df_comparison.to_string(index=False))

# Mathematical formulas and thresholds analysis
print("\\n🧮 MATHEMATICAL FORMULAS AND THRESHOLDS")
print("=" * 55)

for col in outlier_data.columns:
    print(f"\\n📊 {col}:")
    data = outlier_data[col].values
    
    # Z-Score method parameters
    mean = np.mean(data)
    std = np.std(data, ddof=1)
    print(f"   Z-Score: μ = {mean:.3f}, σ = {std:.3f}")
    print(f"   Formula: z = (x - {mean:.3f}) / {std:.3f}")
    print(f"   Outliers: |z| > 3 → |x - {mean:.3f}| > {3*std:.3f}")
    
    # IQR method parameters  
    Q1 = np.percentile(data, 25)
    Q3 = np.percentile(data, 75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    print(f"   IQR: Q₁ = {Q1:.3f}, Q₃ = {Q3:.3f}, IQR = {IQR:.3f}")
    print(f"   Bounds: [{lower_bound:.3f}, {upper_bound:.3f}]")
    
    # Modified Z-Score parameters
    median = np.median(data)
    mad = np.median(np.abs(data - median))
    print(f"   Modified Z-Score: median = {median:.3f}, MAD = {mad:.3f}")
    print(f"   Formula: M = 0.6745 × (x - {median:.3f}) / {mad:.3f}")

# Outlier impact analysis
print("\\n📈 OUTLIER IMPACT ANALYSIS")
print("=" * 40)

impact_analysis = []
for col in outlier_data.columns:
    original_data = outlier_data[col]
    
    # Statistics with outliers
    orig_mean = original_data.mean()
    orig_std = original_data.std()
    orig_median = original_data.median()
    
    # Use IQR method to identify outliers for removal
    iqr_outliers = detection_results['IQR Method'][col]
    cleaned_data = original_data[~iqr_outliers]
    
    # Statistics without outliers
    clean_mean = cleaned_data.mean()
    clean_std = cleaned_data.std()
    clean_median = cleaned_data.median()
    
    # Calculate impact
    mean_change = abs(orig_mean - clean_mean)
    std_change = abs(orig_std - clean_std)
    median_change = abs(orig_median - clean_median)
    
    impact_analysis.append({
        'Feature': col,
        'Original_Mean': f'{orig_mean:.3f}',
        'Clean_Mean': f'{clean_mean:.3f}',
        'Mean_Change': f'{mean_change:.3f}',
        'Original_Std': f'{orig_std:.3f}',
        'Clean_Std': f'{clean_std:.3f}',
        'Std_Change': f'{std_change:.3f}',
        'Median_Change': f'{median_change:.3f}'
    })

df_impact = pd.DataFrame(impact_analysis)
print(df_impact.to_string(index=False))

print("\\n💡 OUTLIER DETECTION METHOD GUIDELINES:")
print("=" * 50)
print("🎯 Z-Score Method:")
print("   • Best for: Normally distributed data")
print("   • Formula: z = (x - μ) / σ")
print("   • Threshold: |z| > 3 (conservative) or |z| > 2 (liberal)")
print("   • Pros: Simple, interpretable")
print("   • Cons: Sensitive to extreme outliers (masking effect)")

print("\\n🎯 IQR Method:")
print("   • Best for: Non-normal data, robust detection")
print("   • Formula: Q₁ - 1.5×IQR < x < Q₃ + 1.5×IQR")
print("   • Threshold: 1.5×IQR (standard), 3×IQR (conservative)")
print("   • Pros: Robust, doesn't assume distribution")
print("   • Cons: May be too aggressive for normal data")

print("\\n🎯 Modified Z-Score:")
print("   • Best for: Data with extreme outliers")
print("   • Formula: M = 0.6745 × (x - median) / MAD")
print("   • Threshold: |M| > 3.5")
print("   • Pros: Very robust, handles multiple outliers")
print("   • Cons: More complex, less intuitive")

print("\\n🎯 Treatment Recommendations:")
print("   • Remove: If outliers are data errors or irrelevant")
print("   • Transform: Apply log, sqrt, or Box-Cox transformation")
print("   • Cap: Replace with percentile values (winsorizing)")
print("   • Keep: If outliers represent valid extreme cases")

# Visualize outlier treatment effects
print("\\n🔧 OUTLIER TREATMENT DEMONSTRATION")
print("=" * 45)

# Demonstrate different treatment approaches
treatment_demo = outlier_data['Normal_Extreme'].copy()
iqr_outliers = detection_results['IQR Method']['Normal_Extreme']

# Method 1: Removal
treated_removal = treatment_demo[~iqr_outliers]

# Method 2: Capping (Winsorizing) at 5th and 95th percentiles
p5, p95 = np.percentile(treatment_demo, [5, 95])
treated_capping = treatment_demo.copy()
treated_capping[treated_capping < p5] = p5
treated_capping[treated_capping > p95] = p95

# Method 3: Log transformation
treated_log = np.log1p(treatment_demo - treatment_demo.min() + 1)

# Visualization
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Original data
axes[0, 0].hist(treatment_demo, bins=30, alpha=0.7, color='blue', edgecolor='black')
axes[0, 0].set_title('Original Data (with outliers)', fontweight='bold')
axes[0, 0].set_ylabel('Frequency')
axes[0, 0].grid(True, alpha=0.3)

# Outlier removal
axes[0, 1].hist(treated_removal, bins=30, alpha=0.7, color='green', edgecolor='black')
axes[0, 1].set_title('After Outlier Removal', fontweight='bold')
axes[0, 1].grid(True, alpha=0.3)

# Capping
axes[1, 0].hist(treated_capping, bins=30, alpha=0.7, color='orange', edgecolor='black')
axes[1, 0].set_title('After Capping (Winsorizing)', fontweight='bold')
axes[1, 0].set_xlabel('Value')
axes[1, 0].set_ylabel('Frequency')
axes[1, 0].grid(True, alpha=0.3)

# Log transformation
axes[1, 1].hist(treated_log, bins=30, alpha=0.7, color='purple', edgecolor='black')
axes[1, 1].set_title('After Log Transformation', fontweight='bold')
axes[1, 1].set_xlabel('Log-transformed Value')
axes[1, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Summary statistics for treatment comparison
treatment_stats = pd.DataFrame({
    'Method': ['Original', 'Removal', 'Capping', 'Log Transform'],
    'Count': [len(treatment_demo), len(treated_removal), len(treated_capping), len(treated_log)],
    'Mean': [treatment_demo.mean(), treated_removal.mean(), treated_capping.mean(), treated_log.mean()],
    'Std': [treatment_demo.std(), treated_removal.std(), treated_capping.std(), treated_log.std()],
    'Skewness': [stats.skew(treatment_demo), stats.skew(treated_removal), 
                stats.skew(treated_capping), stats.skew(treated_log)]
})

print("\\n📊 OUTLIER TREATMENT COMPARISON:")
print(treatment_stats.round(3).to_string(index=False))
```

```python
# 🔄 3. DATA TRANSFORMATION: BOX-COX & YEJO-JOHNSON ANALYSIS
print("\n🔄 3. Data Transformation: Mathematical Implementation & Visual Analysis")
print("-" * 80)

# Create skewed datasets for transformation analysis
np.random.seed(42)
n_samples = 1000

# Generate different types of skewed data
right_skewed = np.random.exponential(2, n_samples)  # Right skewed
left_skewed = -np.random.exponential(2, n_samples) + 10  # Left skewed
heavy_tailed = np.random.pareto(1.5, n_samples)  # Heavy tailed
bimodal = np.concatenate([np.random.normal(2, 1, n_samples//2), 
                         np.random.normal(8, 1.5, n_samples//2)])  # Bimodal

transform_data = pd.DataFrame({
    'Right_Skewed': right_skewed,
    'Left_Skewed': left_skewed,
    'Heavy_Tailed': np.clip(heavy_tailed, 0, 20),  # Clip extreme values
    'Bimodal': bimodal
})

print("📊 Original Skewed Data Statistics:")
print(transform_data.describe())

# Calculate skewness and kurtosis
print("\\n📈 Distribution Characteristics:")
for col in transform_data.columns:
    skewness = stats.skew(transform_data[col])
    kurt = stats.kurtosis(transform_data[col])
    print(f"{col:15s}: Skewness = {skewness:6.3f}, Kurtosis = {kurt:6.3f}")

class MathematicalTransformer:
    def __init__(self, method='boxcox'):
        self.method = method
        self.lambdas_ = {}
        
    def fit_transform(self, X):
        """
        Mathematical implementation of transformations:
        
        Box-Cox: y = (x^λ - 1) / λ  if λ ≠ 0, else y = ln(x)
        Yeo-Johnson: Handles negative values with different formulas per quadrant
        """
        X_transformed = X.copy()
        
        for col in X.columns:
            data = X[col].values
            
            if self.method == 'boxcox':
                # Box-Cox requires positive values
                if np.any(data <= 0):
                    # Shift data to be positive
                    data_shifted = data - np.min(data) + 1
                    print(f"⚠️  {col}: Shifted by {-np.min(data) + 1:.3f} for Box-Cox")
                else:
                    data_shifted = data
                
                # Find optimal lambda using maximum likelihood
                lambdas = np.linspace(-2, 2, 100)
                log_likelihoods = []
                
                for lam in lambdas:
                    if lam == 0:
                        # y = ln(x)
                        y_transformed = np.log(data_shifted)
                    else:
                        # y = (x^λ - 1) / λ
                        y_transformed = (data_shifted**lam - 1) / lam
                    
                    # Log-likelihood calculation
                    y_mean = np.mean(y_transformed)
                    ss_res = np.sum((y_transformed - y_mean)**2)
                    n = len(data_shifted)
                    
                    # ℓ(λ) = -n/2 * ln(SS_res/n) + (λ-1) * Σln(xi)
                    ll = -n/2 * np.log(ss_res/n) + (lam-1) * np.sum(np.log(data_shifted))
                    log_likelihoods.append(ll)
                
                # Find optimal lambda
                optimal_idx = np.argmax(log_likelihoods)
                optimal_lambda = lambdas[optimal_idx]
                self.lambdas_[col] = optimal_lambda
                
                # Apply transformation with optimal lambda
                if optimal_lambda == 0:
                    X_transformed[col] = np.log(data_shifted)
                else:
                    X_transformed[col] = (data_shifted**optimal_lambda - 1) / optimal_lambda
                
            elif self.method == 'yeojohnson':
                # Yeo-Johnson transformation handles negative values
                transformed, optimal_lambda = yeojohnson(data)
                X_transformed[col] = transformed
                self.lambdas_[col] = optimal_lambda
                
            elif self.method == 'log':
                # Simple log transformation (add 1 to handle zeros)
                X_transformed[col] = np.log1p(data - np.min(data) + 1e-8)
                self.lambdas_[col] = None
                
            elif self.method == 'sqrt':
                # Square root transformation
                X_transformed[col] = np.sqrt(data - np.min(data) + 1e-8)
                self.lambdas_[col] = None
        
        return X_transformed

# Apply different transformation methods
transformation_methods = {
    'Original': transform_data,
    'Box-Cox (Custom)': MathematicalTransformer('boxcox').fit_transform(transform_data),
    'Yeo-Johnson (Custom)': MathematicalTransformer('yeojohnson').fit_transform(transform_data),
    'Log Transform': MathematicalTransformer('log').fit_transform(transform_data),
    'Square Root': MathematicalTransformer('sqrt').fit_transform(transform_data)
}

# Add sklearn transformations for comparison
for col in transform_data.columns:
    if np.all(transform_data[col] > 0):  # Box-Cox requires positive values
        try:
            pt_boxcox = PowerTransformer(method='box-cox')
            transformed_boxcox = pt_boxcox.fit_transform(transform_data[[col]])
            if 'Sklearn Box-Cox' not in transformation_methods:
                transformation_methods['Sklearn Box-Cox'] = transform_data.copy()
            transformation_methods['Sklearn Box-Cox'][col] = transformed_boxcox.flatten()
        except:
            pass

# Yeo-Johnson works with any data
pt_yj = PowerTransformer(method='yeo-johnson')
transformation_methods['Sklearn Yeo-Johnson'] = pd.DataFrame(
    pt_yj.fit_transform(transform_data), 
    columns=transform_data.columns
)

# Comprehensive visualization
fig, axes = plt.subplots(len(transform_data.columns), len(transformation_methods), 
                        figsize=(25, 20))

for i, col in enumerate(transform_data.columns):
    for j, (method_name, transformed_data) in enumerate(transformation_methods.items()):
        ax = axes[i, j]
        
        # Histogram
        ax.hist(transformed_data[col], bins=30, alpha=0.7, color='skyblue', 
               edgecolor='black', density=True)
        
        # Overlay normal distribution for comparison
        mean = transformed_data[col].mean()
        std = transformed_data[col].std()
        x_norm = np.linspace(transformed_data[col].min(), transformed_data[col].max(), 100)
        y_norm = stats.norm.pdf(x_norm, mean, std)
        ax.plot(x_norm, y_norm, 'r-', linewidth=2, alpha=0.8, label='Normal')
        
        # Calculate and display statistics
        skewness = stats.skew(transformed_data[col])
        kurt = stats.kurtosis(transformed_data[col])
        
        ax.set_title(f'{method_name}\\n{col}\\nSkew: {skewness:.2f}, Kurt: {kurt:.2f}', 
                    fontweight='bold', fontsize=10)
        ax.grid(True, alpha=0.3)
        
        if i == 0:  # Add method name to top row
            ax.set_xlabel(method_name, fontweight='bold')
        if j == 0:  # Add feature name to first column
            ax.set_ylabel(col, fontweight='bold')

plt.tight_layout()
plt.show()

# Transformation effectiveness analysis
print("\\n📊 TRANSFORMATION EFFECTIVENESS ANALYSIS")
print("=" * 70)

effectiveness_data = []
for method_name, transformed_data in transformation_methods.items():
    for col in transformed_data.columns:
        original_skew = abs(stats.skew(transform_data[col]))
        transformed_skew = abs(stats.skew(transformed_data[col]))
        
        original_kurt = abs(stats.kurtosis(transform_data[col]))
        transformed_kurt = abs(stats.kurtosis(transformed_data[col]))
        
        skew_improvement = ((original_skew - transformed_skew) / original_skew) * 100
        kurt_improvement = ((original_kurt - transformed_kurt) / original_kurt) * 100
        
        effectiveness_data.append({
            'Method': method_name,
            'Feature': col,
            'Original_Skew': f'{original_skew:.3f}',
            'Transformed_Skew': f'{transformed_skew:.3f}',
            'Skew_Improvement': f'{skew_improvement:.1f}%',
            'Kurt_Improvement': f'{kurt_improvement:.1f}%'
        })

df_effectiveness = pd.DataFrame(effectiveness_data)

# Display best transformations for each feature
print("🏆 BEST TRANSFORMATION FOR EACH FEATURE (by skewness reduction):")
print("-" * 65)

for col in transform_data.columns:
    feature_data = df_effectiveness[df_effectiveness['Feature'] == col]
    feature_data = feature_data[feature_data['Method'] != 'Original']
    
    # Convert improvement to numeric for sorting
    feature_data_copy = feature_data.copy()
    feature_data_copy['Skew_Improvement_Numeric'] = feature_data_copy['Skew_Improvement'].str.rstrip('%').astype(float)
    
    best_method = feature_data_copy.loc[feature_data_copy['Skew_Improvement_Numeric'].idxmax()]
    
    print(f"{col:15s}: {best_method['Method']:20s} "
          f"(Skew: {best_method['Original_Skew']} → {best_method['Transformed_Skew']}, "
          f"Improvement: {best_method['Skew_Improvement']})")

# Mathematical formulas recap
print("\\n🧮 MATHEMATICAL FORMULAS IMPLEMENTED:")
print("=" * 50)
print("📐 Box-Cox Transformation:")
print("   y(λ) = (x^λ - 1) / λ     if λ ≠ 0")
print("   y(λ) = ln(x)             if λ = 0")
print("   ")
print("   Optimal λ maximizes log-likelihood:")
print("   ℓ(λ) = -n/2 * ln[(1/n)Σ(y_i^(λ) - ȳ^(λ))²] + (λ-1)Σln(x_i)")

print("\\n📐 Yeo-Johnson Transformation:")
print("   For x ≥ 0, λ ≠ 0: y = ((x+1)^λ - 1) / λ")
print("   For x ≥ 0, λ = 0: y = ln(x + 1)")
print("   For x < 0, λ ≠ 2: y = -((-x+1)^(2-λ) - 1) / (2-λ)")
print("   For x < 0, λ = 2: y = -ln(-x + 1)")

print("\\n📐 Log Transformation:")
print("   y = ln(x + c)    where c > 0 ensures positivity")

print("\\n📐 Square Root Transformation:")
print("   y = √(x + c)     where c ≥ 0 ensures non-negativity")

# Normality test results
print("\\n🧪 NORMALITY TEST RESULTS (Shapiro-Wilk p-values):")
print("=" * 60)
print("📊 p > 0.05 suggests data is approximately normal")

normality_results = []
for method_name, transformed_data in transformation_methods.items():
    for col in transformed_data.columns:
        if len(transformed_data[col]) <= 5000:  # Shapiro-Wilk limitation
            try:
                _, p_value = stats.shapiro(transformed_data[col])
                is_normal = "✅ Normal" if p_value > 0.05 else "❌ Not Normal"
                normality_results.append({
                    'Method': method_name,
                    'Feature': col,
                    'p_value': f'{p_value:.6f}',
                    'Normal': is_normal
                })
            except:
                normality_results.append({
                    'Method': method_name,
                    'Feature': col,
                    'p_value': 'Error',
                    'Normal': 'Error'
                })

df_normality = pd.DataFrame(normality_results)

# Show best normality results for each feature
for col in transform_data.columns:
    feature_normality = df_normality[df_normality['Feature'] == col]
    feature_normality = feature_normality[feature_normality['Method'] != 'Original']
    
    # Find method with highest p-value (most normal)
    feature_normality_numeric = feature_normality[feature_normality['p_value'] != 'Error'].copy()
    if not feature_normality_numeric.empty:
        feature_normality_numeric['p_value_num'] = feature_normality_numeric['p_value'].astype(float)
        best_normal = feature_normality_numeric.loc[feature_normality_numeric['p_value_num'].idxmax()]
        print(f"{col:15s}: Best normality with {best_normal['Method']:20s} (p = {best_normal['p_value']})")

print("\\n💡 TRANSFORMATION SELECTION GUIDELINES:")
print("=" * 45)
print("🎯 Use Box-Cox when:")
print("   • Data is strictly positive")
print("   • Want to find optimal transformation parameter")
print("   • Need to minimize skewness")

print("\\n🎯 Use Yeo-Johnson when:")
print("   • Data contains negative values or zeros")
print("   • Want Box-Cox-like transformation without positivity constraint")
print("   • Need robust transformation for mixed-sign data")

print("\\n🎯 Use Log transformation when:")
print("   • Data is right-skewed")
print("   • Want to compress large values")
print("   • Dealing with multiplicative relationships")

print("\\n🎯 Use Square Root when:")
print("   • Data is count data or right-skewed")
print("   • Want moderate variance stabilization")
print("   • Less aggressive than log transformation")
```

```python
# 📏 2. FEATURE SCALING: MATHEMATICAL IMPLEMENTATIONS & COMPARISONS
print("\n📏 2. Feature Scaling: Mathematical Implementations & Visual Analysis")
print("-" * 75)

# Create diverse dataset for scaling comparison
np.random.seed(42)

# Generate features with different scales and distributions
feature1 = np.random.normal(100, 15, 1000)  # Normal distribution, high mean
feature2 = np.random.exponential(2, 1000)   # Exponential distribution, low values
feature3 = np.random.uniform(-50, 150, 1000)  # Uniform distribution, wide range
feature4 = np.random.lognormal(0, 1, 1000)  # Log-normal distribution, skewed

# Add some outliers
feature1[np.random.choice(1000, 20)] += np.random.normal(0, 50, 20)
feature2[np.random.choice(1000, 15)] += np.random.exponential(10, 15)

scaling_data = pd.DataFrame({
    'Normal': feature1,
    'Exponential': feature2, 
    'Uniform': feature3,
    'LogNormal': feature4
})

print("📊 Original Data Statistics:")
print(scaling_data.describe())

# Mathematical implementation of scaling methods
class MathematicalScaler:
    def __init__(self, method='standard'):
        self.method = method
        self.params_ = {}
        
    def fit(self, X):
        """
        Calculate scaling parameters based on mathematical formulas:
        
        StandardScaler: μ = (1/n)Σxi, σ = √[(1/(n-1))Σ(xi-μ)²]
        MinMaxScaler: min(X), max(X)  
        RobustScaler: median(X), IQR = Q₃ - Q₁
        MaxAbsScaler: max(|X|)
        """
        for col in X.columns:
            if self.method == 'standard':
                # μ = (1/n) * Σ(xi)
                mean = X[col].mean()
                # σ = √[(1/(n-1)) * Σ(xi - μ)²]
                std = X[col].std()
                self.params_[col] = {'mean': mean, 'std': std}
                
            elif self.method == 'minmax':
                # Store min and max values
                min_val = X[col].min()
                max_val = X[col].max()
                self.params_[col] = {'min': min_val, 'max': max_val}
                
            elif self.method == 'robust':
                # median and IQR = Q₃ - Q₁
                median = X[col].median()
                q25 = X[col].quantile(0.25)
                q75 = X[col].quantile(0.75)
                iqr = q75 - q25
                self.params_[col] = {'median': median, 'iqr': iqr}
                
            elif self.method == 'maxabs':
                # max(|X|)
                max_abs = X[col].abs().max()
                self.params_[col] = {'max_abs': max_abs}
        
        return self
    
    def transform(self, X):
        """Apply scaling transformation"""
        X_scaled = X.copy()
        
        for col in X.columns:
            if self.method == 'standard':
                # x_std = (x - μ) / σ
                mean = self.params_[col]['mean']
                std = self.params_[col]['std']
                X_scaled[col] = (X[col] - mean) / std
                
            elif self.method == 'minmax':
                # x_norm = (x - min) / (max - min)
                min_val = self.params_[col]['min']
                max_val = self.params_[col]['max']
                X_scaled[col] = (X[col] - min_val) / (max_val - min_val)
                
            elif self.method == 'robust':
                # x_robust = (x - median) / IQR
                median = self.params_[col]['median']
                iqr = self.params_[col]['iqr']
                X_scaled[col] = (X[col] - median) / iqr
                
            elif self.method == 'maxabs':
                # x_scaled = x / max(|x|)
                max_abs = self.params_[col]['max_abs']
                X_scaled[col] = X[col] / max_abs
        
        return X_scaled
    
    def fit_transform(self, X):
        return self.fit(X).transform(X)

# Apply different scaling methods
scaling_methods = {
    'Original': scaling_data,
    'Standard (Z-Score)': MathematicalScaler('standard').fit_transform(scaling_data),
    'Min-Max [0,1]': MathematicalScaler('minmax').fit_transform(scaling_data),
    'Robust (IQR)': MathematicalScaler('robust').fit_transform(scaling_data),
    'Max Absolute': MathematicalScaler('maxabs').fit_transform(scaling_data),
    'Sklearn Standard': pd.DataFrame(StandardScaler().fit_transform(scaling_data), 
                                   columns=scaling_data.columns),
    'Sklearn MinMax': pd.DataFrame(MinMaxScaler().fit_transform(scaling_data), 
                                 columns=scaling_data.columns)
}

# Comprehensive visualization
fig, axes = plt.subplots(3, 3, figsize=(20, 15))
axes = axes.ravel()

colors = ['blue', 'red', 'green', 'orange']
feature_names = scaling_data.columns

for i, (method_name, scaled_data) in enumerate(scaling_methods.items()):
    if i < len(axes):
        ax = axes[i]
        
        # Box plot for all features
        box_data = [scaled_data[col].values for col in scaled_data.columns]
        bp = ax.boxplot(box_data, labels=feature_names, patch_artist=True)
        
        # Color the boxes
        for patch, color in zip(bp['boxes'], colors):
            patch.set_facecolor(color)
            patch.set_alpha(0.7)
        
        ax.set_title(f'{method_name}', fontweight='bold', fontsize=12)
        ax.grid(True, alpha=0.3)
        ax.tick_params(axis='x', rotation=45)
        
        # Add statistical information
        if method_name != 'Original':
            ranges = [scaled_data[col].max() - scaled_data[col].min() for col in scaled_data.columns]
            avg_range = np.mean(ranges)
            ax.text(0.02, 0.98, f'Avg Range: {avg_range:.2f}', transform=ax.transAxes, 
                   verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

# Remove empty subplots
for i in range(len(scaling_methods), len(axes)):
    fig.delaxes(axes[i])

plt.tight_layout()
plt.show()

# Statistical comparison of scaling methods
print("\\n📈 SCALING METHODS COMPARISON TABLE")
print("=" * 80)

comparison_stats = []
for method_name, scaled_data in scaling_methods.items():
    stats_dict = {
        'Method': method_name,
        'Mean Range': f"{np.mean([scaled_data[col].max() - scaled_data[col].min() for col in scaled_data.columns]):.3f}",
        'Std Range': f"{np.std([scaled_data[col].max() - scaled_data[col].min() for col in scaled_data.columns]):.3f}",
        'Overall Mean': f"{scaled_data.mean().mean():.3f}",
        'Overall Std': f"{scaled_data.std().mean():.3f}"
    }
    
    # Check for bounded output
    all_bounded = all(scaled_data[col].min() >= -0.01 and scaled_data[col].max() <= 1.01 
                     for col in scaled_data.columns)
    stats_dict['Bounded [0,1]'] = 'Yes' if 'MinMax' in method_name and all_bounded else 'No'
    
    comparison_stats.append(stats_dict)

df_scaling_comparison = pd.DataFrame(comparison_stats)
print(df_scaling_comparison.to_string(index=False))

# Distribution analysis before and after scaling
print("\\n🎯 DISTRIBUTION ANALYSIS: BEFORE AND AFTER SCALING")
print("-" * 60)

fig, axes = plt.subplots(2, 4, figsize=(20, 10))

# Original distributions
for i, col in enumerate(scaling_data.columns):
    axes[0, i].hist(scaling_data[col], bins=30, alpha=0.7, color=colors[i], edgecolor='black')
    axes[0, i].set_title(f'Original: {col}', fontweight='bold')
    axes[0, i].set_ylabel('Frequency')
    axes[0, i].grid(True, alpha=0.3)
    
    # Add statistics
    mean_val = scaling_data[col].mean()
    std_val = scaling_data[col].std()
    axes[0, i].axvline(mean_val, color='red', linestyle='--', alpha=0.8, label=f'μ={mean_val:.1f}')
    axes[0, i].axvline(mean_val + std_val, color='orange', linestyle='--', alpha=0.8, label=f'μ+σ')
    axes[0, i].axvline(mean_val - std_val, color='orange', linestyle='--', alpha=0.8, label=f'μ-σ')
    axes[0, i].legend(fontsize=8)

# Standardized distributions
standardized_data = scaling_methods['Standard (Z-Score)']
for i, col in enumerate(standardized_data.columns):
    axes[1, i].hist(standardized_data[col], bins=30, alpha=0.7, color=colors[i], edgecolor='black')
    axes[1, i].set_title(f'Standardized: {col}', fontweight='bold')
    axes[1, i].set_xlabel('Standardized Value')
    axes[1, i].set_ylabel('Frequency')
    axes[1, i].grid(True, alpha=0.3)
    
    # Add statistics
    mean_val = standardized_data[col].mean()
    std_val = standardized_data[col].std()
    axes[1, i].axvline(mean_val, color='red', linestyle='--', alpha=0.8, label=f'μ={mean_val:.2f}')
    axes[1, i].axvline(0, color='green', linestyle='-', alpha=0.8, label='μ=0')
    axes[1, i].legend(fontsize=8)

plt.tight_layout()
plt.show()

print("\\n💡 SCALING METHOD RECOMMENDATIONS:")
print("=" * 45)
print("🎯 StandardScaler (Z-Score):")
print("   • Use when: Features follow normal distribution")
print("   • Formula: (x - μ) / σ")
print("   • Result: μ = 0, σ = 1")
print("   • Best for: Linear regression, logistic regression, neural networks")

print("\\n🎯 MinMaxScaler:")
print("   • Use when: Need bounded output [0,1]") 
print("   • Formula: (x - min) / (max - min)")
print("   • Result: Range [0, 1]")
print("   • Best for: Neural networks, image processing, when you know min/max")

print("\\n🎯 RobustScaler:")
print("   • Use when: Data has outliers")
print("   • Formula: (x - median) / IQR")
print("   • Result: Less sensitive to outliers")
print("   • Best for: Data with outliers, robust statistics")

print("\\n🎯 MaxAbsScaler:")
print("   • Use when: Sparse data, preserve zero values")
print("   • Formula: x / max(|x|)")
print("   • Result: Range [-1, 1], preserves sparsity")
print("   • Best for: Sparse matrices, when zero values are important")
```

```python
# 📊 1. MISSING DATA IMPUTATION: COMPREHENSIVE ANALYSIS
print("📊 1. Missing Data Imputation: Mathematical Implementation & Comparison")
print("-" * 70)

# Create synthetic dataset with missing values
np.random.seed(42)
n_samples = 1000
n_features = 4

# Generate correlated features
mean = [0, 0, 0, 0]
cov = [[1, 0.7, 0.3, 0.1],
       [0.7, 1, 0.5, 0.2],
       [0.3, 0.5, 1, 0.4],
       [0.1, 0.2, 0.4, 1]]

data = np.random.multivariate_normal(mean, cov, n_samples)

# Add some non-linear relationships
data[:, 1] = data[:, 1] + 0.5 * data[:, 0]**2
data[:, 2] = data[:, 2] + np.sin(data[:, 0])
data[:, 3] = data[:, 3] + 0.3 * data[:, 1] * data[:, 2]

# Create DataFrame
feature_names = ['Feature_A', 'Feature_B', 'Feature_C', 'Feature_D']
df_complete = pd.DataFrame(data, columns=feature_names)

# Introduce missing values with different patterns
df_missing = df_complete.copy()

# Missing Completely at Random (MCAR) - 10%
np.random.seed(42)
mcar_mask = np.random.random(n_samples) < 0.1
df_missing.loc[mcar_mask, 'Feature_A'] = np.nan

# Missing at Random (MAR) - correlated with another feature
mar_mask = df_missing['Feature_B'] > 1.0  # 15-20% missing
df_missing.loc[mar_mask, 'Feature_C'] = np.nan

# Missing Not at Random (MNAR) - missing depends on the value itself
mnar_mask = df_missing['Feature_D'] < -0.5  # 12-15% missing  
df_missing.loc[mnar_mask, 'Feature_D'] = np.nan

print(f"📈 Dataset shape: {df_missing.shape}")
print(f"📉 Missing values per feature:")
for col in df_missing.columns:
    missing_count = df_missing[col].isnull().sum()
    missing_pct = (missing_count / len(df_missing)) * 100
    print(f"   {col}: {missing_count} ({missing_pct:.1f}%)")

# Visualization of missing data patterns
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Missing data heatmap
missing_matrix = df_missing.isnull()
sns.heatmap(missing_matrix, cbar=True, ax=axes[0, 0], cmap='viridis')
axes[0, 0].set_title('Missing Data Pattern', fontweight='bold')
axes[0, 0].set_xlabel('Features')
axes[0, 0].set_ylabel('Samples')

# Missing data bar plot
missing_counts = df_missing.isnull().sum()
axes[0, 1].bar(range(len(missing_counts)), missing_counts.values, color='coral', alpha=0.7)
axes[0, 1].set_title('Missing Values per Feature', fontweight='bold')
axes[0, 1].set_xlabel('Features')
axes[0, 1].set_ylabel('Count of Missing Values')
axes[0, 1].set_xticks(range(len(missing_counts)))
axes[0, 1].set_xticklabels(missing_counts.index, rotation=45)

# Distribution comparison (before/after missing)
feature_to_plot = 'Feature_A'
axes[1, 0].hist(df_complete[feature_to_plot], bins=30, alpha=0.7, label='Complete Data', color='blue')
axes[1, 0].hist(df_missing[feature_to_plot].dropna(), bins=30, alpha=0.7, label='After Missing', color='red')
axes[1, 0].set_title(f'{feature_to_plot}: Distribution Impact', fontweight='bold')
axes[1, 0].set_xlabel('Value')
axes[1, 0].set_ylabel('Frequency')
axes[1, 0].legend()

# Correlation matrix comparison
corr_complete = df_complete.corr()
corr_missing = df_missing.corr()
corr_diff = corr_complete - corr_missing

im = axes[1, 1].imshow(corr_diff, cmap='RdBu_r', vmin=-0.1, vmax=0.1)
axes[1, 1].set_title('Correlation Change due to Missing Data', fontweight='bold')
axes[1, 1].set_xticks(range(len(feature_names)))
axes[1, 1].set_yticks(range(len(feature_names)))
axes[1, 1].set_xticklabels(feature_names, rotation=45)
axes[1, 1].set_yticklabels(feature_names)

# Add colorbar
plt.colorbar(im, ax=axes[1, 1], label='Correlation Difference')

# Add correlation values as text
for i in range(len(feature_names)):
    for j in range(len(feature_names)):
        axes[1, 1].text(j, i, f'{corr_diff.iloc[i, j]:.3f}', 
                       ha='center', va='center', fontweight='bold')

plt.tight_layout()
plt.show()

print("\\n🔍 MATHEMATICAL IMPLEMENTATION OF IMPUTATION METHODS")
print("=" * 60)

# Custom implementation of imputation methods for educational purposes
class MathematicalImputer:
    def __init__(self, strategy='mean'):
        self.strategy = strategy
        self.fill_values_ = {}
        
    def fit(self, X):
        """
        Calculate imputation values based on strategy
        
        Mathematical formulas implemented:
        - Mean: μ = (1/n) * Σ(xi) for i=1 to n
        - Median: Middle value when data is sorted
        - Mode: Most frequent value
        """
        for col in X.columns:
            if X[col].isnull().any():
                if self.strategy == 'mean':
                    # μ = (1/n) * Σ(xi)
                    self.fill_values_[col] = X[col].mean()
                elif self.strategy == 'median':
                    # Median = middle value of sorted data
                    self.fill_values_[col] = X[col].median()
                elif self.strategy == 'mode':
                    # Mode = most frequent value
                    self.fill_values_[col] = X[col].mode().iloc[0] if not X[col].mode().empty else 0
                
        return self
    
    def transform(self, X):
        """Apply imputation"""
        X_imputed = X.copy()
        for col, fill_value in self.fill_values_.items():
            X_imputed[col].fillna(fill_value, inplace=True)
        return X_imputed
    
    def fit_transform(self, X):
        return self.fit(X).transform(X)

class KNNImputerMath:
    def __init__(self, k=5):
        self.k = k
        
    def fit_transform(self, X):
        """
        KNN Imputation with mathematical formula:
        
        For missing value xi,m:
        1. Calculate distances: d(i,j) = √(Σ(xif - xjf)²) for all f ≠ m
        2. Find k nearest neighbors: Nk(i)
        3. Impute: x̂i,m = (1/k) * Σ(xj,m) for j ∈ Nk(i)
        """
        X_imputed = X.copy()
        
        for col in X.columns:
            if X[col].isnull().any():
                missing_idx = X[col].isnull()
                
                for idx in X[missing_idx].index:
                    # Calculate distances to all complete samples
                    complete_samples = X[~X[col].isnull()]
                    distances = []
                    
                    for comp_idx in complete_samples.index:
                        # Euclidean distance using available features
                        available_cols = [c for c in X.columns if not pd.isna(X.loc[idx, c]) and not pd.isna(X.loc[comp_idx, c])]
                        if available_cols:
                            dist = np.sqrt(sum((X.loc[idx, c] - X.loc[comp_idx, c])**2 for c in available_cols))
                            distances.append((dist, comp_idx))
                    
                    # Find k nearest neighbors
                    distances.sort()
                    k_neighbors = distances[:min(self.k, len(distances))]
                    
                    # Impute as average of k neighbors
                    if k_neighbors:
                        imputed_value = np.mean([X.loc[neighbor_idx, col] for _, neighbor_idx in k_neighbors])
                        X_imputed.loc[idx, col] = imputed_value
        
        return X_imputed

# Apply different imputation methods
imputation_methods = {
    'Mean': MathematicalImputer('mean'),
    'Median': MathematicalImputer('median'), 
    'KNN (k=5)': KNNImputerMath(k=5),
    'Sklearn KNN': KNNImputer(n_neighbors=5),
    'Iterative': IterativeImputer(random_state=42, max_iter=10)
}

imputed_results = {}
for name, imputer in imputation_methods.items():
    print(f"🔧 Applying {name} imputation...")
    imputed_data = imputer.fit_transform(df_missing)
    imputed_results[name] = imputed_data

print("\\n✅ All imputation methods completed!")
```

## 📚 Mathematical Foundations of Data Preprocessing

Data preprocessing is a crucial step in machine learning that transforms raw data into a format suitable for analysis. Below are the mathematical foundations and formulas for key preprocessing techniques.

---

### 🎯 1. Missing Data Imputation Methods

#### **Mean Imputation**
For a feature $X$ with missing values, replace missing entries with the mean:

$$\hat{x}_i = \begin{cases} 
x_i & \text{if } x_i \text{ is observed} \\
\bar{x} & \text{if } x_i \text{ is missing}
\end{cases}$$

Where $\bar{x} = \frac{1}{n} \sum_{j=1}^{n} x_j$ (mean of observed values)

#### **Median Imputation**
Replace missing values with the median:
$$\hat{x}_i = \begin{cases} 
x_i & \text{if } x_i \text{ is observed} \\
\text{median}(X) & \text{if } x_i \text{ is missing}
\end{cases}$$

#### **K-Nearest Neighbors (KNN) Imputation**
For a missing value in sample $i$, find $k$ nearest neighbors based on Euclidean distance:

$$d(i,j) = \sqrt{\sum_{f=1}^{p} (x_{if} - x_{jf})^2}$$

The imputed value is:
$$\hat{x}_{im} = \frac{1}{k} \sum_{j \in N_k(i)} x_{jm}$$

Where $N_k(i)$ represents the $k$ nearest neighbors of sample $i$.

#### **Iterative Imputation (MICE)**
Models each feature with missing values as a function of other features:

$$X_j = f_j(X_{-j}) + \epsilon_j$$

Where $X_{-j}$ represents all features except $j$, and $\epsilon_j$ is the error term.

---

### 🎯 2. Feature Scaling Techniques

#### **Min-Max Normalization (Min-Max Scaling)**
Scales features to a fixed range $[0, 1]$:

$$x_{norm} = \frac{x - x_{min}}{x_{max} - x_{min}}$$

**Generalized form for range $[a, b]$:**
$$x_{norm} = a + \frac{(x - x_{min})(b - a)}{x_{max} - x_{min}}$$

#### **Standardization (Z-Score Normalization)**
Transforms data to have zero mean and unit variance:

$$x_{std} = \frac{x - \mu}{\sigma}$$

Where:
- $\mu = \frac{1}{n}\sum_{i=1}^{n} x_i$ (sample mean)
- $\sigma = \sqrt{\frac{1}{n-1}\sum_{i=1}^{n} (x_i - \mu)^2}$ (sample standard deviation)

#### **Robust Scaling**
Uses median and interquartile range to reduce impact of outliers:

$$x_{robust} = \frac{x - \text{median}(X)}{\text{IQR}(X)}$$

Where $\text{IQR} = Q_3 - Q_1$ (75th percentile - 25th percentile)

#### **Max Absolute Scaling**
Scales by the maximum absolute value:

$$x_{maxabs} = \frac{x}{\max(|X|)}$$

Where $\max(|X|) = \max_{i} |x_i|$

---

### 🎯 3. Data Transformation Methods

#### **Box-Cox Transformation**
Stabilizes variance and makes data more normal-like:

$$y_i^{(\lambda)} = \begin{cases}
\frac{x_i^{\lambda} - 1}{\lambda} & \text{if } \lambda \neq 0 \\
\ln(x_i) & \text{if } \lambda = 0
\end{cases}$$

**Optimal $\lambda$ maximizes log-likelihood:**
$$\ell(\lambda) = -\frac{n}{2}\ln\left[\frac{1}{n}\sum_{i=1}^{n}(y_i^{(\lambda)} - \bar{y}^{(\lambda)})^2\right] + (\lambda-1)\sum_{i=1}^{n}\ln(x_i)$$

#### **Yeo-Johnson Transformation**
Extends Box-Cox to handle negative values:

$$y_i^{(\lambda)} = \begin{cases}
\frac{(x_i+1)^{\lambda} - 1}{\lambda} & \text{if } x_i \geq 0, \lambda \neq 0 \\
\ln(x_i + 1) & \text{if } x_i \geq 0, \lambda = 0 \\
-\frac{((-x_i)+1)^{2-\lambda} - 1}{2-\lambda} & \text{if } x_i < 0, \lambda \neq 2 \\
-\ln((-x_i) + 1) & \text{if } x_i < 0, \lambda = 2
\end{cases}$$

#### **Quantile Transformation**
Maps data to a uniform or normal distribution:

**Uniform transformation:**
$$F(x) = \frac{\text{rank}(x)}{n+1}$$

**Normal transformation:**
$$G^{-1}(F(x)) = \Phi^{-1}\left(\frac{\text{rank}(x)}{n+1}\right)$$

Where $\Phi^{-1}$ is the inverse CDF of the standard normal distribution.

---

### 🎯 4. Outlier Detection and Treatment

#### **Z-Score Method**
Identify outliers using standardized scores:
$$z_i = \frac{x_i - \mu}{\sigma}$$

**Outlier criterion:** $|z_i| > k$ (typically $k = 2$ or $3$)

#### **Interquartile Range (IQR) Method**
$$\text{Lower Bound} = Q_1 - 1.5 \times \text{IQR}$$
$$\text{Upper Bound} = Q_3 + 1.5 \times \text{IQR}$$

**Outliers:** $x_i < \text{Lower Bound}$ or $x_i > \text{Upper Bound}$

#### **Modified Z-Score (Median Absolute Deviation)**
More robust to outliers:
$$M_i = \frac{0.6745(x_i - \tilde{x})}{\text{MAD}}$$

Where:
- $\tilde{x}$ is the median
- $\text{MAD} = \text{median}(|x_i - \tilde{x}|)$ (Median Absolute Deviation)

**Outlier criterion:** $|M_i| > 3.5$

---

### 🎯 5. Advanced Preprocessing Techniques

#### **Principal Component Analysis (PCA) for Dimensionality Reduction**
Find principal components that maximize variance:

$$\mathbf{Y} = \mathbf{XW}$$

Where $\mathbf{W}$ contains eigenvectors of covariance matrix $\mathbf{C}$:
$$\mathbf{C} = \frac{1}{n-1}\mathbf{X}^T\mathbf{X}$$

**Variance explained by component $i$:**
$$\text{Explained Variance Ratio}_i = \frac{\lambda_i}{\sum_{j=1}^{p} \lambda_j}$$

#### **Feature Selection using Correlation**
Remove highly correlated features using Pearson correlation:

$$r_{xy} = \frac{\sum_{i=1}^{n}(x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum_{i=1}^{n}(x_i - \bar{x})^2}\sqrt{\sum_{i=1}^{n}(y_i - \bar{y})^2}}$$

**Threshold:** Remove feature pairs where $|r_{xy}| > 0.95$

---

### 📊 **Summary of When to Use Each Method**

| **Method** | **Use Case** | **Assumptions** | **Pros** | **Cons** |
|------------|--------------|-----------------|----------|----------|
| **Min-Max Scaling** | Bounded features, neural networks | Known min/max bounds | Preserves relationships | Sensitive to outliers |
| **Standardization** | Gaussian data, linear models | Normal distribution | Handles negative values | Doesn't bound values |
| **Robust Scaling** | Data with outliers | Heavy-tailed distributions | Outlier resistant | Less interpretable |
| **Box-Cox** | Right-skewed data | Positive values only | Normalizes distribution | Requires λ optimization |
| **Yeo-Johnson** | Any skewed data | No restrictions | Handles negative values | More complex |
| **Quantile Transform** | Non-parametric | No distributional assumptions | Very flexible | Loses interpretability |
