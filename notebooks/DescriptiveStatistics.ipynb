## Principal Themes of Descriptive Statistics

Descriptive statistics focus on summarizing and organizing data to make it easily understandable. The principal themes include:

### Measures of Central Tendency
- **Mean ($\mu$)**:

  $\mu = \frac{\sum_{i=1}^{n} x_i}{n}$

  where $x_i$ are the data points and $n$ is the number of data points.

- **Median**:
  - For an odd number of observations: The middle value when data is ordered.
  - For an even number of observations: The average of the two middle values.

- **Mode**: The value that appears most frequently in the data set.

### Measures of Variability (Dispersion)
- **Range**:
  $\text{Range} = x_{\text{max}} - x_{\text{min}}$
  where $x_{\text{max}}$ is the maximum value and $x_{\text{min}}$ is the minimum value.

- **Variance ($\sigma^2$)**:
  $\sigma^2 = \frac{\sum_{i=1}^{n} (x_i - \mu)^2}{n}$
  where $x_i$ are the data points, $\mu$ is the mean, and $n$ is the number of data points.

- **Standard Deviation ($\sigma$)**:
  $\sigma = \sqrt{\sigma^2} = \sqrt{\frac{\sum_{i=1}^{n} (x_i - \mu)^2}{n}}$

- **Interquartile Range (IQR)**:
  $\text{IQR} = Q_3 - Q_1$
  where $Q_3$ is the third quartile (75th percentile) and $Q_1$ is the first quartile (25th percentile).

### Shape of the Distribution
- **Skewness (Sk)**:
  $\text{Skewness} = \frac{\sum_{i=1}^{n} (x_i - \mu)^3}{n \cdot \sigma^3}$
  Positive skewness indicates a distribution with a longer tail on the right, while negative skewness indicates a longer tail on the left.

- **Kurtosis (K)**:
  $\text{Kurtosis} = \frac{\sum_{i=1}^{n} (x_i - \mu)^4}{n \cdot \sigma^4} - 3$
  Kurtosis greater than 3 indicates a distribution with heavier tails, while less than 3 indicates lighter tails compared to a normal distribution.

### Frequency Distribution
- **Frequency ($f$)**: The count of data points within a specified interval or category.

### Graphical Representation
- **Histogram**: A bar graph representing the frequency distribution of continuous data.
- **Bar Chart**: A graph with rectangular bars representing categorical data.
- **Pie Chart**: A circular graph divided into sectors representing proportions.
- **Box Plot**: A graphical depiction of the spread of data showing the median, quartiles, and potential outliers.
- **Scatter Plot**: A graph showing the relationship between two continuous variables.

These themes help in understanding the main characteristics of a dataset, making it easier to interpret and analyze data before applying more complex statistical methods.


```python
import numpy as np
import matplotlib.pyplot as plt

# Generate a dataset
np.random.seed(0)  # For reproducibility
data = np.random.normal(loc=50, scale=10, size=1000)  # Normal distribution with mean=50 and std=10

# Measures of Central Tendency
# Mean
mean = sum(data) / len(data)

# Median
sorted_data = sorted(data)
n = len(data)
if n % 2 == 1:
    median = sorted_data[n // 2]
else:
    median = (sorted_data[n // 2 - 1] + sorted_data[n // 2]) / 2

# Mode calculation (approximate mode for continuous data)
frequency = {}
for value in data:
    if value in frequency:
        frequency[value] += 1
    else:
        frequency[value] = 1
mode = max(frequency, key=frequency.get)

# Measures of Variability (Dispersion)
# Range
range_val = max(data) - min(data)

# Variance
mean_diff_squared = [(x - mean) ** 2 for x in data]
variance = sum(mean_diff_squared) / len(data)

# Standard Deviation
std_deviation = variance ** 0.5

# Interquartile Range (IQR)
Q1 = sorted_data[int(0.25 * (n + 1)) - 1]
Q3 = sorted_data[int(0.75 * (n + 1)) - 1]
IQR = Q3 - Q1

# Shape of the Distribution
# Skewness (Manually computed)
skewness_numerator = sum((x - mean) ** 3 for x in data)
skewness_denominator = (sum((x - mean) ** 2 for x in data) / len(data)) ** 1.5
skewness = skewness_numerator / (len(data) * skewness_denominator)

# Kurtosis (Manually computed)
kurtosis_numerator = sum((x - mean) ** 4 for x in data)
kurtosis_denominator = (sum((x - mean) ** 2 for x in data) / len(data)) ** 2
kurtosis = (kurtosis_numerator / len(data)) / kurtosis_denominator - 3

# Print the metrics
print(f"Mean: {mean}")
print(f"Median: {median}")
print(f"Mode: {mode}")
print(f"Range: {range_val}")
print(f"Variance: {variance}")
print(f"Standard Deviation: {std_deviation}")
print(f"Interquartile Range (IQR): {IQR}")
print(f"Skewness: {skewness}")
print(f"Kurtosis: {kurtosis}")

# Frequency Distribution
plt.hist(data, bins=30, edgecolor='k', alpha=0.7)
plt.title('Histogram')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()

# Box Plot
plt.boxplot(data, vert=False)
plt.title('Box Plot')
plt.xlabel('Value')
plt.show()

# Scatter Plot (using random scatter for illustration)
x = np.linspace(0, len(data) - 1, len(data))
y = data
plt.scatter(x, y, alpha=0.5)
plt.title('Scatter Plot')
plt.xlabel('Index')
plt.ylabel('Value')
plt.show()

# Pie Chart (using categorical data for illustration)
categories = ['A', 'B', 'C', 'D']
counts = [sum(value < 40 for value in data),
          sum(40 <= value < 50 for value in data),
          sum(50 <= value < 60 for value in data),
          sum(value >= 60 for value in data)]
plt.pie(counts, labels=categories, autopct='%1.1f%%', startangle=140)
plt.title('Pie Chart')
plt.show()
```
