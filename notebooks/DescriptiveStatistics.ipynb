    def _calculate_comprehensive_stats(self, data, name):
        """Calculate comprehensive statistical measures for a dataset"""

        data = np.array(data)
        n = len(data)

        # Basic descriptive statistics
        basic_stats = {
            'count': n,
            'mean': np.mean(data),
            'median': np.median(data),
            'mode': self._calculate_mode(data),
            'std': np.std(data, ddof=1),
            'var': np.var(data, ddof=1),
            'min': np.min(data),
            'max': np.max(data),
            'range': np.max(data) - np.min(data)
        }

        # Percentiles and quartiles
        percentiles = {
            'q1': np.percentile(data, 25),
            'q2': np.percentile(data, 50),  # Same as median
            'q3': np.percentile(data, 75),
            'iqr': np.percentile(data, 75) - np.percentile(data, 25),
            'p10': np.percentile(data, 10),
            'p90': np.percentile(data, 90)
        }

        # Shape measures
        shape_stats = {
            'skewness': skew(data),
            'kurtosis': kurtosis(data),
            'skewness_test_stat': stats.skewtest(data)[0] if n > 8 else np.nan,
            'skewness_test_p': stats.skewtest(data)[1] if n > 8 else np.nan,
            'kurtosis_test_stat': stats.kurtosistest(data)[0] if n > 20 else np.nan,
            'kurtosis_test_p': stats.kurtosistest(data)[1] if n > 20 else np.nan
        }

        # Variability measures
        variability_stats = {
            'cv': (basic_stats['std'] / abs(basic_stats['mean'])) * 100 if basic_stats['mean'] != 0 else np.nan,
            'mad_mean': np.mean(np.abs(data - basic_stats['mean'])),
            'mad_median': np.median(np.abs(data - basic_stats['median'])),
            'trimmed_mean_5': stats.trim_mean(data, 0.05),
            'trimmed_mean_10': stats.trim_mean(data, 0.10),
            'gini': self._calculate_gini(data) if np.all(data >= 0) else np.nan
        }

        # Normality tests
        normality_stats = {
            'shapiro_stat': stats.shapiro(data[:5000])[0] if n <= 5000 else np.nan,
            'shapiro_p': stats.shapiro(data[:5000])[1] if n <= 5000 else np.nan,
            'jarque_bera_stat': stats.jarque_bera(data)[0],
            'jarque_bera_p': stats.jarque_bera(data)[1],
            'anderson_stat': stats.anderson(data, dist='norm')[0],
            'anderson_critical_5': stats.anderson(data, dist='norm')[1][2]
        }

        # Outlier analysis
        outlier_stats = self._analyze_outliers(data)

        # Distribution classification
        distribution_type = self._classify_distribution(basic_stats, shape_stats, normality_stats)

        return {
            'dataset_name': name,
            'basic': basic_stats,
            'percentiles': percentiles,
            'shape': shape_stats,
            'variability': variability_stats,
            'normality': normality_stats,
            'outliers': outlier_stats,
            'classification': distribution_type
        }

    def _calculate_mode(self, data):
        """Calculate mode for continuous data using histogram approach"""
        if len(np.unique(data)) == len(data):  # All unique values
            hist, bin_edges = np.histogram(data, bins=30)
            modal_bin_index = np.argmax(hist)
            return (bin_edges[modal_bin_index] + bin_edges[modal_bin_index + 1]) / 2
        else:
            values, counts = np.unique(data, return_counts=True)
            return values[np.argmax(counts)]

    def _calculate_gini(self, data):
        """Calculate Gini coefficient"""
        data = np.array(data)
        if np.amin(data) < 0:
            data = data - np.amin(data)  # Shift to non-negative

        sorted_data = np.sort(data)
        n = len(data)
        cumsum = np.cumsum(sorted_data)

        if cumsum[-1] == 0:
            return 0

        return (2 * np.sum((np.arange(1, n+1) * sorted_data))) / (n * cumsum[-1]) - (n + 1) / n

    def _analyze_outliers(self, data):
        """Comprehensive outlier analysis"""

        # IQR method
        q1 = np.percentile(data, 25)
        q3 = np.percentile(data, 75)
        iqr = q3 - q1
        lower_fence = q1 - 1.5 * iqr
        upper_fence = q3 + 1.5 * iqr

        iqr_outliers = np.sum((data < lower_fence) | (data > upper_fence))

        # Z-score method (|z| > 3)
        if np.std(data) > 0:
            z_scores = np.abs((data - np.mean(data)) / np.std(data))
            z_outliers = np.sum(z_scores > 3)
        else:
            z_outliers = 0

        # Modified Z-score method using MAD
        mad = np.median(np.abs(data - np.median(data)))
        if mad > 0:
            modified_z_scores = 0.6745 * (data - np.median(data)) / mad
            mad_outliers = np.sum(np.abs(modified_z_scores) > 3.5)
        else:
            mad_outliers = 0

        return {
            'iqr_outliers': iqr_outliers,
            'iqr_percentage': (iqr_outliers / len(data)) * 100,
            'z_outliers': z_outliers,
            'z_percentage': (z_outliers / len(data)) * 100,
            'mad_outliers': mad_outliers,
            'mad_percentage': (mad_outliers / len(data)) * 100,
            'lower_fence': lower_fence,
            'upper_fence': upper_fence
        }

    def _classify_distribution(self, basic_stats, shape_stats, normality_stats):
        """Classify distribution type based on statistical properties"""

        skewness = shape_stats['skewness']
        kurtosis = shape_stats['kurtosis']
        jb_p = normality_stats['jarque_bera_p']

        # Normality assessment
        is_normal = jb_p > 0.05

        # Skewness classification
        if abs(skewness) < 0.5:
            skew_type = "symmetric"
        elif skewness > 0.5:
            skew_type = "right_skewed"
        else:
            skew_type = "left_skewed"

        # Kurtosis classification
        if kurtosis < -1:
            kurt_type = "platykurtic"  # Light tails
        elif kurtosis > 1:
            kurt_type = "leptokurtic"  # Heavy tails
        else:
            kurt_type = "mesokurtic"   # Normal tails

        # Overall classification
        if is_normal:
            overall_type = "approximately_normal"
        elif abs(skewness) > 1:
            overall_type = "highly_skewed"
        elif abs(kurtosis) > 2:
            overall_type = "extreme_kurtosis"
        else:
            overall_type = "non_normal"

        return {
            'is_normal': is_normal,
            'skew_type': skew_type,
            'kurtosis_type': kurt_type,
            'overall_type': overall_type
        }

    def _generate_comparison_report(self, stats_results):

        # Create comparison table
        comparison_data = []

            stats = stats_results[name]

                'Mean': f"{stats['basic']['mean']:.3f}",
                'Median': f"{stats['basic']['median']:.3f}",
                'Std Dev': f"{stats['basic']['std']:.3f}",
                'CV (%)': f"{stats['variability']['cv']:.1f}" if not np.isnan(stats['variability']['cv']) else "N/A",
                'Skewness': f"{stats['shape']['skewness']:.3f}",
                'Kurtosis': f"{stats['shape']['kurtosis']:.3f}",
                'JB p-value': f"{stats['normality']['jarque_bera_p']:.4f}",
                'Distribution': stats['classification']['overall_type']
            })

        comparison_df = pd.DataFrame(comparison_data)
        print(comparison_df.to_string(index=False))

        # Distribution classifications
        print(f"\n🔍 DISTRIBUTION CLASSIFICATIONS:")
        print("-" * 40)
        for name in dataset_names:
            classification = stats_results[name]['classification']
            print(f"{name}:")
            print(f"  • Overall Type: {classification['overall_type'].replace('_', ' ').title()}")

            print(f"  • Kurtosis: {classification['kurtosis_type'].replace('_', ' ').title()}")

        # Variability ranking

    def _perform_statistical_tests(self, datasets_dict):
        """Perform statistical tests for comparison"""

        print(f"\n🧪 STATISTICAL TESTS")
        print("-" * 30)

        dataset_names = list(datasets_dict.keys())

        if len(dataset_names) >= 2:
            # Pairwise comparisons
            for i in range(len(dataset_names)):
                for j in range(i+1, len(dataset_names)):
                    name1, name2 = dataset_names[i], dataset_names[j]
                    data1, data2 = datasets_dict[name1], datasets_dict[name2]

                    try:
                        # Mann-Whitney U test (non-parametric)
                        mw_stat, mw_p = stats.mannwhitneyu(data1, data2, alternative='two-sided')
                        print(f"\n{name1} vs {name2}:")
                        print(f"  Mann-Whitney U test: p-value = {mw_p:.4f}")

                        # Kolmogorov-Smirnov test
                        ks_stat, ks_p = stats.ks_2samp(data1, data2)
                        print(f"  Kolmogorov-Smirnov test: p-value = {ks_p:.4f}")

                        # Effect size (Cohen's d)
                        pooled_std = np.sqrt(((len(data1)-1)*np.var(data1, ddof=1) + (len(data2)-1)*np.var(data2, ddof=1)) / (len(data1)+len(data2)-2))
                        if pooled_std > 0:
                            cohens_d = (np.mean(data1) - np.mean(data2)) / pooled_std
                            print(f"  Cohen's d (effect size): {cohens_d:.3f}")

                    except Exception as e:
                        print(f"  Error in statistical tests: {str(e)}")

```python
class StatisticalInterpretationGuide:
    """
    Comprehensive guide for interpreting descriptive statistics in context
    """

    def __init__(self):
        """Initialize the interpretation guide"""
        self.interpretation_frameworks = {}
        self.domain_specific_guidelines = {}

    def interpret_central_tendency(self, mean, median, mode, context="general"):
        """
        Provide contextual interpretation of central tendency measures
        """

        interpretation = {
            'measures': {'mean': mean, 'median': median, 'mode': mode},
            'relationships': {},
            'practical_implications': {},
            'recommendations': []
        }

        # Analyze relationships between measures
        mean_median_diff = abs(mean - median)

        interpretation['relationships'] = {
            'mean_median_difference': mean_median_diff
        }

        # Determine distribution characteristics
        if mean_median_diff < 0.1 * abs(mean) if mean != 0 else mean_median_diff < 0.1:
            distribution_shape = "approximately_symmetric"
            interpretation['practical_implications']['distribution'] = (
                "Data is approximately symmetric around the center. "
                "Mean and median are reliable measures of central tendency."
            )
        elif mean > median:
            distribution_shape = "right_skewed"
            interpretation['practical_implications']['distribution'] = (
                "Data is right-skewed with a longer tail on the upper end. "
                "Median is more representative of typical values than mean."
            )
        else:
            distribution_shape = "left_skewed"
            interpretation['practical_implications']['distribution'] = (
                "Data is left-skewed with a longer tail on the lower end. "
                "Median is more representative of typical values than mean."
            )

        # Context-specific interpretations
        if context == "financial":
            if distribution_shape == "right_skewed":
                interpretation['practical_implications']['financial'] = (
                    "Typical returns are lower than average, with occasional large gains. "
                    "Consider median for conservative estimates."
                )
        elif context == "medical":
            if distribution_shape == "right_skewed":
                interpretation['practical_implications']['medical'] = (
                    "Most patients have shorter treatment times, with some requiring extended care. "
                    "Plan resources based on median, prepare for outliers."
                )

        # Generate recommendations
        interpretation['recommendations'] = self._generate_central_tendency_recommendations(
            mean, median, mode, distribution_shape, context
        )

        return interpretation

    def interpret_variability(self, std, var, cv, iqr, range_val, context="general"):
        """
        Provide contextual interpretation of variability measures
        """

        interpretation = {
            'measures': {
                'standard_deviation': std,
                'variance': var,
                'coefficient_variation': cv,
                'iqr': iqr,
                'range': range_val
            },
            'variability_level': {},
            'practical_implications': {},
            'recommendations': []
        }

        # Assess variability level
        if cv < 15:
            variability_level = "low"
            variability_desc = "Data points are closely clustered around the mean"
        elif cv < 30:
            variability_level = "moderate"
            variability_desc = "Data shows moderate dispersion"
        else:
            variability_level = "high"
            variability_desc = "Data points are widely dispersed"

        interpretation['variability_level'] = {
            'level': variability_level,
            'description': variability_desc,
            'cv_percent': cv
        }

        # Context-specific interpretations
        if context == "quality_control":
            if variability_level == "low":
                interpretation['practical_implications']['quality'] = (
                    "Process is well-controlled with consistent output. "
                    "Good quality control maintained."
                )
            elif variability_level == "high":
                interpretation['practical_implications']['quality'] = (
                    "Process shows excessive variation. "
                    "Investigation and process improvement needed."
                )
        elif context == "financial":
            if variability_level == "high":
                interpretation['practical_implications']['financial'] = (
                    "High volatility indicates significant risk. "
                    "Consider risk management strategies."
                )

        # Robustness assessment
        robust_ratio = iqr / (1.349 * std) if std > 0 else 0  # For normal distribution, this should be ~1
        if 0.8 <= robust_ratio <= 1.2:
            interpretation['practical_implications']['robustness'] = (
                "IQR and standard deviation are consistent, suggesting few outliers."
            )
        elif robust_ratio < 0.8:
            interpretation['practical_implications']['robustness'] = (
                "Standard deviation is inflated by outliers. "
                "IQR may be more representative of typical variability."
            )

        # Generate recommendations
        interpretation['recommendations'] = self._generate_variability_recommendations(
            std, cv, variability_level, context
        )

        return interpretation

    def interpret_distribution_shape(self, skewness, kurtosis, normality_p_value, context="general"):
        """
        Provide interpretation of distribution shape characteristics
        """

        interpretation = {
            'shape_measures': {
                'skewness': skewness,
                'kurtosis': kurtosis,
                'normality_p_value': normality_p_value
            },
            'shape_classification': {},
            'practical_implications': {},
            'statistical_implications': {},
            'recommendations': []
        }

        # Classify skewness
        if abs(skewness) < 0.5:
            skew_classification = "approximately_symmetric"
            skew_description = "Distribution is approximately symmetric"
        elif skewness > 0.5:
            skew_classification = "right_skewed"
            skew_description = "Distribution has a longer right tail"
        else:
            skew_classification = "left_skewed"
            skew_description = "Distribution has a longer left tail"

        # Classify kurtosis (excess kurtosis)
        excess_kurtosis = kurtosis
        if abs(excess_kurtosis) < 0.5:
            kurt_classification = "mesokurtic"
            kurt_description = "Normal tail behavior"
        elif excess_kurtosis > 0.5:
            kurt_classification = "leptokurtic"
            kurt_description = "Heavy tails, more extreme values"
        else:
            kurt_classification = "platykurtic"
            kurt_description = "Light tails, fewer extreme values"

        # Normality assessment
        is_normal = normality_p_value > 0.05

        interpretation['shape_classification'] = {
            'skewness_type': skew_classification,
            'skewness_description': skew_description,
            'kurtosis_type': kurt_classification,
            'kurtosis_description': kurt_description,
            'is_approximately_normal': is_normal
        }

        # Statistical implications
        if not is_normal:
            interpretation['statistical_implications']['parametric_tests'] = (
                "Non-normal distribution may violate assumptions of parametric tests. "
                "Consider non-parametric alternatives or data transformation."
            )

        if abs(skewness) > 1:
            interpretation['statistical_implications']['outliers'] = (
                "High skewness may indicate presence of outliers or extreme values."
            )

        if abs(excess_kurtosis) > 2:
            interpretation['statistical_implications']['tail_risk'] = (
                "Extreme kurtosis indicates unusual tail behavior. "
                "Standard risk models may underestimate extreme events."
            )

        # Generate recommendations
        interpretation['recommendations'] = self._generate_shape_recommendations(
            skewness, kurtosis, is_normal, context
        )

        return interpretation

    def interpret_outliers(self, outlier_count, total_count, outlier_percentage, detection_method, context="general"):
        """
        Provide interpretation of outlier analysis results
        """

        interpretation = {
            'outlier_summary': {
                'count': outlier_count,
                'percentage': outlier_percentage,
                'detection_method': detection_method,
                'total_observations': total_count
            },
            'severity_assessment': {},
            'potential_causes': [],
            'practical_implications': {},
            'recommendations': []
        }

        # Assess outlier severity
        if outlier_percentage < 2:
            severity = "minimal"
            severity_desc = "Few outliers present, within normal range"
        elif outlier_percentage < 5:
            severity = "moderate"
            severity_desc = "Moderate number of outliers, worth investigating"
        else:
            severity = "high"
            severity_desc = "High number of outliers, requires attention"

        interpretation['severity_assessment'] = {
            'level': severity,
            'description': severity_desc
        }

        # Generate recommendations
        interpretation['recommendations'] = self._generate_outlier_recommendations(
            outlier_percentage, severity, context
        )

        return interpretation

    def create_comprehensive_report(self, data, context="general", dataset_name="Dataset"):
        """
        Create a comprehensive interpretation report for a dataset
        """

        print(f"📋 COMPREHENSIVE STATISTICAL INTERPRETATION REPORT")
        print(f"Dataset: {dataset_name}")
        print(f"Context: {context.title()}")
        print("=" * 80)

        # Calculate basic statistics
        mean = np.mean(data)
        median = np.median(data)
        try:
            mode_result = stats.mode(data, keepdims=True)
            mode = mode_result[0][0] if len(mode_result[0]) > 0 else None
        except:
            mode = None

        std = np.std(data, ddof=1)
        var = np.var(data, ddof=1)
        cv = (std / abs(mean)) * 100 if mean != 0 else np.nan

        q1 = np.percentile(data, 25)
        q3 = np.percentile(data, 75)
        iqr = q3 - q1
        range_val = np.max(data) - np.min(data)

        skewness_val = skew(data)
        kurtosis_val = kurtosis(data)
        _, normality_p = stats.jarque_bera(data)

        # Outlier analysis
        lower_fence = q1 - 1.5 * iqr
        upper_fence = q3 + 1.5 * iqr
        outliers = np.sum((data < lower_fence) | (data > upper_fence))
        outlier_pct = (outliers / len(data)) * 100

        # Generate interpretations
        central_interp = self.interpret_central_tendency(mean, median, mode, context)
        variability_interp = self.interpret_variability(std, var, cv, iqr, range_val, context)
        shape_interp = self.interpret_distribution_shape(skewness_val, kurtosis_val, normality_p, context)
        outlier_interp = self.interpret_outliers(outliers, len(data), outlier_pct, "IQR", context)

        # Print sections
        self._print_interpretation_section("CENTRAL TENDENCY ANALYSIS", central_interp)
        self._print_interpretation_section("VARIABILITY ANALYSIS", variability_interp)
        self._print_interpretation_section("DISTRIBUTION SHAPE ANALYSIS", shape_interp)
        self._print_interpretation_section("OUTLIER ANALYSIS", outlier_interp)

        # Overall assessment
        print(f"\n🎯 OVERALL ASSESSMENT")
        print("-" * 40)

        # Combine recommendations from all sections
        all_recommendations = (
            central_interp['recommendations'] +
            variability_interp['recommendations'] +
            shape_interp['recommendations'] +
            outlier_interp['recommendations']
        )

        # Prioritize and deduplicate recommendations
        priority_recommendations = list(set(all_recommendations))

        for i, rec in enumerate(priority_recommendations[:5], 1):  # Top 5 recommendations
            print(f"{i}. {rec}")

        return {
            'central_tendency': central_interp,
            'variability': variability_interp,
            'shape': shape_interp,
            'outliers': outlier_interp,
            'overall_recommendations': priority_recommendations
        }

    def _generate_central_tendency_recommendations(self, mean, median, mode, distribution_shape, context):
        """Generate recommendations based on central tendency analysis"""
        recommendations = []

        if distribution_shape == "right_skewed":
            recommendations.append("Use median rather than mean for typical value reporting")
            recommendations.append("Consider log transformation to reduce right skewness")
        elif distribution_shape == "left_skewed":
            recommendations.append("Use median for robust central tendency measure")
            recommendations.append("Investigate causes of left tail values")

        return recommendations

    def _generate_variability_recommendations(self, std, cv, variability_level, context):
        """Generate recommendations based on variability analysis"""
        recommendations = []

        if variability_level == "high":
            recommendations.append("Investigate sources of high variability")
            recommendations.append("Consider using robust statistical methods")

        if cv > 100:
            recommendations.append("Extremely high coefficient of variation suggests need for data review")

        return recommendations

    def _generate_shape_recommendations(self, skewness, kurtosis, is_normal, context):
        """Generate recommendations based on distribution shape"""
        recommendations = []

        if not is_normal:
            recommendations.append("Consider non-parametric statistical tests for analysis")
            recommendations.append("Evaluate data transformation options (log, Box-Cox)")

        if abs(skewness) > 2:
            recommendations.append("Extreme skewness detected - investigate data quality")

        return recommendations

    def _generate_outlier_recommendations(self, outlier_percentage, severity, context):
        """Generate recommendations based on outlier analysis"""
        recommendations = []

        if severity == "high":
            recommendations.append("Investigate high outlier rate - may indicate data quality issues")
            recommendations.append("Consider separate analysis of outliers vs main data")

        return recommendations

    def _print_interpretation_section(self, title, interpretation):
        """Print a formatted interpretation section"""

        print(f"\n📊 {title}")
        print("-" * len(title))

        # Print practical implications
        if 'practical_implications' in interpretation:
            for key, value in interpretation['practical_implications'].items():
                print(f"• {key.replace('_', ' ').title()}: {value}")

        # Print statistical implications
        if 'statistical_implications' in interpretation:
            for key, value in interpretation['statistical_implications'].items():
                print(f"• {key.replace('_', ' ').title()}: {value}")

        # Print key recommendations for this section
        if 'recommendations' in interpretation and interpretation['recommendations']:
            print(f"\n💡 Key Recommendations:")
            for rec in interpretation['recommendations'][:3]:  # Top 3 per section
                print(f"  - {rec}")

# Initialize the interpretation guide
guide = StatisticalInterpretationGuide()
print("✅ Statistical Interpretation Guide initialized!")
```

```python
class DescriptiveStatisticsComparator:
    """
    Advanced framework for comparing descriptive statistics across different scenarios
    FIXED: All matplotlib and scipy compatibility issues resolved
    """

    def __init__(self):
        """Initialize the comparator"""
        self.comparison_results = {}

    def compare_distributions(self, datasets_dict, figsize=(20, 15)):
        """
        Compare multiple datasets using comprehensive statistical measures
        """

        print("📊 COMPREHENSIVE DISTRIBUTION COMPARISON")
        print("=" * 60)

        # Calculate comprehensive statistics for each dataset
        stats_results = {}
        for name, data in datasets_dict.items():
            stats_results[name] = self._calculate_comprehensive_stats(data, name)

        # Create comparison visualizations
        self._create_comparison_visualizations(datasets_dict, stats_results, figsize)

        # Generate comparison report
        self._generate_comparison_report(stats_results)

        # Perform statistical tests
        self._perform_statistical_tests(datasets_dict)

        self.comparison_results = stats_results
        return stats_results

    def _calculate_comprehensive_stats(self, data, name):
        """Calculate comprehensive statistical measures for a dataset"""

        data = np.array(data)
        n = len(data)

        # Basic descriptive statistics
        basic_stats = {
            'count': n,
            'mean': np.mean(data),
            'median': np.median(data),
            'mode': self._calculate_mode(data),
            'std': np.std(data, ddof=1),
            'var': np.var(data, ddof=1),
            'min': np.min(data),
            'max': np.max(data),
            'range': np.max(data) - np.min(data)
        }

        # Percentiles and quartiles
        percentiles = {
            'q1': np.percentile(data, 25),
            'q2': np.percentile(data, 50),  # Same as median
            'q3': np.percentile(data, 75),
            'iqr': np.percentile(data, 75) - np.percentile(data, 25),
            'p10': np.percentile(data, 10),
            'p90': np.percentile(data, 90)
        }

        # Shape measures
        shape_stats = {
            'skewness': skew(data),
            'kurtosis': kurtosis(data),
            'skewness_test_stat': stats.skewtest(data)[0] if n > 8 else np.nan,
            'skewness_test_p': stats.skewtest(data)[1] if n > 8 else np.nan,
            'kurtosis_test_stat': stats.kurtosistest(data)[0] if n > 20 else np.nan,
            'kurtosis_test_p': stats.kurtosistest(data)[1] if n > 20 else np.nan
        }

        # Variability measures
        variability_stats = {
            'cv': (basic_stats['std'] / abs(basic_stats['mean'])) * 100 if basic_stats['mean'] != 0 else np.nan,
            'mad_mean': np.mean(np.abs(data - basic_stats['mean'])),
            'mad_median': np.median(np.abs(data - basic_stats['median'])),
            'trimmed_mean_5': stats.trim_mean(data, 0.05),
            'trimmed_mean_10': stats.trim_mean(data, 0.10),
            'gini': self._calculate_gini(data) if np.all(data >= 0) else np.nan
        }

        # Normality tests
        normality_stats = {
            'shapiro_stat': stats.shapiro(data[:5000])[0] if n <= 5000 else np.nan,
            'shapiro_p': stats.shapiro(data[:5000])[1] if n <= 5000 else np.nan,
            'jarque_bera_stat': stats.jarque_bera(data)[0],
            'jarque_bera_p': stats.jarque_bera(data)[1],
            'anderson_stat': stats.anderson(data, dist='norm')[0],
            'anderson_critical_5': stats.anderson(data, dist='norm')[1][2]
        }

        # Outlier analysis
        outlier_stats = self._analyze_outliers(data)

        # Distribution classification
        distribution_type = self._classify_distribution(basic_stats, shape_stats, normality_stats)

        return {
            'dataset_name': name,
            'basic': basic_stats,
            'percentiles': percentiles,
            'shape': shape_stats,
            'variability': variability_stats,
            'normality': normality_stats,
            'outliers': outlier_stats,
            'classification': distribution_type
        }

    def _calculate_mode(self, data):
        """Calculate mode for continuous data using histogram approach"""
        if len(np.unique(data)) == len(data):  # All unique values
            hist, bin_edges = np.histogram(data, bins=30)
            modal_bin_index = np.argmax(hist)
            return (bin_edges[modal_bin_index] + bin_edges[modal_bin_index + 1]) / 2
        else:
            values, counts = np.unique(data, return_counts=True)
            return values[np.argmax(counts)]

    def _calculate_gini(self, data):
        """Calculate Gini coefficient"""
        data = np.array(data)
        if np.amin(data) < 0:
            data = data - np.amin(data)  # Shift to non-negative

        sorted_data = np.sort(data)
        n = len(data)
        cumsum = np.cumsum(sorted_data)

        if cumsum[-1] == 0:
            return 0

        return (2 * np.sum((np.arange(1, n+1) * sorted_data))) / (n * cumsum[-1]) - (n + 1) / n

    def _analyze_outliers(self, data):
        """Comprehensive outlier analysis"""

        # IQR method
        q1 = np.percentile(data, 25)
        q3 = np.percentile(data, 75)
        iqr = q3 - q1
        lower_fence = q1 - 1.5 * iqr
        upper_fence = q3 + 1.5 * iqr

        iqr_outliers = np.sum((data < lower_fence) | (data > upper_fence))

        # Z-score method (|z| > 3)
        if np.std(data) > 0:
            z_scores = np.abs((data - np.mean(data)) / np.std(data))
            z_outliers = np.sum(z_scores > 3)
        else:
            z_outliers = 0

        # Modified Z-score method using MAD
        mad = np.median(np.abs(data - np.median(data)))
        if mad > 0:
            modified_z_scores = 0.6745 * (data - np.median(data)) / mad
            mad_outliers = np.sum(np.abs(modified_z_scores) > 3.5)
        else:
            mad_outliers = 0

        return {
            'iqr_outliers': iqr_outliers,
            'iqr_percentage': (iqr_outliers / len(data)) * 100,
            'z_outliers': z_outliers,
            'z_percentage': (z_outliers / len(data)) * 100,
            'mad_outliers': mad_outliers,
            'mad_percentage': (mad_outliers / len(data)) * 100,
            'lower_fence': lower_fence,
            'upper_fence': upper_fence
        }

    def _classify_distribution(self, basic_stats, shape_stats, normality_stats):
        """Classify distribution type based on statistical properties"""

        skewness = shape_stats['skewness']
        kurtosis = shape_stats['kurtosis']
        jb_p = normality_stats['jarque_bera_p']

        # Normality assessment
        is_normal = jb_p > 0.05

        # Skewness classification
        if abs(skewness) < 0.5:
            skew_type = "symmetric"
        elif skewness > 0.5:
            skew_type = "right_skewed"
        else:
            skew_type = "left_skewed"

        # Kurtosis classification
        if kurtosis < -1:
            kurt_type = "platykurtic"  # Light tails
        elif kurtosis > 1:
            kurt_type = "leptokurtic"  # Heavy tails
        else:
            kurt_type = "mesokurtic"   # Normal tails

        # Overall classification
        if is_normal:
            overall_type = "approximately_normal"
        elif abs(skewness) > 1:
            overall_type = "highly_skewed"
        elif abs(kurtosis) > 2:
            overall_type = "extreme_kurtosis"
        else:
            overall_type = "non_normal"

        return {
            'is_normal': is_normal,
            'skew_type': skew_type,
            'kurtosis_type': kurt_type,
            'overall_type': overall_type
        }

    def _create_comparison_visualizations(self, datasets_dict, stats_results, figsize):
        """Create comprehensive comparison visualizations - FIXED VERSION"""

        fig, axes = plt.subplots(3, 4, figsize=figsize)
        fig.suptitle('Comprehensive Statistical Comparison', fontsize=16, fontweight='bold')

        dataset_names = list(datasets_dict.keys())
        colors = plt.cm.Set3(np.linspace(0, 1, len(dataset_names)))

        # 1. Overlaid histograms
        ax1 = axes[0, 0]
        for i, (name, data) in enumerate(datasets_dict.items()):
            ax1.hist(data, bins=30, alpha=0.6, color=colors[i], label=name, density=True)
        ax1.set_title('Distribution Comparison')
        ax1.set_xlabel('Value')
        ax1.set_ylabel('Density')
        ax1.legend()
        ax1.grid(True, alpha=0.3)

        # 2. Box plots - FIXED: No more deprecation warning
        ax2 = axes[0, 1]
        data_list = [data for data in datasets_dict.values()]
        box_plot = ax2.boxplot(data_list, patch_artist=True)
        for patch, color in zip(box_plot['boxes'], colors):
            patch.set_facecolor(color)
            patch.set_alpha(0.7)
        ax2.set_title('Box Plot Comparison')
        ax2.set_xticklabels(dataset_names, rotation=45)
        ax2.grid(True, alpha=0.3)

        # 3. Q-Q plots against normal - FIXED: No more TypeError or ValueError
        ax3 = axes[0, 2]
        for i, (name, data) in enumerate(datasets_dict.items()):
            # Generate theoretical and sample quantiles
            theoretical_quantiles, sample_quantiles = stats.probplot(data, dist="norm", fit=False)
            # Plot the points using scatter
            ax3.scatter(theoretical_quantiles, sample_quantiles, color=colors[i], alpha=0.6, s=20, label=name)

        ax3.set_title('Q-Q Plots vs Normal')
        ax3.set_xlabel('Theoretical Quantiles')
        ax3.set_ylabel('Sample Quantiles')
        ax3.grid(True, alpha=0.3)
        ax3.legend()

        # 4. Violin plots
        ax4 = axes[0, 3]
        parts = ax4.violinplot(data_list, positions=range(1, len(dataset_names)+1), widths=0.8)
        for i, part in enumerate(parts['bodies']):
            part.set_facecolor(colors[i])
            part.set_alpha(0.7)
        ax4.set_xticks(range(1, len(dataset_names)+1))
        ax4.set_xticklabels(dataset_names, rotation=45)
        ax4.set_title('Violin Plot Comparison')
        ax4.grid(True, alpha=0.3)

        # 5. Central tendency comparison
        ax5 = axes[1, 0]
        means = [stats_results[name]['basic']['mean'] for name in dataset_names]
        medians = [stats_results[name]['basic']['median'] for name in dataset_names]
        modes = [stats_results[name]['basic']['mode'] for name in dataset_names]

        x = np.arange(len(dataset_names))
        width = 0.25

        ax5.bar(x - width, means, width, label='Mean', alpha=0.8, color='red')
        ax5.bar(x, medians, width, label='Median', alpha=0.8, color='green')
        ax5.bar(x + width, modes, width, label='Mode', alpha=0.8, color='blue')

        ax5.set_title('Central Tendency Comparison')
        ax5.set_xticks(x)
        ax5.set_xticklabels(dataset_names, rotation=45)
        ax5.legend()
        ax5.grid(True, alpha=0.3)

        # 6. Variability measures
        ax6 = axes[1, 1]
        stds = [stats_results[name]['basic']['std'] for name in dataset_names]
        iqrs = [stats_results[name]['percentiles']['iqr'] for name in dataset_names]
        ranges = [stats_results[name]['basic']['range'] for name in dataset_names]

        x = np.arange(len(dataset_names))
        ax6.bar(x - width, stds, width, label='Std Dev', alpha=0.8, color='orange')
        ax6.bar(x, iqrs, width, label='IQR', alpha=0.8, color='purple')
        ax6.bar(x + width, ranges, width, label='Range', alpha=0.8, color='brown')

        ax6.set_title('Variability Measures')
        ax6.set_xticks(x)
        ax6.set_xticklabels(dataset_names, rotation=45)
        ax6.legend()
        ax6.grid(True, alpha=0.3)

        # 7. Coefficient of Variation
        ax7 = axes[1, 2]
        cvs = [stats_results[name]['variability']['cv'] for name in dataset_names]
        bars = ax7.bar(range(len(dataset_names)), cvs, color=colors, alpha=0.8)
        ax7.set_title('Coefficient of Variation')
        ax7.set_xticks(range(len(dataset_names)))
        ax7.set_xticklabels(dataset_names, rotation=45)
        ax7.set_ylabel('CV (%)')
        ax7.grid(True, alpha=0.3)

        # Add value labels
        for bar, cv in zip(bars, cvs):
            if not np.isnan(cv):
                height = bar.get_height()
                ax7.text(bar.get_x() + bar.get_width()/2., height,
                        f'{cv:.1f}%', ha='center', va='bottom')

        # 8. Skewness vs Kurtosis scatter
        ax8 = axes[1, 3]
        skewnesses = [stats_results[name]['shape']['skewness'] for name in dataset_names]
        kurtoses = [stats_results[name]['shape']['kurtosis'] for name in dataset_names]

        for i, name in enumerate(dataset_names):
            ax8.scatter(skewnesses[i], kurtoses[i], color=colors[i], s=200, alpha=0.8, label=name)
            ax8.annotate(name, (skewnesses[i], kurtoses[i]),
                        xytext=(5, 5), textcoords='offset points', fontsize=8)

        ax8.axvline(0, color='gray', linestyle='--', alpha=0.5)
        ax8.axhline(0, color='gray', linestyle='--', alpha=0.5)
        ax8.set_title('Skewness vs Kurtosis')
        ax8.set_xlabel('Skewness')
        ax8.set_ylabel('Kurtosis')
        ax8.grid(True, alpha=0.3)

        # 9. Outlier comparison
        ax9 = axes[2, 0]
        iqr_outliers = [stats_results[name]['outliers']['iqr_percentage'] for name in dataset_names]
        z_outliers = [stats_results[name]['outliers']['z_percentage'] for name in dataset_names]
        mad_outliers = [stats_results[name]['outliers']['mad_percentage'] for name in dataset_names]

        x = np.arange(len(dataset_names))
        ax9.bar(x - width, iqr_outliers, width, label='IQR Method', alpha=0.8, color='red')
        ax9.bar(x, z_outliers, width, label='Z-Score Method', alpha=0.8, color='blue')
        ax9.bar(x + width, mad_outliers, width, label='MAD Method', alpha=0.8, color='green')

        ax9.set_title('Outlier Detection Comparison')
        ax9.set_xticks(x)
        ax9.set_xticklabels(dataset_names, rotation=45)
        ax9.set_ylabel('Outlier Percentage (%)')
        ax9.legend()
        ax9.grid(True, alpha=0.3)

        # 10. Normality test p-values
        ax10 = axes[2, 1]
        jb_p_values = [stats_results[name]['normality']['jarque_bera_p'] for name in dataset_names]
        shapiro_p_values = [stats_results[name]['normality']['shapiro_p'] for name in dataset_names]

        x = np.arange(len(dataset_names))
        ax10.bar(x - width/2, jb_p_values, width, label='Jarque-Bera', alpha=0.8, color='cyan')
        ax10.bar(x + width/2, shapiro_p_values, width, label='Shapiro-Wilk', alpha=0.8, color='magenta')

        ax10.axhline(0.05, color='red', linestyle='--', alpha=0.7, label='α = 0.05')
        ax10.set_title('Normality Test P-values')
        ax10.set_xticks(x)
        ax10.set_xticklabels(dataset_names, rotation=45)
        ax10.set_ylabel('P-value')
        ax10.legend()
        ax10.grid(True, alpha=0.3)
        ax10.set_ylim(0, 1)

        # 11. Percentile comparison
        ax11 = axes[2, 2]
        percentiles = ['p10', 'q1', 'q2', 'q3', 'p90']
        for i, name in enumerate(dataset_names):
            percentile_values = [stats_results[name]['percentiles'][p] for p in percentiles]
            ax11.plot(percentiles, percentile_values, 'o-', color=colors[i],
                     label=name, linewidth=2, markersize=6)

        ax11.set_title('Percentile Comparison')
        ax11.set_xlabel('Percentile')
        ax11.set_ylabel('Value')
        ax11.legend()
        ax11.grid(True, alpha=0.3)
        ax11.tick_params(axis='x', rotation=45)

        # 12. Summary statistics heatmap
        ax12 = axes[2, 3]

        # Create heatmap data
        heatmap_metrics = ['Mean', 'Std Dev', 'Skewness', 'Kurtosis', 'CV']
        heatmap_data = []

        for metric in heatmap_metrics:
            if metric == 'Mean':
                values = [stats_results[name]['basic']['mean'] for name in dataset_names]
            elif metric == 'Std Dev':
                values = [stats_results[name]['basic']['std'] for name in dataset_names]
            elif metric == 'Skewness':
                values = [abs(stats_results[name]['shape']['skewness']) for name in dataset_names]
            elif metric == 'Kurtosis':
                values = [abs(stats_results[name]['shape']['kurtosis']) for name in dataset_names]
            elif metric == 'CV':
                values = [abs(stats_results[name]['variability']['cv']) if not np.isnan(stats_results[name]['variability']['cv']) else 0 for name in dataset_names]

            # Normalize to 0-1 scale
            if max(values) != min(values):
                normalized = [(v - min(values)) / (max(values) - min(values)) for v in values]
            else:
                normalized = [0.5] * len(values)
            heatmap_data.append(normalized)

        heatmap_array = np.array(heatmap_data)
        im = ax12.imshow(heatmap_array, cmap='YlOrRd', aspect='auto')

        ax12.set_xticks(range(len(dataset_names)))
        ax12.set_yticks(range(len(heatmap_metrics)))
        ax12.set_xticklabels(dataset_names, rotation=45)
        ax12.set_yticklabels(heatmap_metrics)
        ax12.set_title('Normalized Statistics Heatmap')

        plt.colorbar(im, ax=ax12, shrink=0.8)

        plt.tight_layout()
        plt.show()

    def _generate_comparison_report(self, stats_results):
        """Generate comprehensive comparison report"""

        print("\n📋 DETAILED COMPARISON REPORT")
        print("=" * 80)

        dataset_names = list(stats_results.keys())

        # Create comparison table
        comparison_data = []
        for name in dataset_names:
            stats = stats_results[name]
            comparison_data.append({
                'Dataset': name,
                'Count': stats['basic']['count'],
                'Mean': f"{stats['basic']['mean']:.3f}",
                'Median': f"{stats['basic']['median']:.3f}",
                'Std Dev': f"{stats['basic']['std']:.3f}",
                'CV (%)': f"{stats['variability']['cv']:.1f}" if not np.isnan(stats['variability']['cv']) else "N/A",
                'Skewness': f"{stats['shape']['skewness']:.3f}",
                'Kurtosis': f"{stats['shape']['kurtosis']:.3f}",
                'JB p-value': f"{stats['normality']['jarque_bera_p']:.4f}",
                'Distribution': stats['classification']['overall_type']
            })

        comparison_df = pd.DataFrame(comparison_data)
        print(comparison_df.to_string(index=False))

        # Distribution classifications
        print(f"\n🔍 DISTRIBUTION CLASSIFICATIONS:")
        print("-" * 40)
        for name in dataset_names:
            classification = stats_results[name]['classification']
            print(f"{name}:")
            print(f"  • Overall Type: {classification['overall_type'].replace('_', ' ').title()}")
            print(f"  • Skewness: {classification['skew_type'].replace('_', ' ').title()}")
            print(f"  • Kurtosis: {classification['kurtosis_type'].replace('_', ' ').title()}")
            print(f"  • Normal?: {'Yes' if classification['is_normal'] else 'No'}")

        # Variability ranking
        print(f"\n📊 VARIABILITY RANKING (by CV):")
        print("-" * 30)
        cv_data = [(name, stats_results[name]['variability']['cv']) for name in dataset_names]
        cv_data = [(name, cv) for name, cv in cv_data if not np.isnan(cv)]
        cv_data.sort(key=lambda x: x[1], reverse=True)

        for i, (name, cv) in enumerate(cv_data, 1):
            print(f"{i}. {name}: {cv:.1f}%")

    def _perform_statistical_tests(self, datasets_dict):
        """Perform statistical tests for comparison"""

        print(f"\n🧪 STATISTICAL TESTS")
        print("-" * 30)

        dataset_names = list(datasets_dict.keys())

        if len(dataset_names) >= 2:
            # Pairwise comparisons
            for i in range(len(dataset_names)):
                for j in range(i+1, len(dataset_names)):
                    name1, name2 = dataset_names[i], dataset_names[j]
                    data1, data2 = datasets_dict[name1], datasets_dict[name2]

                    try:
                        # Mann-Whitney U test (non-parametric)
                        mw_stat, mw_p = stats.mannwhitneyu(data1, data2, alternative='two-sided')
                        print(f"\n{name1} vs {name2}:")
                        print(f"  Mann-Whitney U test: p-value = {mw_p:.4f}")

                        # Kolmogorov-Smirnov test
                        ks_stat, ks_p = stats.ks_2samp(data1, data2)
                        print(f"  Kolmogorov-Smirnov test: p-value = {ks_p:.4f}")

                        # Effect size (Cohen's d)
                        pooled_std = np.sqrt(((len(data1)-1)*np.var(data1, ddof=1) + (len(data2)-1)*np.var(data2, ddof=1)) / (len(data1)+len(data2)-2))
                        if pooled_std > 0:
                            cohens_d = (np.mean(data1) - np.mean(data2)) / pooled_std
                            print(f"  Cohen's d (effect size): {cohens_d:.3f}")

                    except Exception as e:
                        print(f"  Error in statistical tests: {str(e)}")

    def create_summary_dashboard(self, datasets_dict, figsize=(24, 16)):
        """Create a comprehensive summary dashboard"""

        print("🎯 CREATING COMPREHENSIVE SUMMARY DASHBOARD")
        print("=" * 60)

        # Perform full comparison
        stats_results = self.compare_distributions(datasets_dict, figsize)

        return stats_results

# Initialize the comparator
comparator = DescriptiveStatisticsComparator()
print("✅ Comparative Analysis Framework initialized (FIXED VERSION)!")
```
