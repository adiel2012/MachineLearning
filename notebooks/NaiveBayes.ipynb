## Mathematical Foundation & Bayes' Theorem

### What is Naive Bayes?

Naive Bayes is a family of **probabilistic classifiers** based on **Bayes' theorem** with the **"naive" assumption** of **conditional independence** between features. Despite this strong assumption, Naive Bayes often performs surprisingly well in practice, especially for text classification and when the number of features is large.

### 1. Bayes' Theorem

**Bayes' theorem** forms the mathematical foundation of Naive Bayes classifiers:

$$P(C \mid \mathbf{X}) = \frac{P(\mathbf{X} \mid C) \cdot P(C)}{P(\mathbf{X})}$$

Where:
- $P(C \mid \mathbf{X})$ = **Posterior probability**: Probability of class $C$ given feature vector $\mathbf{X}$
- $P(\mathbf{X} \mid C)$ = **Likelihood**: Probability of observing features $\mathbf{X}$ given class $C$
- $P(C)$ = **Prior probability**: Probability of class $C$ before observing data
- $P(\mathbf{X})$ = **Evidence**: Probability of observing features $\mathbf{X}$ (normalization factor)

### 2. Classification Decision Rule

The **Maximum A Posteriori (MAP)** decision rule selects the class with highest posterior probability:

$$\hat{C} = \arg\max_C P(C \mid \mathbf{X}) = \arg\max_C \frac{P(\mathbf{X} \mid C) \cdot P(C)}{P(\mathbf{X})}$$

Since $P(\mathbf{X})$ is constant for all classes, we can ignore it:

$$\hat{C} = \arg\max_C P(\mathbf{X} \mid C) \cdot P(C)$$

### 3. The Challenge: High-Dimensional Likelihood

For a feature vector $\mathbf{X} = (X_1, X_2, \ldots, X_n)$, computing $P(\mathbf{X} \mid C)$ directly requires estimating the **joint distribution** of all features, which becomes computationally intractable and requires exponentially more data as dimensionality increases.

**Solution**: The **Naive Independence Assumption**

---

## The Naive Independence Assumption

### Mathematical Formulation

The **"naive" assumption** states that all features are **conditionally independent** given the class:

$$P(\mathbf{X} \mid C) = P(X_1, X_2, \ldots, X_n \mid C) = \prod_{i=1}^{n} P(X_i \mid C)$$

This transforms the complex joint distribution into a product of simpler **univariate distributions**.

### Classification Formula

Combining Bayes' theorem with the naive assumption:

$$\hat{C} = \arg\max_C P(C) \prod_{i=1}^{n} P(X_i \mid C)$$

**Log-probability version** (to prevent numerical underflow):

$$\hat{C} = \arg\max_C \left[ \log P(C) + \sum_{i=1}^{n} \log P(X_i \mid C) \right]$$

### Why "Naive" Works in Practice

Despite the strong independence assumption being violated in many real-world scenarios, Naive Bayes often performs well because:

1. **Classification vs. Probability Estimation**: We only need **correct ranking** of classes, not accurate probability values
2. **Bias-Variance Tradeoff**: The assumption introduces **bias** but significantly reduces **variance**
3. **High-Dimensional Advantage**: Independence assumption provides **regularization** in high-dimensional spaces
4. **Empirical Success**: Particularly effective in text classification where feature correlations often cancel out

## Types of Naive Bayes Classifiers

The choice of Naive Bayes variant depends on the **nature of features**:

### 1. **Gaussian Naive Bayes** 
- **Features**: Continuous (real-valued)
- **Distribution**: $P(X_i \mid C) \sim \mathcal{N}(\mu_{i,C}, \sigma_{i,C}^2)$
- **Applications**: Iris classification, medical diagnosis, sensor data

### 2. **Multinomial Naive Bayes**
- **Features**: Discrete counts (non-negative integers)  
- **Distribution**: $P(\mathbf{X} \mid C) \sim \text{Multinomial}(\mathbf{p}_C)$
- **Applications**: Text classification, document categorization, genomics

### 3. **Bernoulli Naive Bayes**
- **Features**: Binary (0/1, True/False)
- **Distribution**: $P(X_i \mid C) \sim \text{Bernoulli}(p_{i,C})$
- **Applications**: Spam detection, sentiment analysis, medical diagnosis

## Gaussian Naive Bayes

### Mathematical Foundation

**Gaussian Naive Bayes** assumes that continuous features follow a **normal distribution** within each class. This is the most commonly used variant for continuous numerical features.

### Probability Density Function

For each feature $X_i$ and class $C$:

$$P(X_i = x \mid C) = \frac{1}{\sqrt{2\pi\sigma_{i,C}^2}} \exp\left(-\frac{(x - \mu_{i,C})^2}{2\sigma_{i,C}^2}\right)$$

Where:
- $\mu_{i,C}$ = **mean** of feature $i$ for class $C$
- $\sigma_{i,C}^2$ = **variance** of feature $i$ for class $C$

### Parameter Estimation

**Maximum Likelihood Estimates (MLE)**:

$$\mu_{i,C} = \frac{1}{N_C} \sum_{j: y_j = C} x_{j,i}$$

$$\sigma_{i,C}^2 = \frac{1}{N_C} \sum_{j: y_j = C} (x_{j,i} - \mu_{i,C})^2$$

Where $N_C$ is the number of training samples in class $C$.

### Classification Formula

$$\hat{C} = \arg\max_C \left[ \log P(C) + \sum_{i=1}^{n} \log P(X_i \mid C) \right]$$

$$= \arg\max_C \left[ \log P(C) - \frac{1}{2} \sum_{i=1}^{n} \left[ \log(2\pi\sigma_{i,C}^2) + \frac{(x_i - \mu_{i,C})^2}{\sigma_{i,C}^2} \right] \right]$$

## Multinomial Naive Bayes

### Mathematical Foundation

**Multinomial Naive Bayes** is designed for **discrete count features**, where each feature represents the count or frequency of an event. It's most commonly used in **text classification**.

### Probability Mass Function

For a document with word counts $\mathbf{x} = (x_1, x_2, \ldots, x_n)$ where $x_i$ is the count of word $i$:

$$P(\mathbf{x} \mid C) = \frac{(\sum_i x_i)!}{\prod_i x_i!} \prod_{i=1}^{n} p_{i,C}^{x_i}$$

Where:
- $p_{i,C}$ = probability of word $i$ occurring in class $C$
- $\sum_i x_i$ = total word count in the document

### Simplified Likelihood (Proportional Form)

Since the multinomial coefficient doesn't depend on the class, we can use:

$$P(\mathbf{x} \mid C) \propto \prod_{i=1}^{n} p_{i,C}^{x_i}$$

### Parameter Estimation with Laplace Smoothing

To handle **zero probabilities** (words not seen in training), we use **Laplace smoothing**:

$$p_{i,C} = \frac{N_{i,C} + \alpha}{\sum_j (N_{j,C} + \alpha)} = \frac{N_{i,C} + \alpha}{N_C + \alpha|V|}$$

Where:
- $N_{i,C}$ = count of word $i$ in documents of class $C$
- $N_C$ = total word count in class $C$
- $|V|$ = vocabulary size
- $\alpha$ = smoothing parameter (typically $\alpha = 1$ for Laplace smoothing)

### Classification Formula

$$\hat{C} = \arg\max_C \left[ \log P(C) + \sum_{i=1}^{n} x_i \log p_{i,C} \right]$$

## Bernoulli Naive Bayes

### Mathematical Foundation

**Bernoulli Naive Bayes** is designed for **binary features**, where each feature indicates the presence (1) or absence (0) of a characteristic. Unlike Multinomial NB, it explicitly models non-occurring features.

### Probability Mass Function

For binary feature vector $\mathbf{x} = (x_1, x_2, \ldots, x_n)$ where $x_i \in \{0, 1\}$:

$$P(\mathbf{x} \mid C) = \prod_{i=1}^{n} p_{i,C}^{x_i} (1 - p_{i,C})^{(1-x_i)}$$

Where $p_{i,C}$ is the probability that feature $i$ is present (equals 1) in class $C$.

### Parameter Estimation with Smoothing

$$p_{i,C} = \frac{N_{i,C} + \alpha}{N_C + 2\alpha}$$

Where:
- $N_{i,C}$ = number of documents in class $C$ where feature $i$ is present
- $N_C$ = total number of documents in class $C$  
- $\alpha$ = smoothing parameter

### Classification Formula

$$\hat{C} = \arg\max_C \left[ \log P(C) + \sum_{i=1}^{n} \left[ x_i \log p_{i,C} + (1-x_i) \log(1-p_{i,C}) \right] \right]$$

### Key Difference from Multinomial

- **Bernoulli NB**: Explicitly models both presence AND absence of features
- **Multinomial NB**: Only considers the frequency of present features
- **When to use**: Bernoulli for small documents with sparse features; Multinomial for longer documents with word counts

---

## Maximum Likelihood Estimation & Smoothing

### The Zero Probability Problem

**Problem**: If a feature value never appears with a class in training data, $P(X_i \mid C) = 0$, making the entire posterior probability zero.

**Example**: Word "fantastic" never appears in spam emails during training → $P(\text{fantastic} \mid \text{spam}) = 0$ → Any email containing "fantastic" gets $P(\text{spam} \mid \text{email}) = 0$

### Laplace Smoothing (Add-One Smoothing)

**Solution**: Add a small count $\alpha$ to all feature counts:

$$P(X_i = v \mid C) = \frac{\text{count}(X_i = v, C) + \alpha}{\text{count}(C) + \alpha \cdot |\text{values}(X_i)|}$$

**Benefits**:
- Ensures no zero probabilities
- Minimal impact on well-estimated probabilities
- Provides Bayesian interpretation (uniform prior)

### Alternative Smoothing Methods

1. **Lidstone Smoothing**: $\alpha \in (0, 1)$ instead of $\alpha = 1$
2. **Good-Turing Smoothing**: Uses frequency of frequencies
3. **Kneser-Ney Smoothing**: Advanced method for language modeling

````python
# Complete Implementation of Naive Bayes Classifiers from Scratch

class GaussianNaiveBayes:
    """
    Gaussian Naive Bayes for continuous features
    Assumes features follow normal distribution within each class
    """
    
    def __init__(self):
        self.classes_ = None
        self.class_priors_ = {}
        self.feature_params_ = {}  # {class: {feature: {'mean': ..., 'var': ...}}}
        
    def fit(self, X, y):
        """
        Train the Gaussian Naive Bayes classifier
        
        Parameters:
        X : array-like, shape (n_samples, n_features)
        y : array-like, shape (n_samples,)
        """
        X = np.array(X)
        y = np.array(y)
        
        self.classes_ = np.unique(y)
        n_samples, n_features = X.shape
        
        # Calculate class priors P(C)
        for class_val in self.classes_:
            class_count = np.sum(y == class_val)
            self.class_priors_[class_val] = class_count / n_samples
            
        # Calculate feature parameters (mean and variance) for each class
        for class_val in self.classes_:
            class_mask = (y == class_val)
            X_class = X[class_mask]
            
            self.feature_params_[class_val] = {}
            for feature_idx in range(n_features):
                feature_values = X_class[:, feature_idx]
                mean = np.mean(feature_values)
                # Add small epsilon to prevent zero variance
                var = np.var(feature_values) + 1e-9
                
                self.feature_params_[class_val][feature_idx] = {
                    'mean': mean,
                    'var': var
                }
                
        return self
    
    def _gaussian_pdf(self, x, mean, var):
        """Calculate Gaussian probability density function"""
        coefficient = 1.0 / np.sqrt(2.0 * np.pi * var)
        exponent = np.exp(-0.5 * ((x - mean) ** 2) / var)
        return coefficient * exponent
    
    def _predict_sample(self, x):
        """Predict single sample using MAP estimation"""
        posteriors = {}
        
        for class_val in self.classes_:
            # Log probability to prevent underflow
            log_posterior = np.log(self.class_priors_[class_val])
            
            # Add log likelihood for each feature
            for feature_idx in range(len(x)):
                mean = self.feature_params_[class_val][feature_idx]['mean']
                var = self.feature_params_[class_val][feature_idx]['var']
                
                # Log of Gaussian PDF
                log_likelihood = np.log(self._gaussian_pdf(x[feature_idx], mean, var))
                log_posterior += log_likelihood
                
            posteriors[class_val] = log_posterior
            
        # Return class with maximum posterior probability
        return max(posteriors, key=posteriors.get)
    
    def predict(self, X):
        """Predict class labels for samples in X"""
        X = np.array(X)
        predictions = []
        
        for sample in X:
            predictions.append(self._predict_sample(sample))
            
        return np.array(predictions)
    
    def predict_proba(self, X):
        """Predict class probabilities for samples in X"""
        X = np.array(X)
        probabilities = []
        
        for sample in X:
            posteriors = {}
            
            for class_val in self.classes_:
                log_posterior = np.log(self.class_priors_[class_val])
                
                for feature_idx in range(len(sample)):
                    mean = self.feature_params_[class_val][feature_idx]['mean']
                    var = self.feature_params_[class_val][feature_idx]['var']
                    log_likelihood = np.log(self._gaussian_pdf(sample[feature_idx], mean, var))
                    log_posterior += log_likelihood
                    
                posteriors[class_val] = log_posterior
            
            # Convert log probabilities to probabilities and normalize
            max_log_prob = max(posteriors.values())
            exp_probs = {c: np.exp(prob - max_log_prob) for c, prob in posteriors.items()}
            total_prob = sum(exp_probs.values())
            normalized_probs = {c: prob / total_prob for c, prob in exp_probs.items()}
            
            # Convert to array in class order
            prob_array = [normalized_probs[class_val] for class_val in self.classes_]
            probabilities.append(prob_array)
            
        return np.array(probabilities)


class MultinomialNaiveBayes:
    """
    Multinomial Naive Bayes for discrete count features
    Commonly used for text classification with word counts
    """
    
    def __init__(self, alpha=1.0):
        self.alpha = alpha  # Smoothing parameter
        self.classes_ = None
        self.class_priors_ = {}
        self.feature_probs_ = {}  # {class: {feature: probability}}
        self.vocabulary_size_ = 0
        
    def fit(self, X, y):
        """
        Train the Multinomial Naive Bayes classifier
        
        Parameters:
        X : array-like, shape (n_samples, n_features) - count features
        y : array-like, shape (n_samples,)
        """
        X = np.array(X)
        y = np.array(y)
        
        self.classes_ = np.unique(y)
        n_samples, n_features = X.shape
        self.vocabulary_size_ = n_features
        
        # Calculate class priors P(C)
        for class_val in self.classes_:
            class_count = np.sum(y == class_val)
            self.class_priors_[class_val] = class_count / n_samples
        
        # Calculate feature probabilities with Laplace smoothing
        for class_val in self.classes_:
            class_mask = (y == class_val)
            X_class = X[class_mask]
            
            # Sum of all feature counts for this class
            total_count = np.sum(X_class) + self.alpha * self.vocabulary_size_
            
            # Feature probabilities with smoothing
            feature_counts = np.sum(X_class, axis=0) + self.alpha
            self.feature_probs_[class_val] = feature_counts / total_count
            
        return self
    
    def predict(self, X):
        """Predict class labels for samples in X"""
        X = np.array(X)
        predictions = []
        
        for sample in X:
            posteriors = {}
            
            for class_val in self.classes_:
                # Log probability to prevent underflow
                log_posterior = np.log(self.class_priors_[class_val])
                
                # Add log likelihood for each feature
                for feature_idx in range(len(sample)):
                    count = sample[feature_idx]
                    if count > 0:  # Only consider non-zero counts
                        feature_prob = self.feature_probs_[class_val][feature_idx]
                        log_posterior += count * np.log(feature_prob)
                        
                posteriors[class_val] = log_posterior
            
            # Return class with maximum posterior probability
            predictions.append(max(posteriors, key=posteriors.get))
            
        return np.array(predictions)


class BernoulliNaiveBayes:
    """
    Bernoulli Naive Bayes for binary features
    Models both presence and absence of features explicitly
    """
    
    def __init__(self, alpha=1.0):
        self.alpha = alpha  # Smoothing parameter
        self.classes_ = None
        self.class_priors_ = {}
        self.feature_probs_ = {}  # {class: {feature: probability}}
        
    def fit(self, X, y):
        """
        Train the Bernoulli Naive Bayes classifier
        
        Parameters:
        X : array-like, shape (n_samples, n_features) - binary features
        y : array-like, shape (n_samples,)
        """
        X = np.array(X)
        y = np.array(y)
        
        self.classes_ = np.unique(y)
        n_samples, n_features = X.shape
        
        # Calculate class priors P(C)
        for class_val in self.classes_:
            class_count = np.sum(y == class_val)
            self.class_priors_[class_val] = class_count / n_samples
        
        # Calculate feature probabilities with smoothing
        for class_val in self.classes_:
            class_mask = (y == class_val)
            X_class = X[class_mask]
            n_class_samples = len(X_class)
            
            self.feature_probs_[class_val] = {}
            
            for feature_idx in range(n_features):
                # Count of feature=1 in this class + smoothing
                feature_count = np.sum(X_class[:, feature_idx]) + self.alpha
                # Total class count + smoothing for both 0 and 1
                total_count = n_class_samples + 2 * self.alpha
                
                self.feature_probs_[class_val][feature_idx] = feature_count / total_count
                
        return self
    
    def predict(self, X):
        """Predict class labels for samples in X"""
        X = np.array(X)
        predictions = []
        
        for sample in X:
            posteriors = {}
            
            for class_val in self.classes_:
                # Log probability to prevent underflow
                log_posterior = np.log(self.class_priors_[class_val])
                
                # Add log likelihood for each feature (both presence and absence)
                for feature_idx in range(len(sample)):
                    feature_val = sample[feature_idx]
                    p_feature_given_class = self.feature_probs_[class_val][feature_idx]
                    
                    if feature_val == 1:
                        log_posterior += np.log(p_feature_given_class)
                    else:  # feature_val == 0
                        log_posterior += np.log(1 - p_feature_given_class)
                        
                posteriors[class_val] = log_posterior
            
            # Return class with maximum posterior probability
            predictions.append(max(posteriors, key=posteriors.get))
            
        return np.array(predictions)


# Utility function to evaluate model performance
def evaluate_model(y_true, y_pred, model_name):
    """Comprehensive model evaluation"""
    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, average='weighted', zero_division=0)
    recall = recall_score(y_true, y_pred, average='weighted', zero_division=0)
    f1 = f1_score(y_true, y_pred, average='weighted', zero_division=0)
    
    print(f"\n📊 {model_name} Performance:")
    print(f"   • Accuracy:  {accuracy:.4f}")
    print(f"   • Precision: {precision:.4f}")
    print(f"   • Recall:    {recall:.4f}")
    print(f"   • F1-Score:  {f1:.4f}")
    
    return {
        'accuracy': accuracy,
        'precision': precision, 
        'recall': recall,
        'f1_score': f1
    }

print("\n✅ All Naive Bayes implementations ready!")
print("   • GaussianNaiveBayes: For continuous features")
print("   • MultinomialNaiveBayes: For count features")  
print("   • BernoulliNaiveBayes: For binary features")
````

## 🎯 Model Interpretation & Practical Insights

### When to Use Each Naive Bayes Variant

#### **Gaussian Naive Bayes** 🌸
**Best for**: Continuous numerical features that roughly follow normal distributions
- **Applications**: Medical diagnosis, sensor data, financial predictions, iris classification
- **Advantages**: 
  - Works well with continuous features
  - Fast training and prediction
  - Handles multi-class naturally
  - Good baseline for continuous data
- **Limitations**: 
  - Assumes Gaussian distribution (may not hold in practice)
  - Sensitive to outliers
  - Strong independence assumption
- **When to use**: When features are continuous and roughly normally distributed

#### **Multinomial Naive Bayes** 📰
**Best for**: Discrete count features (text analysis, genomics)
- **Applications**: Document classification, spam detection, sentiment analysis, topic modeling
- **Advantages**: 
  - Excellent for text classification
  - Handles high-dimensional sparse data efficiently
  - Built-in smoothing prevents zero probabilities
  - Scales well with vocabulary size
- **Limitations**: 
  - Only works with non-negative features
  - Ignores word absence (unlike Bernoulli)
  - Assumes word independence in text
- **When to use**: Text classification, document categorization, any count-based features

#### **Bernoulli Naive Bayes** 🔍
**Best for**: Binary features (presence/absence)
- **Applications**: Spam detection, sentiment analysis with binary features, medical diagnosis
- **Advantages**: 
  - Explicitly models feature absence
  - Good for small documents
  - Works with binary features naturally
  - Less sensitive to document length
- **Limitations**: 
  - Only binary features
  - May lose information by ignoring counts
  - Still assumes feature independence
- **When to use**: Binary classification with sparse binary features

  - Scale features if they have very different ranges
  - Remove outliers that might skew mean/variance estimates

- **Multinomial NB**: 
  - Use TF-IDF weighting for better performance
  - Remove very rare and very common words
  - Consider n-gram features for context

- **Bernoulli NB**: 
  - Carefully choose vocabulary size
  - Consider feature selection to reduce noise
  - Good for small documents with distinct vocabularies

#### **Hyperparameter Tuning**
- **Smoothing parameter (α)**: Start with α=1.0 (Laplace smoothing)
  - Increase α if overfitting (too many zero probabilities)
  - Decrease α if you have sufficient data and want less regularization
  - Use cross-validation to optimize

#### **When NOT to Use Naive Bayes**
- **Strong feature dependencies**: When features are highly correlated (use logistic regression instead)
- **Complex non-linear relationships**: Consider SVM, Random Forest, or Neural Networks
- **Very small datasets**: May not have enough data to estimate probabilities reliably
- **When prediction probability calibration is critical**: Naive Bayes probabilities can be poorly calibrated
