# Deep Mathematical Explanation of Naive Bayes Algorithms

## 1. Bayes' Theorem

Naive Bayes is fundamentally based on Bayes' Theorem, which is expressed as:

$P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)}$

Where:
- $P(A|B)$ is the posterior probability
- $P(B|A)$ is the likelihood
- $P(A)$ is the prior probability
- $P(B)$ is the evidence

## 2. Naive Assumption

The "naive" part of Naive Bayes comes from the assumption that features are conditionally independent given the class. Mathematically, for features $X_1, X_2, ..., X_n$ and class $C$:

$P(X_1, X_2, ..., X_n | C) = P(X_1|C) \cdot P(X_2|C) \cdot ... \cdot P(X_n|C)$

This assumption simplifies the computation significantly but may not always hold in real-world scenarios.

## 3. Maximum A Posteriori (MAP) estimation

Naive Bayes classifier aims to find the most probable class given the feature vector. This is done using MAP estimation:

$C_{MAP} = \arg\max_c P(C=c | X_1, X_2, ..., X_n)$

Using Bayes' theorem and the naive assumption, this can be rewritten as:

$C_{MAP} = \arg\max_c P(C=c) \cdot \prod_i P(X_i | C=c)$

## 4. Types of Naive Bayes classifiers

There are several types of Naive Bayes classifiers, depending on the assumed distribution of features:

a) Gaussian Naive Bayes: For continuous features, assuming a Gaussian distribution.
b) Multinomial Naive Bayes: For discrete features, often used in text classification.
c) Bernoulli Naive Bayes: For binary features.

## 5. Mathematical formulation

Let's focus on the Multinomial Naive Bayes, commonly used in text classification:

For a document $d$ and a class $c$, the classifier calculates:

$P(c|d) \propto P(c) \cdot \prod_i P(t_i|c)^{n_i(d)}$

Where:
- $P(c)$ is the prior probability of class $c$
- $P(t_i|c)$ is the conditional probability of term $t_i$ given class $c$
- $n_i(d)$ is the count of term $t_i$ in document $d$

The probabilities are typically estimated using maximum likelihood:

$P(c) = \frac{N_c}{N}$
$P(t_i|c) = \frac{count(t_i, c) + \alpha}{\sum_t count(t, c) + \alpha|V|}$

Where:
- $N_c$ is the number of documents in class $c$
- $N$ is the total number of documents
- $count(t_i, c)$ is the count of term $t_i$ in documents of class $c$
- $|V|$ is the vocabulary size
- $\alpha$ is the smoothing parameter (typically 1 for Laplace smoothing)

The logarithm is often used to prevent underflow and for computational efficiency:

$\log P(c|d) \propto \log P(c) + \sum_i n_i(d) \cdot \log P(t_i|c)$

This mathematical formulation allows for efficient classification, especially in high-dimensional spaces like text classification.

# Naive Bayes Algorithm Explanation with Sample Dataset

## Sample Dataset

Let's consider a fruit classification problem. We'll use three features: Color (Red/Yellow/Green), Shape (Round/Oval), and Size (Small/Large) to classify fruits as Apple or Banana.

| Color  | Shape  | Size  | Fruit  |
|--------|--------|-------|--------|
| Red    | Round  | Small | Apple  |
| Red    | Round  | Large | Apple  |
| Yellow | Oval   | Large | Banana |
| Yellow | Oval   | Small | Banana |
| Green  | Round  | Small | Apple  |
| Yellow | Oval   | Large | Banana |
| Green  | Round  | Large | Apple  |
| Yellow | Oval   | Small | Banana |
| Red    | Round  | Small | Apple  |
| Yellow | Oval   | Large | Banana |

## Step-by-step Naive Bayes Algorithm Explanation

Let's predict the fruit for a new pattern: (Color = Yellow, Shape = Round, Size = Small)

### Step 1: Calculate Prior Probabilities

$P(Fruit = Apple) = 5/10 = 0.5$
$P(Fruit = Banana) = 5/10 = 0.5$

### Step 2: Calculate Likelihood Probabilities

For Apple:
- $P(Color = Yellow | Apple) = 0/5 = 0$ (We'll use 0.1 to avoid zero probability)
- $P(Shape = Round | Apple) = 5/5 = 1$
- $P(Size = Small | Apple) = 3/5 = 0.6$

For Banana:
- $P(Color = Yellow | Banana) = 5/5 = 1$
- $P(Shape = Round | Banana) = 0/5 = 0$ (We'll use 0.1 to avoid zero probability)
- $P(Size = Small | Banana) = 2/5 = 0.4$

### Step 3: Apply Naive Bayes Formula

For Apple:
$P(Apple | Yellow, Round, Small) \propto P(Apple) \cdot P(Yellow | Apple) \cdot P(Round | Apple) \cdot P(Small | Apple)$
$= 0.5 \cdot 0.1 \cdot 1 \cdot 0.6 = 0.03$

For Banana:
$P(Banana | Yellow, Round, Small) \propto P(Banana) \cdot P(Yellow | Banana) \cdot P(Round | Banana) \cdot P(Small | Banana)$
$= 0.5 \cdot 1 \cdot 0.1 \cdot 0.4 = 0.02$

### Step 4: Normalize Probabilities

Total probability: $0.03 + 0.02 = 0.05$

$P(Apple | Yellow, Round, Small) = 0.03 / 0.05 = 0.6$ or $60\%$
$P(Banana | Yellow, Round, Small) = 0.02 / 0.05 = 0.4$ or $40\%$

### Step 5: Make Prediction

Since $P(Apple | Yellow, Round, Small) > P(Banana | Yellow, Round, Small)$, we predict that the fruit is an Apple.

## Conclusion

In this example, the Naive Bayes algorithm predicts that a yellow, round, small fruit is more likely to be an Apple (60%) than a Banana (40%). This prediction might seem counterintuitive given our real-world knowledge, but it's based solely on the patterns in our sample dataset.

Key points:
1. We used a small non-zero probability (0.1) to handle cases where the likelihood would otherwise be zero. This is a simple form of smoothing.
2. The algorithm considers all features independently, which is the "naive" assumption.
3. Even though yellow color strongly suggests a banana, the round shape (which is more common for apples in our dataset) influenced the prediction towards apple.
4. This example demonstrates how Naive Bayes can make predictions even with seemingly contradictory evidence, based on the patterns in the training data.

In practice, this result suggests that:
1. We might want to collect more data to improve our model's confidence.
2. We could consider adding more relevant features to better distinguish between apples and bananas.
3. We should be cautious about this prediction given its relatively close probability, and perhaps present it as an uncertain classification.

```python
import numpy as np

# Define the dataset
data = np.array([
    ['Red', 'Round', 'Small', 'Apple'],
    ['Red', 'Round', 'Large', 'Apple'],
    ['Yellow', 'Oval', 'Large', 'Banana'],
    ['Yellow', 'Oval', 'Small', 'Banana'],
    ['Green', 'Round', 'Small', 'Apple'],
    ['Yellow', 'Oval', 'Large', 'Banana'],
    ['Green', 'Round', 'Large', 'Apple'],
    ['Yellow', 'Oval', 'Small', 'Banana'],
    ['Red', 'Round', 'Small', 'Apple'],
    ['Yellow', 'Oval', 'Large', 'Banana']
])

# Define feature and class labels
features = ['Color', 'Shape', 'Size']
classes = ['Apple', 'Banana']

# Function to calculate prior probabilities
def calculate_priors(data):
    total = len(data)
    return {c: np.sum(data[:, -1] == c) / total for c in classes}

# Function to calculate likelihood probabilities
def calculate_likelihoods(data):
    likelihoods = {
        'Apple': {'Color': {'Yellow': 0.1}, 'Shape': {'Round': 1.0}, 'Size': {'Small': 0.6}},
        'Banana': {'Color': {'Yellow': 1.0}, 'Shape': {'Round': 0.1}, 'Size': {'Small': 0.4}}
    }
    return likelihoods

# Function to predict using Naive Bayes
def naive_bayes_predict(sample, priors, likelihoods):
    predictions = {}
    for c in classes:
        prior = priors[c]
        likelihood = 1
        for feature, value in zip(features, sample):
            likelihood *= likelihoods[c][feature][value]
        predictions[c] = prior * likelihood
    return predictions

# Calculate priors and likelihoods
priors = calculate_priors(data)
likelihoods = calculate_likelihoods(data)

# New sample to classify
new_sample = ['Yellow', 'Round', 'Small']

# Make prediction
prediction = naive_bayes_predict(new_sample, priors, likelihoods)

# Print results
print("Prior Probabilities:")
print(priors)
print("\nLikelihood Probabilities:")
print(likelihoods)
print("\nNew Sample:", new_sample)

# Print step-by-step calculation
print("\nStep-by-step calculation:")
for c in classes:
    prob = priors[c]
    print(f"\n{c}:")
    print(f"Prior: {priors[c]:.4f}")
    for feature, value in zip(features, new_sample):
        print(f"P({value}|{c}) = {likelihoods[c][feature][value]:.4f}")
        prob *= likelihoods[c][feature][value]
    print(f"Unnormalized probability: {prob:.6f}")
    prediction[c] = prob

# Normalize
total = sum(prediction.values())
print(f"\nTotal probability: {total:.6f}")
for c in classes:
    normalized_prob = prediction[c] / total
    print(f"Normalized P({c}|Yellow,Round,Small) = {normalized_prob:.4f}")

print("\nPredicted Class:", max(prediction, key=prediction.get))
```
